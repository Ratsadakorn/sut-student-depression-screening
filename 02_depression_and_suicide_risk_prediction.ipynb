```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import rankdata


def chatterjee_xi_tie_average(x, y):
    """
    Calculate Chatterjee's xi(X, Y), averaging exactly over
    all possible orderings within tied X groups.

    X is the explanatory variable.
    Y is the response variable.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    if x.ndim != 1 or y.ndim != 1 or len(x) != len(y):
        raise ValueError(
            "X and Y must be equally sized one-dimensional arrays."
        )

    if len(x) < 3:
        raise ValueError("At least three observations are required.")

    if not np.isfinite(x).all() or not np.isfinite(y).all():
        raise ValueError("Missing or non-finite values found in X or Y.")

    n = len(y)

    # Original ranks start at 1.
    # r_i = number of observations with Y <= Y_i
    # l_i = number of observations with Y >= Y_i
    ranks = rankdata(y, method="max")
    upper_counts = n + 1 - rankdata(y, method="min")

    denominator = 2 * np.sum(
        upper_counts * (n - upper_counts)
    )

    # Xi is undefined when Y is constant.
    if denominator == 0:
        return np.nan

    # Sort by X and identify groups of identical X values.
    order = np.argsort(x, kind="stable")
    sorted_x = x[order]
    sorted_ranks = ranks[order]

    boundaries = np.flatnonzero(
        sorted_x[1:] != sorted_x[:-1]
    ) + 1

    groups = np.split(sorted_ranks, boundaries)

    expected_rank_difference = 0.0

    # Expected sum of adjacent rank differences within each X group.
    for group in groups:
        size = len(group)
        ordered_group = np.sort(group)

        coefficients = (
            2 * np.arange(1, size + 1) - size - 1
        )

        expected_rank_difference += (
            2 * np.dot(coefficients, ordered_group) / size
        )

    # Expected rank difference between consecutive X groups.
    for left_group, right_group in zip(groups[:-1], groups[1:]):
        ordered_right = np.sort(right_group)

        prefix_sum = np.concatenate([
            [0.0],
            np.cumsum(ordered_right),
        ])

        positions = np.searchsorted(
            ordered_right,
            left_group,
            side="left",
        )

        differences_to_lower = (
            left_group * positions
            - prefix_sum[positions]
        )

        differences_to_upper = (
            prefix_sum[-1]
            - prefix_sum[positions]
            - left_group * (len(right_group) - positions)
        )

        expected_rank_difference += (
            np.sum(differences_to_lower + differences_to_upper)
            / (len(left_group) * len(right_group))
        )

    return float(
        1 - n * expected_rank_difference / denominator
    )


# Validate the analysis dataframe.
columns = df1.columns

if not columns.is_unique:
    raise ValueError("Duplicate column names found in df1.")

if len(df1) < 3 or len(columns) < 2:
    raise ValueError(
        "At least three observations and two columns are required."
    )

if df1.isna().any().any():
    raise ValueError(
        "Missing values found in df1. Check the analysis data first."
    )

non_numeric = df1.select_dtypes(exclude="number").columns.tolist()

if non_numeric:
    raise ValueError(
        "Explicit numeric coding is required for: "
        + ", ".join(map(str, non_numeric))
    )

values = df1.to_numpy(dtype=float)

if not np.isfinite(values).all():
    raise ValueError("Non-finite values found in df1.")

# Initialize the dataframe for storing Xi values.
xi_corr_matrix = pd.DataFrame(
    np.nan,
    columns=columns,
    index=columns,
)

# Rows represent X; columns represent Y.
for i in range(len(columns)):
    for j in range(len(columns)):
        xi_corr_matrix.iloc[i, j] = chatterjee_xi_tie_average(
            values[:, i],
            values[:, j],
        )

# Display the Xi correlation matrix.
print("New Correlation")
print(xi_corr_matrix)

# Create a heatmap.
plt.figure(figsize=(10, 8))
sns.heatmap(
    xi_corr_matrix,
    annot=False,
    cmap=plt.cm.Greens,
)

plt.title(
    "New Correlation Heatmap between Student Characteristics and Outcomes",
    fontsize=16,
)

plt.xticks(fontsize=13, rotation=45, ha="right")
plt.yticks(fontsize=13)

plt.tight_layout()

# ==== EXPORT HIGH QUALITY ====
plt.savefig(
    "xi_correlation_heatmap.png",
    dpi=600,
    bbox_inches="tight",
)
# plt.savefig("xi_correlation_heatmap.pdf", bbox_inches="tight")
# plt.savefig("xi_correlation_heatmap.svg", bbox_inches="tight")

plt.show()
```

```python
# ============================================================
# PART S0 VERSION 2.0
# SECONDARY OUTCOME DEFINITIONS, DATA INTEGRITY,
# FEATURE-TYPE LOCK, AND CIRCULARITY AUDIT
# ============================================================
#
# Secondary outcomes:
# 1. Depression risk derived from the 9Q-based depression level.
# 2. Suicide risk derived from the 8Q-based suicide-risk level.
#
# Prespecified predictor strategy:
# - M2_Admin_2Q
# - 8 administrative predictors
# - 2 items from the Thai 2Q depression screener
#
# Important interpretation:
# - The depression-risk model has no direct 9Q item predictors, but
#   the Thai 2Q predictors and the 9Q-derived outcome assess closely
#   related depressive constructs. Substantial construct overlap
#   therefore remains.
# - The suicide-risk model contains no 8Q-derived predictors, but its
#   outcome remains questionnaire-defined rather than independently
#   adjudicated.
#
# This PART performs no outcome dichotomization, data split, model
# fitting, resampling, prediction, threshold selection, calibration,
# bootstrap analysis, or validation.
#
# Privacy safeguard:
# - Participant-level rows, identifiers, and row indices are not
#   exported.
# - Only aggregate audit metadata are written to CSV.
# ============================================================

import numpy as np
import pandas as pd


# ============================================================
# 1. SETTINGS
# ============================================================

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

DEVELOPMENT_PATH = "2567-2Q9Q8Q.csv"
TEMPORAL_PATH = "2568-2Q9Q8Q.csv"

EXPECTED_2024_N = 4090
EXPECTED_2025_N = 815

PRESPECIFIED_STRATEGY = "M2_Admin_2Q"
PRESPECIFIED_ALGORITHM = "AdaBoostClassifier"

EXPECTED_M2_FEATURES = [
    "institute",
    "GPAX",
    "dadstatus",
    "momstatus",
    "dadeducation",
    "momeducation",
    "parent",
    "gender",
    "2Q1",
    "2Q2",
]

ADMIN_FEATURES = EXPECTED_M2_FEATURES[:8]
THAI_2Q_FEATURES = ["2Q1", "2Q2"]
M2_FEATURES = EXPECTED_M2_FEATURES.copy()

CONTINUOUS_FEATURES = ["GPAX"]
CATEGORICAL_OR_ORDINAL_FEATURES = [
    feature
    for feature in M2_FEATURES
    if feature not in CONTINUOUS_FEATURES
]

CATEGORICAL_INDICES = [
    index
    for index, feature in enumerate(M2_FEATURES)
    if feature in CATEGORICAL_OR_ORDINAL_FEATURES
]

SECONDARY_OUTCOMES = {
    "Depression_Risk": {
        "source_column": "depressionlevel",
        "derived_from": "9Q-based depression level",
        "binary_rule_for_part_s1": "depressionlevel >= 1",
        "prohibited_predictor_prefixes": ["9Q"],
        "direct_item_overlap_expected": False,
        "construct_overlap": "Substantial",
        "interpretation": (
            "The Thai 2Q predictors and the 9Q-derived outcome assess "
            "closely related depressive constructs."
        ),
    },
    "Suicide_Risk": {
        "source_column": "suicidelevel",
        "derived_from": "8Q-based suicide-risk level",
        "binary_rule_for_part_s1": "suicidelevel >= 1",
        "prohibited_predictor_prefixes": ["8Q"],
        "direct_item_overlap_expected": False,
        "construct_overlap": "No direct item overlap",
        "interpretation": (
            "No 8Q-derived predictor is included; however, the outcome "
            "remains questionnaire-defined rather than independently "
            "adjudicated."
        ),
    },
}


# ============================================================
# 2. LOAD SOURCE DATA
# ============================================================

df_2024 = pd.read_csv(DEVELOPMENT_PATH)
df_2025 = pd.read_csv(TEMPORAL_PATH)

if not isinstance(df_2024, pd.DataFrame):
    raise TypeError("The 2024 source is not a DataFrame.")

if not isinstance(df_2025, pd.DataFrame):
    raise TypeError("The 2025 source is not a DataFrame.")

print("=" * 105)
print("PART S0 VERSION 2.0")
print("SECONDARY OUTCOME DEFINITIONS, DATA INTEGRITY,")
print("FEATURE-TYPE LOCK, AND CIRCULARITY AUDIT")
print("=" * 105)
print(f"2024 source: {DEVELOPMENT_PATH}; shape={df_2024.shape}")
print(f"2025 source: {TEMPORAL_PATH}; shape={df_2025.shape}")


# ============================================================
# 3. VERIFY COHORT SIZE AND REQUIRED COLUMNS
# ============================================================

outcome_source_columns = [
    specification["source_column"]
    for specification in SECONDARY_OUTCOMES.values()
]

required_columns = list(dict.fromkeys(
    M2_FEATURES + outcome_source_columns + ["mdddx"]
))

dataset_audit_rows = []

for dataset_name, data, expected_n in [
    ("2024 analytic source", df_2024, EXPECTED_2024_N),
    ("2025 temporal source", df_2025, EXPECTED_2025_N),
]:
    missing_columns = [
        column for column in required_columns
        if column not in data.columns
    ]
    duplicate_columns = data.columns[
        data.columns.duplicated()
    ].tolist()

    dataset_audit_rows.append({
        "Dataset": dataset_name,
        "Observed_N": len(data),
        "Expected_N": expected_n,
        "N_Correct": len(data) == expected_n,
        "Observed_Column_Count": data.shape[1],
        "Required_Columns_Present": len(missing_columns) == 0,
        "Missing_Required_Columns": " | ".join(missing_columns),
        "Duplicate_Column_Count": len(duplicate_columns),
        "No_Duplicate_Columns": len(duplicate_columns) == 0,
    })

partS0_dataset_audit = pd.DataFrame(dataset_audit_rows)

if not partS0_dataset_audit[
    ["N_Correct", "Required_Columns_Present", "No_Duplicate_Columns"]
].all(axis=None):
    display(partS0_dataset_audit)
    raise RuntimeError("PART S0 dataset-structure audit failed.")


# ============================================================
# 4. LOCK M2 FEATURE DEFINITION AND RESAMPLING TYPES
# ============================================================

feature_definition_rows = []

for position, feature in enumerate(M2_FEATURES):
    if feature == "GPAX":
        feature_group = "Administrative"
        resampling_type = "Continuous"
    elif feature in THAI_2Q_FEATURES:
        feature_group = "Thai 2Q"
        resampling_type = "Categorical_or_Ordinal"
    else:
        feature_group = "Administrative"
        resampling_type = "Categorical_or_Ordinal"

    feature_definition_rows.append({
        "Feature_Order": position + 1,
        "Feature": feature,
        "Feature_Group": feature_group,
        "Resampling_Type": resampling_type,
        "SMOTENC_Feature_Index": position,
        "Categorical_For_SMOTENC": position in CATEGORICAL_INDICES,
    })

partS0_feature_definition = pd.DataFrame(feature_definition_rows)

feature_definition_checks = {
    "Administrative_Feature_Count_Equals_8": len(ADMIN_FEATURES) == 8,
    "Thai_2Q_Item_Count_Equals_2": len(THAI_2Q_FEATURES) == 2,
    "M2_Feature_Count_Equals_10": len(M2_FEATURES) == 10,
    "M2_Feature_Order_Correct": M2_FEATURES == EXPECTED_M2_FEATURES,
    "No_Duplicate_M2_Features": len(set(M2_FEATURES)) == 10,
    "GPAX_Is_Only_Continuous_Feature": CONTINUOUS_FEATURES == ["GPAX"],
    "Nine_Categorical_Or_Ordinal_Features": (
        len(CATEGORICAL_OR_ORDINAL_FEATURES) == 9
    ),
    "Categorical_Indices_Correct": (
        CATEGORICAL_INDICES == [0, 2, 3, 4, 5, 6, 7, 8, 9]
    ),
}

if not all(feature_definition_checks.values()):
    failed = [
        name for name, passed in feature_definition_checks.items()
        if not passed
    ]
    raise RuntimeError(
        "M2 feature-definition audit failed: " + ", ".join(failed)
    )


# ============================================================
# 5. FEATURE MISSINGNESS AND TYPE AUDIT
# ============================================================

feature_audit_rows = []

for dataset_name, data in [
    ("2024 analytic source", df_2024),
    ("2025 temporal source", df_2025),
]:
    for feature in M2_FEATURES:
        numeric_conversion = pd.to_numeric(data[feature], errors="coerce")
        nonnumeric_nonmissing = int(
            (data[feature].notna() & numeric_conversion.isna()).sum()
        )
        finite_numeric = bool(
            numeric_conversion.dropna().empty
            or np.isfinite(
                numeric_conversion.dropna().to_numpy(dtype=float)
            ).all()
        )

        feature_audit_rows.append({
            "Dataset": dataset_name,
            "Feature": feature,
            "Resampling_Type": (
                "Continuous"
                if feature == "GPAX"
                else "Categorical_or_Ordinal"
            ),
            "Missing_Cells": int(data[feature].isna().sum()),
            "Observed_Unique_Values": int(data[feature].nunique(dropna=True)),
            "Nonnumeric_Nonmissing_Cells": nonnumeric_nonmissing,
            "Numeric_Values_Finite": finite_numeric,
        })

partS0_feature_audit = pd.DataFrame(feature_audit_rows)

all_m2_complete = bool(
    (partS0_feature_audit["Missing_Cells"] == 0).all()
)
all_m2_numeric_encoded = bool(
    (partS0_feature_audit["Nonnumeric_Nonmissing_Cells"] == 0).all()
)
all_m2_finite = bool(
    partS0_feature_audit["Numeric_Values_Finite"].all()
)


# ============================================================
# 6. OUTCOME-LEVEL DISTRIBUTION AND MISSINGNESS AUDIT
# ============================================================

outcome_distribution_rows = []
outcome_audit_rows = []

for dataset_name, data in [
    ("2024 analytic source", df_2024),
    ("2025 temporal source", df_2025),
]:
    for outcome_name, specification in SECONDARY_OUTCOMES.items():
        column = specification["source_column"]
        numeric = pd.to_numeric(data[column], errors="coerce")
        nonnumeric_nonmissing = int(
            (data[column].notna() & numeric.isna()).sum()
        )
        observed_levels = sorted(
            numeric.dropna().astype(int).unique().tolist()
        )
        allowed_levels = set(observed_levels).issubset({0, 1, 2, 3})

        outcome_audit_rows.append({
            "Dataset": dataset_name,
            "Outcome": outcome_name,
            "Source_Column": column,
            "Derived_From": specification["derived_from"],
            "Observed_N": len(data),
            "Missing_Cells": int(data[column].isna().sum()),
            "Nonnumeric_Nonmissing_Cells": nonnumeric_nonmissing,
            "Observed_Levels": " | ".join(map(str, observed_levels)),
            "Levels_Within_0_to_3": allowed_levels,
            "Binary_Rule_Locked_For_PART_S1": (
                specification["binary_rule_for_part_s1"]
            ),
        })

        for level, count in (
            numeric.value_counts(dropna=False).sort_index().items()
        ):
            level_label = "Missing" if pd.isna(level) else str(int(level))
            outcome_distribution_rows.append({
                "Dataset": dataset_name,
                "Outcome": outcome_name,
                "Source_Column": column,
                "Original_Level": level_label,
                "N": int(count),
                "Percent": float(100 * count / len(data)),
            })

partS0_outcome_audit = pd.DataFrame(outcome_audit_rows)
partS0_outcome_distribution = pd.DataFrame(outcome_distribution_rows)

outcomes_complete = bool(
    (partS0_outcome_audit["Missing_Cells"] == 0).all()
)
outcomes_numeric = bool(
    (partS0_outcome_audit["Nonnumeric_Nonmissing_Cells"] == 0).all()
)
outcome_levels_valid = bool(
    partS0_outcome_audit["Levels_Within_0_to_3"].all()
)


# ============================================================
# 7. DIRECT ITEM-OVERLAP AND CONSTRUCT-OVERLAP AUDIT
# ============================================================

circularity_rows = []

for outcome_name, specification in SECONDARY_OUTCOMES.items():
    prohibited_prefixes = [
        prefix.upper()
        for prefix in specification["prohibited_predictor_prefixes"]
    ]
    prohibited_features_found = [
        feature
        for feature in M2_FEATURES
        if any(
            str(feature).upper().startswith(prefix)
            for prefix in prohibited_prefixes
        )
    ]
    direct_overlap = len(prohibited_features_found) > 0

    circularity_rows.append({
        "Outcome": outcome_name,
        "Outcome_Source_Column": specification["source_column"],
        "Outcome_Derived_From": specification["derived_from"],
        "Predictor_Strategy": PRESPECIFIED_STRATEGY,
        "Prohibited_Predictor_Families": " | ".join(
            specification["prohibited_predictor_prefixes"]
        ),
        "Prohibited_Features_Found": " | ".join(
            prohibited_features_found
        ),
        "Direct_Item_Overlap": direct_overlap,
        "Direct_Item_Overlap_Check_Passed": not direct_overlap,
        "Construct_Overlap": specification["construct_overlap"],
        "Required_Interpretation": specification["interpretation"],
        "Independently_Adjudicated_Outcome": False,
    })

partS0_circularity_construct_audit = pd.DataFrame(circularity_rows)

direct_overlap_check_passed = bool(
    partS0_circularity_construct_audit[
        "Direct_Item_Overlap_Check_Passed"
    ].all()
)

if not direct_overlap_check_passed:
    display(partS0_circularity_construct_audit)
    raise RuntimeError("Direct prohibited item overlap was detected.")


# ============================================================
# 8. CROSS-COHORT CATEGORY SUPPORT AUDIT
# ============================================================
# This audit describes whether temporal categorical values were
# observed in 2024. It does not modify, exclude, or encode values.
# ============================================================

category_support_rows = []

for feature in CATEGORICAL_OR_ORDINAL_FEATURES:
    values_2024 = set(df_2024[feature].dropna().tolist())
    values_2025 = set(df_2025[feature].dropna().tolist())
    unseen_temporal_values = sorted(
        values_2025.difference(values_2024),
        key=lambda value: str(value),
    )

    category_support_rows.append({
        "Feature": feature,
        "Development_2024_Category_Count": len(values_2024),
        "Temporal_2025_Category_Count": len(values_2025),
        "Unseen_Temporal_Category_Count": len(unseen_temporal_values),
        "No_Unseen_Temporal_Categories": (
            len(unseen_temporal_values) == 0
        ),
    })

partS0_cross_cohort_category_support = pd.DataFrame(
    category_support_rows
)

# Unseen categories are reported, not used as an automatic exclusion rule.


# ============================================================
# 9. DESIGN-LOCK TABLE
# ============================================================

partS0_design_lock = pd.DataFrame([
    {
        "Analysis": "Secondary questionnaire-defined outcomes",
        "Predictor_Strategy": PRESPECIFIED_STRATEGY,
        "Algorithm": PRESPECIFIED_ALGORITHM,
        "Feature_Count": len(M2_FEATURES),
        "Continuous_Features": "GPAX",
        "Categorical_or_Ordinal_Feature_Count": len(
            CATEGORICAL_OR_ORDINAL_FEATURES
        ),
        "Planned_Over_Sampler": "SMOTENC",
        "SMOTENC_Sampling_Strategy": 0.20,
        "SMOTENC_K_Neighbors": 5,
        "Planned_Under_Sampler": "RandomUnderSampler",
        "Under_Sampling_Strategy": 0.50,
        "Resampling_Location": "Development training folds only",
        "OOF_Folds": 5,
        "OOF_Random_State": 42,
        "Secondary_Threshold_Minimum_Sensitivity": 0.80,
        "Algorithm_Competition_Planned": False,
        "Heldout_Used_For_Development": False,
        "Temporal_Used_For_Development": False,
    }
])


# ============================================================
# 10. GLOBAL VERIFICATION
# ============================================================

global_checks = {
    "Dataset_Structure_Ready": bool(
        partS0_dataset_audit[
            ["N_Correct", "Required_Columns_Present", "No_Duplicate_Columns"]
        ].all(axis=None)
    ),
    **feature_definition_checks,
    "All_M2_Values_Complete": all_m2_complete,
    "All_M2_Values_Numeric_Encoded": all_m2_numeric_encoded,
    "All_M2_Numeric_Values_Finite": all_m2_finite,
    "Secondary_Outcome_Values_Complete": outcomes_complete,
    "Secondary_Outcome_Values_Numeric": outcomes_numeric,
    "Secondary_Outcome_Levels_Valid": outcome_levels_valid,
    "No_Direct_9Q_or_8Q_Item_Overlap": direct_overlap_check_passed,
    "Depression_Construct_Overlap_Documented": bool(
        partS0_circularity_construct_audit.loc[
            partS0_circularity_construct_audit["Outcome"]
            == "Depression_Risk",
            "Construct_Overlap",
        ].iloc[0] == "Substantial"
    ),
    "Suicide_Outcome_Not_Independently_Adjudicated_Documented": bool(
        not partS0_circularity_construct_audit.loc[
            partS0_circularity_construct_audit["Outcome"]
            == "Suicide_Risk",
            "Independently_Adjudicated_Outcome",
        ].iloc[0]
    ),
}

PARTS0_READY = bool(all(global_checks.values()))

partS0_verification = pd.DataFrame([{
    "Prespecified_Strategy": PRESPECIFIED_STRATEGY,
    "Prespecified_Algorithm": PRESPECIFIED_ALGORITHM,
    **global_checks,
    "Binary_Outcome_Encoding_Performed": False,
    "Development_Heldout_Split_Performed": False,
    "Model_Fitted": False,
    "Resampling_Performed": False,
    "Prediction_Generated": False,
    "Threshold_Selected": False,
    "Calibration_Performed": False,
    "Bootstrap_Performed": False,
    "Participant_Level_Data_Exported": False,
    "PARTS0_Ready": PARTS0_READY,
}])


# ============================================================
# 11. DISPLAY AGGREGATE AUDIT RESULTS
# ============================================================

print("\nDataset audit:")
display(partS0_dataset_audit)

print("\nLocked M2 feature definition:")
display(partS0_feature_definition)

print("\nFeature audit:")
display(partS0_feature_audit)

print("\nOutcome audit:")
display(partS0_outcome_audit)

print("\nOriginal outcome-level distributions:")
display(partS0_outcome_distribution)

print("\nDirect item-overlap and construct-overlap audit:")
display(partS0_circularity_construct_audit)

print("\nCross-cohort category-support audit:")
display(partS0_cross_cohort_category_support)

print("\nDesign lock:")
display(partS0_design_lock)

print("\nGlobal verification:")
display(partS0_verification)


# ============================================================
# 12. EXPORT AGGREGATE METADATA ONLY
# ============================================================

partS0_dataset_audit.to_csv(
    "partS0_dataset_audit.csv",
    index=False,
    encoding="utf-8-sig",
)

partS0_feature_definition.to_csv(
    "partS0_m2_feature_definition.csv",
    index=False,
    encoding="utf-8-sig",
)

partS0_feature_audit.to_csv(
    "partS0_feature_audit.csv",
    index=False,
    encoding="utf-8-sig",
)

partS0_outcome_audit.to_csv(
    "partS0_outcome_audit.csv",
    index=False,
    encoding="utf-8-sig",
)

partS0_outcome_distribution.to_csv(
    "partS0_outcome_distribution_aggregate.csv",
    index=False,
    encoding="utf-8-sig",
)

partS0_circularity_construct_audit.to_csv(
    "partS0_circularity_construct_audit.csv",
    index=False,
    encoding="utf-8-sig",
)

partS0_cross_cohort_category_support.to_csv(
    "partS0_cross_cohort_category_support.csv",
    index=False,
    encoding="utf-8-sig",
)

partS0_design_lock.to_csv(
    "partS0_design_lock.csv",
    index=False,
    encoding="utf-8-sig",
)

partS0_verification.to_csv(
    "partS0_verification.csv",
    index=False,
    encoding="utf-8-sig",
)


# ============================================================
# 13. FINAL SAFETY GATE AND SUMMARY
# ============================================================

if not PARTS0_READY:
    failed_checks = [
        name for name, passed in global_checks.items()
        if not bool(passed)
    ]
    raise RuntimeError(
        "PART S0 verification failed: " + ", ".join(failed_checks)
    )

print("\n" + "=" * 105)
print("PART S0 VERSION 2.0 COMPLETED")
print("=" * 105)
print(f"2024 source                    : N={len(df_2024):,}")
print(f"2025 temporal source           : N={len(df_2025):,}")
print(f"Predictor strategy             : {PRESPECIFIED_STRATEGY}")
print(f"Algorithm                      : {PRESPECIFIED_ALGORITHM}")
print(f"Administrative predictors      : {len(ADMIN_FEATURES)}")
print(f"Thai 2Q items                  : {len(THAI_2Q_FEATURES)}")
print(f"Continuous predictors          : {CONTINUOUS_FEATURES}")
print(f"Categorical/ordinal predictors : {len(CATEGORICAL_OR_ORDINAL_FEATURES)}")
print("Direct 9Q/8Q item overlap      : No")
print("Depression construct overlap   : Substantial; documented")
print("Suicide outcome adjudication   : Questionnaire-defined")
print("Binary outcomes created        : No")
print("Split created                  : No")
print("Model fitted                   : No")
print("Resampling performed           : No")
print("Participant rows exported      : No")
print("\nPARTS0_READY:", PARTS0_READY)
print("\nProceed to PART S1 only after PARTS0_READY=True.")
```

## PART S1: Binary Secondary Outcome Encoding and Secondary Development/Held-out Split

```python
# ============================================================
# PART S1 VERSION 2.0
# BINARY SECONDARY OUTCOMES AND SHARED DEVELOPMENT-HELD-OUT SPLIT
# ============================================================
#
# Purpose:
# - Create the two locked questionnaire-defined binary outcomes.
# - Create one shared 70:30 participant split in the 2024 cohort.
# - Stratify the shared split by clinically diagnosed MDD, as stated
#   in the manuscript.
# - Apply the same participants to both secondary outcomes.
# - Preserve the complete 2025 cohort for temporal validation.
#
# Locked binary definitions:
# - Depression_Risk = 1 when depressionlevel >= 1.
# - Suicide_Risk = 1 when suicidelevel >= 1.
#
# Safeguards:
# - The split is created once in this PART and must not be recreated.
# - No model is fitted.
# - No resampling is performed.
# - No prediction or threshold selection is performed.
# - Participant-level split manifests and row positions are retained
#   in memory only and are not exported.
# ============================================================

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split


# ============================================================
# 1. SETTINGS
# ============================================================

SECONDARY_TEST_SIZE = 0.30
SECONDARY_RANDOM_STATE = 42

EXPECTED_FULL_2024_N = 4090
EXPECTED_FULL_2024_MDD_EVENTS = 46
EXPECTED_DEVELOPMENT_N = 2863
EXPECTED_DEVELOPMENT_MDD_EVENTS = 32
EXPECTED_HELDOUT_N = 1227
EXPECTED_HELDOUT_MDD_EVENTS = 14
EXPECTED_TEMPORAL_N = 815
EXPECTED_TEMPORAL_MDD_EVENTS = 16

SECONDARY_TARGETS = [
    "Depression_Risk",
    "Suicide_Risk",
]

BINARY_OUTCOME_DEFINITIONS = {
    "Depression_Risk": {
        "source_column": "depressionlevel",
        "positive_rule": "depressionlevel >= 1",
        "derived_from": "9Q-based depression level",
    },
    "Suicide_Risk": {
        "source_column": "suicidelevel",
        "positive_rule": "suicidelevel >= 1",
        "derived_from": "8Q-based suicide-risk level",
    },
}


# ============================================================
# 2. VERIFY PART S0
# ============================================================

required_objects = [
    "PARTS0_READY",
    "partS0_verification",
    "df_2024",
    "df_2025",
    "M2_FEATURES",
    "EXPECTED_M2_FEATURES",
    "CATEGORICAL_INDICES",
    "PRESPECIFIED_STRATEGY",
    "PRESPECIFIED_ALGORITHM",
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required PART S0 objects: "
        + ", ".join(missing_objects)
    )

if not bool(PARTS0_READY):
    raise RuntimeError("PART S0 is not ready.")

if not bool(partS0_verification["PARTS0_Ready"].iloc[0]):
    raise RuntimeError("PART S0 verification table is not ready.")

if list(M2_FEATURES) != list(EXPECTED_M2_FEATURES):
    raise RuntimeError("The locked M2 feature definition has changed.")

if list(CATEGORICAL_INDICES) != [0, 2, 3, 4, 5, 6, 7, 8, 9]:
    raise RuntimeError("The locked SMOTENC categorical indices changed.")


# ============================================================
# 3. COPY SOURCE DATA AND PRESERVE SOURCE ROW POSITION IN MEMORY
# ============================================================

secondary_2024 = df_2024.copy()
secondary_2025 = df_2025.copy()

secondary_2024.insert(
    0,
    "Source_Row_Position",
    np.arange(len(secondary_2024), dtype=int),
)

secondary_2025.insert(
    0,
    "Temporal_Row_Position",
    np.arange(len(secondary_2025), dtype=int),
)

if len(secondary_2024) != EXPECTED_FULL_2024_N:
    raise RuntimeError("Unexpected full 2024 cohort size.")

if len(secondary_2025) != EXPECTED_TEMPORAL_N:
    raise RuntimeError("Unexpected temporal 2025 cohort size.")


# ============================================================
# 4. VERIFY CLINICALLY DIAGNOSED MDD STRATIFICATION VARIABLE
# ============================================================

for dataset_name, data, expected_events in [
    ("2024", secondary_2024, EXPECTED_FULL_2024_MDD_EVENTS),
    ("2025", secondary_2025, EXPECTED_TEMPORAL_MDD_EVENTS),
]:
    if "mdddx" not in data.columns:
        raise KeyError(f"{dataset_name}: mdddx is missing.")

    mdd = pd.to_numeric(data["mdddx"], errors="raise")
    if mdd.isna().any():
        raise RuntimeError(f"{dataset_name}: mdddx contains missing values.")
    mdd = mdd.astype(int)

    if set(mdd.unique()) != {0, 1}:
        raise RuntimeError(f"{dataset_name}: mdddx is not binary.")
    if int(mdd.sum()) != expected_events:
        raise RuntimeError(
            f"{dataset_name}: expected {expected_events} MDD events, "
            f"observed {int(mdd.sum())}."
        )

    data["mdddx"] = mdd


# ============================================================
# 5. CREATE LOCKED BINARY SECONDARY OUTCOMES
# ============================================================

binary_definition_rows = []

for outcome, specification in BINARY_OUTCOME_DEFINITIONS.items():
    source_column = specification["source_column"]

    for dataset_name, data in [
        ("2024 full cohort", secondary_2024),
        ("2025 temporal cohort", secondary_2025),
    ]:
        if source_column not in data.columns:
            raise KeyError(
                f"{dataset_name}: {source_column} is missing."
            )

        original_level = pd.to_numeric(
            data[source_column],
            errors="raise",
        )
        if original_level.isna().any():
            raise RuntimeError(
                f"{dataset_name}: {source_column} contains missing values."
            )
        if not set(original_level.astype(int).unique()).issubset(
            {0, 1, 2, 3}
        ):
            raise RuntimeError(
                f"{dataset_name}: invalid levels in {source_column}."
            )

        data[outcome] = (original_level >= 1).astype(int)

        binary_definition_rows.append({
            "Dataset": dataset_name,
            "Outcome": outcome,
            "Source_Column": source_column,
            "Derived_From": specification["derived_from"],
            "Positive_Rule": specification["positive_rule"],
            "N": len(data),
            "Events": int(data[outcome].sum()),
            "Nonevents": int((data[outcome] == 0).sum()),
            "Prevalence": float(data[outcome].mean()),
            "Missing": int(data[outcome].isna().sum()),
            "Binary_Outcome": set(data[outcome].unique()) == {0, 1},
        })

partS1_binary_outcome_definitions = pd.DataFrame(
    binary_definition_rows
)

if not partS1_binary_outcome_definitions["Binary_Outcome"].all():
    raise RuntimeError("A secondary binary outcome is invalid.")

if not (partS1_binary_outcome_definitions["Missing"] == 0).all():
    raise RuntimeError("A secondary binary outcome contains missing values.")


# ============================================================
# 6. CREATE ONE SHARED 2024 DEVELOPMENT-HELD-OUT SPLIT
# ============================================================

all_secondary_positions = np.arange(
    len(secondary_2024),
    dtype=int,
)

secondary_train_idx, secondary_heldout_idx = train_test_split(
    all_secondary_positions,
    test_size=SECONDARY_TEST_SIZE,
    random_state=SECONDARY_RANDOM_STATE,
    shuffle=True,
    stratify=secondary_2024["mdddx"].to_numpy(dtype=int),
)

# Sorting fixes row order for reproducibility without changing membership.
secondary_train_idx = np.sort(secondary_train_idx)
secondary_heldout_idx = np.sort(secondary_heldout_idx)

secondary_train_2024 = (
    secondary_2024.iloc[secondary_train_idx]
    .copy()
    .reset_index(drop=True)
)

secondary_heldout_2024 = (
    secondary_2024.iloc[secondary_heldout_idx]
    .copy()
    .reset_index(drop=True)
)

secondary_temporal_2025 = (
    secondary_2025.copy()
    .reset_index(drop=True)
)


# ============================================================
# 7. VERIFY SPLIT MEMBERSHIP AND INDEPENDENCE
# ============================================================

train_position_set = set(secondary_train_idx.tolist())
heldout_position_set = set(secondary_heldout_idx.tolist())
full_position_set = set(all_secondary_positions.tolist())

train_heldout_overlap = train_position_set.intersection(
    heldout_position_set
)
split_union = train_position_set.union(heldout_position_set)

split_membership_checks = {
    "Development_N_Correct": (
        len(secondary_train_2024) == EXPECTED_DEVELOPMENT_N
    ),
    "Heldout_N_Correct": (
        len(secondary_heldout_2024) == EXPECTED_HELDOUT_N
    ),
    "No_Development_Heldout_Overlap": (
        len(train_heldout_overlap) == 0
    ),
    "Every_2024_Participant_Assigned_Once": (
        split_union == full_position_set
        and len(train_position_set) + len(heldout_position_set)
        == len(full_position_set)
    ),
    "Development_Source_Positions_Unique": (
        secondary_train_2024["Source_Row_Position"].nunique()
        == len(secondary_train_2024)
    ),
    "Heldout_Source_Positions_Unique": (
        secondary_heldout_2024["Source_Row_Position"].nunique()
        == len(secondary_heldout_2024)
    ),
    "Temporal_Row_Positions_Unique": (
        secondary_temporal_2025["Temporal_Row_Position"].nunique()
        == len(secondary_temporal_2025)
    ),
}

if not all(split_membership_checks.values()):
    failed = [
        name for name, passed in split_membership_checks.items()
        if not passed
    ]
    raise RuntimeError(
        "Shared split membership verification failed: "
        + ", ".join(failed)
    )


# ============================================================
# 8. VERIFY MDD STRATIFICATION RESULT
# ============================================================

mdd_stratification_checks = {
    "Full_2024_MDD_Events_Correct": (
        int(secondary_2024["mdddx"].sum())
        == EXPECTED_FULL_2024_MDD_EVENTS
    ),
    "Development_MDD_Events_Correct": (
        int(secondary_train_2024["mdddx"].sum())
        == EXPECTED_DEVELOPMENT_MDD_EVENTS
    ),
    "Heldout_MDD_Events_Correct": (
        int(secondary_heldout_2024["mdddx"].sum())
        == EXPECTED_HELDOUT_MDD_EVENTS
    ),
    "Temporal_MDD_Events_Correct": (
        int(secondary_temporal_2025["mdddx"].sum())
        == EXPECTED_TEMPORAL_MDD_EVENTS
    ),
}

if not all(mdd_stratification_checks.values()):
    raise RuntimeError("MDD stratification-count verification failed.")


# ============================================================
# 9. COHORT AND SECONDARY-OUTCOME SUMMARY
# ============================================================

cohort_objects = {
    "Development 2024": secondary_train_2024,
    "Held-out 2024": secondary_heldout_2024,
    "Temporal 2025": secondary_temporal_2025,
}

cohort_summary_rows = []

for cohort_name, data in cohort_objects.items():
    for outcome in SECONDARY_TARGETS:
        X = data[M2_FEATURES]
        y = data[outcome]

        cohort_summary_rows.append({
            "Cohort": cohort_name,
            "Outcome": outcome,
            "N": len(data),
            "Events": int(y.sum()),
            "Nonevents": int((y == 0).sum()),
            "Prevalence": float(y.mean()),
            "MDD_Events": int(data["mdddx"].sum()),
            "M2_Feature_Count": X.shape[1],
            "M2_Feature_Order_Correct": (
                X.columns.tolist() == M2_FEATURES
            ),
            "Predictor_Missing_Cells": int(X.isna().sum().sum()),
            "Outcome_Missing_Cells": int(y.isna().sum()),
            "Outcome_Binary": set(y.unique()) == {0, 1},
        })

partS1_cohort_outcome_summary = pd.DataFrame(
    cohort_summary_rows
)


# ============================================================
# 10. CHECK OOF FEASIBILITY FOR EACH SECONDARY OUTCOME
# ============================================================

oof_feasibility_rows = []

for outcome in SECONDARY_TARGETS:
    y = secondary_train_2024[outcome].astype(int)
    events = int(y.sum())
    nonevents = int((y == 0).sum())
    minority_n = min(events, nonevents)

    oof_feasibility_rows.append({
        "Outcome": outcome,
        "Development_N": len(y),
        "Events": events,
        "Nonevents": nonevents,
        "Minority_Class_N": minority_n,
        "Five_Fold_Stratification_Feasible": minority_n >= 5,
        "SMOTENC_K5_Feasible_In_Principle": minority_n >= 6,
    })

partS1_oof_feasibility = pd.DataFrame(oof_feasibility_rows)

if not partS1_oof_feasibility[
    [
        "Five_Fold_Stratification_Feasible",
        "SMOTENC_K5_Feasible_In_Principle",
    ]
].all(axis=None):
    raise RuntimeError(
        "A secondary outcome is not feasible for planned 5-fold OOF "
        "SMOTENC development."
    )


# ============================================================
# 11. AGGREGATE SPLIT DESIGN TABLE
# ============================================================

partS1_split_design = pd.DataFrame([
    {
        "Source_Cohort": "2024 analytic cohort",
        "Split_Role": "Development",
        "N": len(secondary_train_2024),
        "MDD_Events": int(secondary_train_2024["mdddx"].sum()),
        "Split_Fraction": 0.70,
        "Random_State": SECONDARY_RANDOM_STATE,
        "Stratification_Variable": "Clinically diagnosed MDD (mdddx)",
        "Used_For_Model_Fitting": True,
        "Used_For_Threshold_Selection": True,
    },
    {
        "Source_Cohort": "2024 analytic cohort",
        "Split_Role": "Held-out validation",
        "N": len(secondary_heldout_2024),
        "MDD_Events": int(secondary_heldout_2024["mdddx"].sum()),
        "Split_Fraction": 0.30,
        "Random_State": SECONDARY_RANDOM_STATE,
        "Stratification_Variable": "Clinically diagnosed MDD (mdddx)",
        "Used_For_Model_Fitting": False,
        "Used_For_Threshold_Selection": False,
    },
    {
        "Source_Cohort": "2025 temporal cohort",
        "Split_Role": "Temporal validation",
        "N": len(secondary_temporal_2025),
        "MDD_Events": int(secondary_temporal_2025["mdddx"].sum()),
        "Split_Fraction": np.nan,
        "Random_State": np.nan,
        "Stratification_Variable": "Not applicable",
        "Used_For_Model_Fitting": False,
        "Used_For_Threshold_Selection": False,
    },
])


# ============================================================
# 12. GLOBAL VERIFICATION
# ============================================================

global_checks = {
    "PARTS0_Ready": bool(PARTS0_READY),
    "Binary_Outcomes_Valid": bool(
        partS1_binary_outcome_definitions["Binary_Outcome"].all()
        and (partS1_binary_outcome_definitions["Missing"] == 0).all()
    ),
    **split_membership_checks,
    **mdd_stratification_checks,
    "All_Cohort_Outcome_Rows_Ready": bool(
        partS1_cohort_outcome_summary[
            ["M2_Feature_Order_Correct", "Outcome_Binary"]
        ].all(axis=None)
        and (
            partS1_cohort_outcome_summary["Predictor_Missing_Cells"] == 0
        ).all()
        and (
            partS1_cohort_outcome_summary["Outcome_Missing_Cells"] == 0
        ).all()
    ),
    "Both_Outcomes_Five_Fold_OOF_Feasible": bool(
        partS1_oof_feasibility[
            "Five_Fold_Stratification_Feasible"
        ].all()
    ),
    "Both_Outcomes_SMOTENC_K5_Feasible": bool(
        partS1_oof_feasibility[
            "SMOTENC_K5_Feasible_In_Principle"
        ].all()
    ),
}

PARTS1_READY = bool(all(global_checks.values()))

partS1_verification = pd.DataFrame([{
    "Predictor_Strategy": PRESPECIFIED_STRATEGY,
    "Algorithm": PRESPECIFIED_ALGORITHM,
    "Shared_Split_Created_Once": True,
    "Split_Applied_To_Both_Secondary_Outcomes": True,
    **global_checks,
    "Model_Fitted": False,
    "Resampling_Performed": False,
    "Prediction_Generated": False,
    "Threshold_Selected": False,
    "Heldout_Used_For_Development": False,
    "Temporal_Used_For_Development": False,
    "Participant_Level_Manifest_Exported": False,
    "PARTS1_Ready": PARTS1_READY,
}])


# ============================================================
# 13. DISPLAY AGGREGATE RESULTS
# ============================================================

print("=" * 105)
print("PART S1 VERSION 2.0")
print("BINARY SECONDARY OUTCOMES AND SHARED 70:30 SPLIT")
print("=" * 105)

print("\nBinary outcome definitions:")
display(partS1_binary_outcome_definitions)

print("\nShared split design:")
display(partS1_split_design)

print("\nCohort and secondary-outcome summary:")
display(partS1_cohort_outcome_summary)

print("\nOOF feasibility:")
display(partS1_oof_feasibility)

print("\nGlobal verification:")
display(partS1_verification)


# ============================================================
# 14. EXPORT AGGREGATE METADATA ONLY
# ============================================================

partS1_binary_outcome_definitions.to_csv(
    "partS1_binary_outcome_definitions.csv",
    index=False,
    encoding="utf-8-sig",
)

partS1_split_design.to_csv(
    "partS1_shared_split_design.csv",
    index=False,
    encoding="utf-8-sig",
)

partS1_cohort_outcome_summary.to_csv(
    "partS1_cohort_outcome_summary.csv",
    index=False,
    encoding="utf-8-sig",
)

partS1_oof_feasibility.to_csv(
    "partS1_oof_feasibility.csv",
    index=False,
    encoding="utf-8-sig",
)

partS1_verification.to_csv(
    "partS1_verification.csv",
    index=False,
    encoding="utf-8-sig",
)


# ============================================================
# 15. FINAL SAFETY GATE AND SUMMARY
# ============================================================

if not PARTS1_READY:
    failed_checks = [
        name for name, passed in global_checks.items()
        if not bool(passed)
    ]
    raise RuntimeError(
        "PART S1 verification failed: " + ", ".join(failed_checks)
    )

print("\n" + "=" * 105)
print("PART S1 VERSION 2.0 COMPLETED")
print("=" * 105)
print("Binary depression risk : depressionlevel >= 1")
print("Binary suicide risk    : suicidelevel >= 1")
print("Shared split source    : Full 2024 cohort")
print("Split stratification   : Clinically diagnosed MDD")
print("Random state           : 42")
print(
    f"Development 2024      : N={len(secondary_train_2024):,}; "
    f"MDD events={int(secondary_train_2024['mdddx'].sum())}"
)
print(
    f"Held-out 2024         : N={len(secondary_heldout_2024):,}; "
    f"MDD events={int(secondary_heldout_2024['mdddx'].sum())}"
)
print(
    f"Temporal 2025         : N={len(secondary_temporal_2025):,}; "
    f"MDD events={int(secondary_temporal_2025['mdddx'].sum())}"
)
print("Split recreated later : Prohibited")
print("Model fitted           : No")
print("Resampling performed   : No")
print("Participant rows export: No")
print("\nPARTS1_READY:", PARTS1_READY)
print("\nProceed to PART S2 only after PARTS1_READY=True.")
```

## PART S2: Prespecified M2_Admin_2Q + AdaBoost Outcome-Specific Model Development

```python
# ============================================================
# PART S2 VERSION 2.1
# OUTCOME-SPECIFIC SMOTENC-ADABOOST DEVELOPMENT,
# GENUINE OOF PREDICTION, AND THRESHOLD LOCKING
# ============================================================
#
# Secondary outcomes:
# - Depression_Risk
# - Suicide_Risk
#
# Fixed model specification:
# - M2_Admin_2Q predictors
# - AdaBoostClassifier
# - GPAX is the only continuous predictor
# - All remaining predictors, including 2Q1 and 2Q2, are treated
#   as categorical/ordinal for SMOTENC
#
# Resampling specification:
# - SMOTENC minority-to-majority ratio: 0.20
# - SMOTENC k_neighbors: 5
# - Random undersampling final minority-to-majority ratio: 0.50
# - Resampling occurs within development training folds only
#
# Threshold specification:
# - Genuine 5-fold stratified OOF predictions only
# - Maximize F1 subject to sensitivity >= 0.80 when feasible
# - Eligible ties: precision, specificity, higher threshold
# - Fallback ties: sensitivity, precision, specificity,
#   higher threshold
#
# Held-out 2024 and temporal 2025 cohorts are not evaluated here.
# Participant-level OOF predictions remain in memory and are not
# exported.
# ============================================================

import numpy as np
import pandas as pd
from sklearn.base import clone
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
)
from imblearn.over_sampling import SMOTENC
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline as ImbPipeline


# ============================================================
# 1. SETTINGS
# ============================================================

SECONDARY_MODEL_NAME = "M2_Admin_2Q"
SECONDARY_ALGORITHM_NAME = "AdaBoostClassifier"

SECONDARY_TARGETS = [
    "Depression_Risk",
    "Suicide_Risk",
]

N_SPLITS = 5
CV_RANDOM_STATE = 42
BASE_RESAMPLING_RANDOM_STATE = 42
TARGET_SENSITIVITY = 0.80
THRESHOLD_EPSILON = 1e-12

SMOTENC_SAMPLING_STRATEGY = 0.20
SMOTENC_K_NEIGHBORS = 5
UNDER_SAMPLING_STRATEGY = 0.50

ADABOOST_N_ESTIMATORS = 100
ADABOOST_LEARNING_RATE = 0.5
ADABOOST_RANDOM_STATE = 42

EXPECTED_DEVELOPMENT_N = 2863


# ============================================================
# 2. VERIFY PART S1 AND LOCKED DESIGN
# ============================================================

required_objects = [
    "PARTS1_READY",
    "partS1_verification",
    "secondary_train_2024",
    "secondary_heldout_2024",
    "secondary_temporal_2025",
    "M2_FEATURES",
    "EXPECTED_M2_FEATURES",
    "CONTINUOUS_FEATURES",
    "CATEGORICAL_OR_ORDINAL_FEATURES",
    "CATEGORICAL_INDICES",
    "PRESPECIFIED_STRATEGY",
    "PRESPECIFIED_ALGORITHM",
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required PART S1 objects: "
        + ", ".join(missing_objects)
    )

if not bool(PARTS1_READY):
    raise RuntimeError("PART S1 is not ready.")

if not bool(partS1_verification["PARTS1_Ready"].iloc[0]):
    raise RuntimeError("PART S1 verification table is not ready.")

if list(M2_FEATURES) != list(EXPECTED_M2_FEATURES):
    raise RuntimeError("The locked M2 feature definition changed.")

if list(CONTINUOUS_FEATURES) != ["GPAX"]:
    raise RuntimeError("GPAX must be the only continuous predictor.")

if list(CATEGORICAL_INDICES) != [0, 2, 3, 4, 5, 6, 7, 8, 9]:
    raise RuntimeError("The SMOTENC categorical indices changed.")

if PRESPECIFIED_STRATEGY != SECONDARY_MODEL_NAME:
    raise RuntimeError("The prespecified secondary strategy changed.")

if PRESPECIFIED_ALGORITHM != SECONDARY_ALGORITHM_NAME:
    raise RuntimeError("The prespecified secondary algorithm changed.")


# ============================================================
# 3. DEVELOPMENT DATA
# ============================================================

required_development_columns = M2_FEATURES + SECONDARY_TARGETS + [
    "Source_Row_Position"
]

missing_development_columns = [
    column for column in required_development_columns
    if column not in secondary_train_2024.columns
]

if missing_development_columns:
    raise KeyError(
        "Development cohort is missing: "
        + ", ".join(missing_development_columns)
    )

X_secondary_development = (
    secondary_train_2024[M2_FEATURES]
    .copy()
    .reset_index(drop=True)
)

development_source_positions = (
    secondary_train_2024["Source_Row_Position"]
    .to_numpy(dtype=int)
)

if len(X_secondary_development) != EXPECTED_DEVELOPMENT_N:
    raise RuntimeError("Unexpected development cohort size.")

if X_secondary_development.columns.tolist() != M2_FEATURES:
    raise RuntimeError("Development M2 feature order is incorrect.")

if X_secondary_development.isna().any().any():
    raise RuntimeError("Development M2 predictors contain missing values.")

for feature in M2_FEATURES:
    converted = pd.to_numeric(
        X_secondary_development[feature],
        errors="raise",
    )
    if not np.isfinite(converted.to_numpy(dtype=float)).all():
        raise RuntimeError(f"{feature} contains nonfinite values.")
    X_secondary_development[feature] = converted


# ============================================================
# 4. MODEL AND SAMPLER FACTORIES
# ============================================================

def make_adaboost():
    return AdaBoostClassifier(
        n_estimators=ADABOOST_N_ESTIMATORS,
        learning_rate=ADABOOST_LEARNING_RATE,
        random_state=ADABOOST_RANDOM_STATE,
    )


def make_smotenc(random_state):
    return SMOTENC(
        categorical_features=CATEGORICAL_INDICES,
        sampling_strategy=SMOTENC_SAMPLING_STRATEGY,
        k_neighbors=SMOTENC_K_NEIGHBORS,
        random_state=random_state,
    )


def make_under_sampler(random_state):
    return RandomUnderSampler(
        sampling_strategy=UNDER_SAMPLING_STRATEGY,
        random_state=random_state,
    )


def make_final_pipeline(random_state):
    return ImbPipeline(steps=[
        ("over_sampler", make_smotenc(random_state)),
        ("under_sampler", make_under_sampler(random_state)),
        ("estimator", make_adaboost()),
    ])


# ============================================================
# 5. PROBABILITY AND METRIC HELPERS
# ============================================================

def positive_probability(estimator, X):
    matrix = np.asarray(estimator.predict_proba(X), dtype=float)
    classes = np.asarray(estimator.classes_)
    positive_positions = np.flatnonzero(classes == 1)

    if matrix.ndim != 2 or len(positive_positions) != 1:
        raise ValueError("Unable to identify class-1 probability.")

    probability = matrix[:, int(positive_positions[0])]

    if (
        not np.isfinite(probability).all()
        or (probability < 0).any()
        or (probability > 1).any()
    ):
        raise ValueError("Predicted probabilities are invalid.")

    return probability


def calculate_binary_metrics(y_true, y_prob, threshold):
    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)
    y_pred = (y_prob >= float(threshold)).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1],
    ).ravel()

    sensitivity = recall_score(y_true, y_pred, zero_division=0)
    specificity = tn / (tn + fp) if (tn + fp) else np.nan
    precision = precision_score(y_true, y_pred, zero_division=0)

    return {
        "AUC": float(roc_auc_score(y_true, y_prob)),
        "Average_Precision": float(
            average_precision_score(y_true, y_prob)
        ),
        "Brier_Score": float(brier_score_loss(y_true, y_prob)),
        "Sensitivity": float(sensitivity),
        "Specificity": float(specificity),
        "Precision": float(precision),
        "F1": float(f1_score(y_true, y_pred, zero_division=0)),
        "Predicted_Positive_N": int(y_pred.sum()),
        "Predicted_Positive_Rate": float(y_pred.mean()),
        "TP": int(tp),
        "FP": int(fp),
        "FN": int(fn),
        "TN": int(tn),
        "y_pred": y_pred,
    }


# ============================================================
# 6. EMPIRICAL THRESHOLD SEARCH
# ============================================================

def create_threshold_candidates(y_prob):
    y_prob = np.asarray(y_prob, dtype=float)

    if not np.isfinite(y_prob).all():
        raise ValueError("OOF probabilities contain nonfinite values.")

    if (y_prob < 0).any() or (y_prob > 1).any():
        raise ValueError("OOF probabilities must be within [0, 1].")

    unique_probability = np.unique(y_prob)
    lower_boundary = max(
        0.0,
        np.nextafter(float(unique_probability.min()), -np.inf),
    )
    upper_boundary = min(
        1.0 + THRESHOLD_EPSILON,
        np.nextafter(float(unique_probability.max()), np.inf),
    )

    return np.unique(np.concatenate([
        [lower_boundary],
        unique_probability,
        [upper_boundary],
    ]))


def search_and_select_threshold(
    y_true,
    y_prob,
    target_sensitivity=TARGET_SENSITIVITY,
):
    search_rows = []

    for threshold in create_threshold_candidates(y_prob):
        metrics = calculate_binary_metrics(y_true, y_prob, threshold)
        search_rows.append({
            "Threshold": float(threshold),
            "Sensitivity": metrics["Sensitivity"],
            "Specificity": metrics["Specificity"],
            "Precision": metrics["Precision"],
            "F1": metrics["F1"],
            "TP": metrics["TP"],
            "FP": metrics["FP"],
            "FN": metrics["FN"],
            "TN": metrics["TN"],
            "Predicted_Positive_N": metrics["Predicted_Positive_N"],
        })

    search_table = pd.DataFrame(search_rows)
    eligible = search_table.loc[
        search_table["Sensitivity"] >= target_sensitivity
    ].copy()

    if not eligible.empty:
        selected = eligible.sort_values(
            by=["F1", "Precision", "Specificity", "Threshold"],
            ascending=[False, False, False, False],
            kind="mergesort",
        ).iloc[0]
        selection_rule = (
            f"Sensitivity>={target_sensitivity:.2f}; max F1; "
            "tie: precision, specificity, higher threshold"
        )
        target_achieved = True
    else:
        selected = search_table.sort_values(
            by=[
                "F1",
                "Sensitivity",
                "Precision",
                "Specificity",
                "Threshold",
            ],
            ascending=[False, False, False, False, False],
            kind="mergesort",
        ).iloc[0]
        selection_rule = (
            "Fallback: max F1 overall; tie: sensitivity, precision, "
            "specificity, higher threshold"
        )
        target_achieved = False

    return (
        float(selected["Threshold"]),
        selected.copy(),
        search_table,
        selection_rule,
        target_achieved,
        int(len(eligible)),
    )


# ============================================================
# 7. CATEGORY-VALIDITY HELPER
# ============================================================

def validate_resampled_categories(
    X_original,
    X_resampled,
    outcome,
    fold,
):
    if not isinstance(X_resampled, pd.DataFrame):
        X_resampled = pd.DataFrame(
            X_resampled,
            columns=M2_FEATURES,
        )

    rows = []
    all_valid = True

    for feature_index in CATEGORICAL_INDICES:
        feature = M2_FEATURES[feature_index]
        original_values = set(X_original[feature].dropna().tolist())
        resampled_values = set(X_resampled[feature].dropna().tolist())
        unexpected_values = resampled_values.difference(original_values)
        valid = len(unexpected_values) == 0
        all_valid = all_valid and valid

        rows.append({
            "Outcome": outcome,
            "Fold": fold,
            "Feature": feature,
            "Feature_Index": feature_index,
            "Original_Category_Count": len(original_values),
            "Resampled_Category_Count": len(resampled_values),
            "Unexpected_Category_Count": len(unexpected_values),
            "Category_Valid": valid,
        })

    return all_valid, rows


# ============================================================
# 8. STORAGE OBJECTS
# ============================================================

secondary_oof_predictions = {}
secondary_threshold_tables = {}
secondary_locked_thresholds = {}
secondary_fitted_pipelines = {}
secondary_fold_assignments = {}

oof_result_rows = []
fold_manifest_rows = []
resampling_rows = []
category_validation_rows = []


# ============================================================
# 9. OUTCOME-SPECIFIC OOF DEVELOPMENT
# ============================================================

for outcome_index, outcome in enumerate(SECONDARY_TARGETS):
    y_development = (
        secondary_train_2024[outcome]
        .astype(int)
        .copy()
        .reset_index(drop=True)
    )

    if set(y_development.unique()) != {0, 1}:
        raise RuntimeError(f"{outcome}: development outcome is not binary.")

    n = len(y_development)
    events = int(y_development.sum())
    nonevents = int((y_development == 0).sum())

    oof_probability = np.full(n, np.nan, dtype=float)
    oof_fold = np.full(n, -1, dtype=int)
    validation_counter = np.zeros(n, dtype=int)

    cv = StratifiedKFold(
        n_splits=N_SPLITS,
        shuffle=True,
        random_state=CV_RANDOM_STATE,
    )

    for fold_number, (train_position, validation_position) in enumerate(
        cv.split(X_secondary_development, y_development),
        start=1,
    ):
        X_fold_train = X_secondary_development.iloc[
            train_position
        ].copy()
        y_fold_train = y_development.iloc[train_position].copy()
        X_fold_validation = X_secondary_development.iloc[
            validation_position
        ].copy()
        y_fold_validation = y_development.iloc[
            validation_position
        ].copy()

        fold_random_state = (
            BASE_RESAMPLING_RANDOM_STATE
            + outcome_index * 100
            + fold_number
        )

        before_events = int(y_fold_train.sum())
        before_nonevents = int((y_fold_train == 0).sum())
        before_ratio = min(before_events, before_nonevents) / max(
            before_events, before_nonevents
        )

        if before_ratio >= SMOTENC_SAMPLING_STRATEGY:
            # This safeguard is reported explicitly. It prevents SMOTENC
            # from attempting to reduce an already larger minority class.
            X_after_over = X_fold_train.copy()
            y_after_over = y_fold_train.copy()
            over_sampler_applied = False
        else:
            over_sampler = make_smotenc(fold_random_state)
            X_after_over, y_after_over = over_sampler.fit_resample(
                X_fold_train,
                y_fold_train,
            )
            over_sampler_applied = True

        if not isinstance(X_after_over, pd.DataFrame):
            X_after_over = pd.DataFrame(
                X_after_over,
                columns=M2_FEATURES,
            )
        y_after_over = pd.Series(y_after_over).astype(int)

        after_over_events = int(y_after_over.sum())
        after_over_nonevents = int((y_after_over == 0).sum())
        after_over_ratio = min(
            after_over_events,
            after_over_nonevents,
        ) / max(after_over_events, after_over_nonevents)

        if after_over_ratio >= UNDER_SAMPLING_STRATEGY:
            X_resampled = X_after_over.copy()
            y_resampled = y_after_over.copy()
            under_sampler_applied = False
        else:
            under_sampler = make_under_sampler(fold_random_state)
            X_resampled, y_resampled = under_sampler.fit_resample(
                X_after_over,
                y_after_over,
            )
            under_sampler_applied = True

        if not isinstance(X_resampled, pd.DataFrame):
            X_resampled = pd.DataFrame(
                X_resampled,
                columns=M2_FEATURES,
            )
        y_resampled = pd.Series(y_resampled).astype(int)

        category_valid, category_rows = validate_resampled_categories(
            X_original=X_fold_train,
            X_resampled=X_resampled,
            outcome=outcome,
            fold=fold_number,
        )
        category_validation_rows.extend(category_rows)

        if not category_valid:
            raise RuntimeError(
                f"{outcome} fold {fold_number}: invalid synthetic category."
            )

        estimator = make_adaboost()
        estimator.fit(X_resampled, y_resampled)
        fold_probability = positive_probability(
            estimator,
            X_fold_validation,
        )

        oof_probability[validation_position] = fold_probability
        oof_fold[validation_position] = fold_number
        validation_counter[validation_position] += 1

        after_events = int(y_resampled.sum())
        after_nonevents = int((y_resampled == 0).sum())

        fold_manifest_rows.append({
            "Outcome": outcome,
            "Fold": fold_number,
            "Training_N": len(train_position),
            "Training_Events": int(y_fold_train.sum()),
            "Validation_N": len(validation_position),
            "Validation_Events": int(y_fold_validation.sum()),
            "Training_Validation_Position_Overlap": bool(
                set(train_position).intersection(validation_position)
            ),
            "Validation_Probability_Min": float(fold_probability.min()),
            "Validation_Probability_Max": float(fold_probability.max()),
        })

        resampling_rows.append({
            "Outcome": outcome,
            "Fold": fold_number,
            "Sampler": "SMOTENC followed by RandomUnderSampler",
            "SMOTENC_Applied": over_sampler_applied,
            "RandomUnderSampler_Applied": under_sampler_applied,
            "Before_N": len(y_fold_train),
            "Before_Events": before_events,
            "Before_Nonevents": before_nonevents,
            "Before_Minority_Majority_Ratio": before_ratio,
            "After_SMOTENC_N": len(y_after_over),
            "After_SMOTENC_Events": after_over_events,
            "After_SMOTENC_Nonevents": after_over_nonevents,
            "After_SMOTENC_Minority_Majority_Ratio": after_over_ratio,
            "Final_N": len(y_resampled),
            "Final_Events": after_events,
            "Final_Nonevents": after_nonevents,
            "Final_Minority_Majority_Ratio": (
                min(after_events, after_nonevents)
                / max(after_events, after_nonevents)
            ),
            "Validation_Resampled": False,
            "Categories_Valid": category_valid,
        })

    if np.isnan(oof_probability).any():
        raise RuntimeError(f"{outcome}: missing OOF probabilities.")
    if not np.all(validation_counter == 1):
        raise RuntimeError(
            f"{outcome}: each participant was not predicted once."
        )
    if set(np.unique(oof_fold)) != {1, 2, 3, 4, 5}:
        raise RuntimeError(f"{outcome}: incomplete OOF fold assignment.")

    (
        locked_threshold,
        selected_threshold_row,
        threshold_table,
        threshold_rule,
        target_achieved,
        eligible_threshold_count,
    ) = search_and_select_threshold(
        y_true=y_development.to_numpy(dtype=int),
        y_prob=oof_probability,
        target_sensitivity=TARGET_SENSITIVITY,
    )

    oof_metrics = calculate_binary_metrics(
        y_development,
        oof_probability,
        locked_threshold,
    )

    secondary_oof_predictions[outcome] = {
        "y_true": y_development.to_numpy(dtype=int),
        "y_prob": oof_probability.copy(),
        "y_pred": oof_metrics["y_pred"].copy(),
        "fold_id": oof_fold.copy(),
        "validation_counter": validation_counter.copy(),
        "source_row_position": development_source_positions.copy(),
        "locked_threshold": locked_threshold,
        "n": n,
        "events": events,
        "prevalence": float(events / n),
    }

    secondary_fold_assignments[outcome] = oof_fold.copy()
    secondary_threshold_tables[outcome] = threshold_table.copy()
    secondary_locked_thresholds[outcome] = locked_threshold

    # Fit the final development-only pipeline. Samplers are used only
    # during fit; predict_proba applies the estimator to untouched data.
    final_random_state = (
        BASE_RESAMPLING_RANDOM_STATE + outcome_index * 100
    )
    final_pipeline = make_final_pipeline(final_random_state)
    final_pipeline.fit(
        X_secondary_development,
        y_development,
    )
    secondary_fitted_pipelines[outcome] = final_pipeline

    oof_result_rows.append({
        "Outcome": outcome,
        "Screening_Model": SECONDARY_MODEL_NAME,
        "Algorithm": SECONDARY_ALGORITHM_NAME,
        "Development_N": n,
        "Development_Events": events,
        "Development_Prevalence": float(events / n),
        "CV_Folds": N_SPLITS,
        "CV_Random_State": CV_RANDOM_STATE,
        "Sampler": "SMOTENC followed by RandomUnderSampler",
        "SMOTENC_Sampling_Strategy": SMOTENC_SAMPLING_STRATEGY,
        "SMOTENC_K_Neighbors": SMOTENC_K_NEIGHBORS,
        "Under_Sampling_Strategy": UNDER_SAMPLING_STRATEGY,
        "Locked_Threshold": locked_threshold,
        "Threshold_Target_Sensitivity": TARGET_SENSITIVITY,
        "Target_Sensitivity_Achieved": target_achieved,
        "Eligible_Threshold_Count": eligible_threshold_count,
        "Threshold_Rule": threshold_rule,
        **{
            f"OOF_{key}": value
            for key, value in oof_metrics.items()
            if key != "y_pred"
        },
    })


# ============================================================
# 10. AGGREGATE RESULT TABLES
# ============================================================

partS2_oof_results = pd.DataFrame(oof_result_rows)
partS2_fold_manifest = pd.DataFrame(fold_manifest_rows)
partS2_resampling_audit = pd.DataFrame(resampling_rows)
partS2_category_validation = pd.DataFrame(
    category_validation_rows
)

partS2_threshold_summary = partS2_oof_results.loc[
    :,
    [
        "Outcome",
        "Screening_Model",
        "Algorithm",
        "Development_N",
        "Development_Events",
        "Locked_Threshold",
        "Threshold_Target_Sensitivity",
        "Target_Sensitivity_Achieved",
        "Threshold_Rule",
        "OOF_AUC",
        "OOF_Average_Precision",
        "OOF_Brier_Score",
        "OOF_Sensitivity",
        "OOF_Specificity",
        "OOF_Precision",
        "OOF_F1",
        "OOF_TP",
        "OOF_FP",
        "OOF_FN",
        "OOF_TN",
    ],
].copy()


# ============================================================
# 11. METHODOLOGICAL VERIFICATION
# ============================================================

verification_rows = []

for outcome in SECONDARY_TARGETS:
    record = secondary_oof_predictions[outcome]
    outcome_fold_manifest = partS2_fold_manifest.loc[
        partS2_fold_manifest["Outcome"] == outcome
    ]
    outcome_resampling = partS2_resampling_audit.loc[
        partS2_resampling_audit["Outcome"] == outcome
    ]
    outcome_category_validation = partS2_category_validation.loc[
        partS2_category_validation["Outcome"] == outcome
    ]

    checks = {
        "Development_N_Correct": record["n"] == EXPECTED_DEVELOPMENT_N,
        "Exactly_One_OOF_Prediction_Per_Participant": bool(
            np.all(record["validation_counter"] == 1)
        ),
        "All_5_Folds_Present": set(record["fold_id"]) == {1, 2, 3, 4, 5},
        "No_Missing_OOF_Probabilities": bool(
            np.isfinite(record["y_prob"]).all()
        ),
        "All_Probabilities_In_Unit_Interval": bool(
            (record["y_prob"] >= 0).all()
            and (record["y_prob"] <= 1).all()
        ),
        "No_Training_Validation_Overlap": bool(
            not outcome_fold_manifest[
                "Training_Validation_Position_Overlap"
            ].any()
        ),
        "Validation_Folds_Not_Resampled": bool(
            not outcome_resampling["Validation_Resampled"].any()
        ),
        "SMOTENC_Configuration_Correct": bool(
            (
                partS2_oof_results.loc[
                    partS2_oof_results["Outcome"] == outcome,
                    "SMOTENC_Sampling_Strategy",
                ] == SMOTENC_SAMPLING_STRATEGY
            ).all()
        ),
        "All_Resampled_Categories_Valid": bool(
            outcome_category_validation["Category_Valid"].all()
        ),
        "One_Locked_Threshold": bool(
            outcome in secondary_locked_thresholds
            and np.isfinite(secondary_locked_thresholds[outcome])
        ),
        "Final_Development_Pipeline_Fitted": bool(
            outcome in secondary_fitted_pipelines
        ),
        "Heldout_Excluded_From_Development": True,
        "Temporal_Excluded_From_Development": True,
    }

    verification_rows.append({
        "Outcome": outcome,
        **checks,
        "Outcome_Ready": bool(all(checks.values())),
    })

partS2_verification = pd.DataFrame(verification_rows)

PARTS2_READY = bool(
    len(partS2_verification) == len(SECONDARY_TARGETS)
    and partS2_verification["Outcome_Ready"].all()
)

partS2_global_verification = pd.DataFrame([{
    "Predictor_Strategy": SECONDARY_MODEL_NAME,
    "Algorithm": SECONDARY_ALGORITHM_NAME,
    "Secondary_Outcomes": len(SECONDARY_TARGETS),
    "Development_N": len(X_secondary_development),
    "OOF_Folds": N_SPLITS,
    "Sampler": "SMOTENC followed by RandomUnderSampler",
    "Continuous_Feature_Count": len(CONTINUOUS_FEATURES),
    "Categorical_or_Ordinal_Feature_Count": len(
        CATEGORICAL_OR_ORDINAL_FEATURES
    ),
    "All_Outcomes_Ready": bool(
        partS2_verification["Outcome_Ready"].all()
    ),
    "Heldout_Evaluated": False,
    "Temporal_Evaluated": False,
    "Participant_Level_OOF_Exported": False,
    "PARTS2_Ready": PARTS2_READY,
}])


# ============================================================
# 12. DISPLAY AGGREGATE RESULTS
# ============================================================

print("=" * 110)
print("PART S2 VERSION 2.1")
print("OUTCOME-SPECIFIC SMOTENC-ADABOOST OOF DEVELOPMENT")
print("=" * 110)

print("\nOOF development results:")
display(partS2_oof_results)

print("\nLocked thresholds:")
display(partS2_threshold_summary)

print("\nFold-level resampling audit:")
display(partS2_resampling_audit)

print("\nCategorical-value validity audit:")
display(partS2_category_validation)

print("\nMethodological verification:")
display(partS2_verification)

print("\nGlobal verification:")
display(partS2_global_verification)


# ============================================================
# 13. EXPORT AGGREGATE METADATA ONLY
# ============================================================

partS2_oof_results.to_csv(
    "partS2_oof_development_results.csv",
    index=False,
    encoding="utf-8-sig",
)

partS2_threshold_summary.to_csv(
    "partS2_locked_thresholds.csv",
    index=False,
    encoding="utf-8-sig",
)

partS2_fold_manifest.to_csv(
    "partS2_fold_manifest_aggregate.csv",
    index=False,
    encoding="utf-8-sig",
)

partS2_resampling_audit.to_csv(
    "partS2_resampling_audit.csv",
    index=False,
    encoding="utf-8-sig",
)

partS2_category_validation.to_csv(
    "partS2_category_validation.csv",
    index=False,
    encoding="utf-8-sig",
)

partS2_verification.to_csv(
    "partS2_verification.csv",
    index=False,
    encoding="utf-8-sig",
)

partS2_global_verification.to_csv(
    "partS2_global_verification.csv",
    index=False,
    encoding="utf-8-sig",
)


# ============================================================
# 14. FINAL SAFETY GATE AND SUMMARY
# ============================================================

if not PARTS2_READY:
    failed_outcomes = partS2_verification.loc[
        ~partS2_verification["Outcome_Ready"],
        "Outcome",
    ].tolist()
    raise RuntimeError(
        "PART S2 verification failed for: "
        + ", ".join(failed_outcomes)
    )

print("\n" + "=" * 110)
print("PART S2 VERSION 2.1 COMPLETED")
print("=" * 110)
print(f"Predictor strategy       : {SECONDARY_MODEL_NAME}")
print(f"Algorithm                : {SECONDARY_ALGORITHM_NAME}")
print(f"Development cohort       : N={len(X_secondary_development):,}")
print(f"Secondary outcomes       : {len(SECONDARY_TARGETS)}")
print(f"OOF folds                : {N_SPLITS}")
print(f"OOF random state         : {CV_RANDOM_STATE}")
print("Continuous predictors    : GPAX only")
print(f"Categorical/ordinal       : {len(CATEGORICAL_OR_ORDINAL_FEATURES)}")
print("Over-sampler             : SMOTENC")
print(f"SMOTENC target ratio     : {SMOTENC_SAMPLING_STRATEGY:.2f}")
print(f"SMOTENC k_neighbors      : {SMOTENC_K_NEIGHBORS}")
print("Under-sampler            : RandomUnderSampler")
print(f"Final target ratio       : {UNDER_SAMPLING_STRATEGY:.2f}")
print(f"Threshold sensitivity    : {TARGET_SENSITIVITY:.2f}")
print("Held-out evaluated       : No")
print("Temporal evaluated       : No")
print("Participant OOF exported : No")
print("\nPARTS2_READY:", PARTS2_READY)
print("\nProceed to PART S3 only after PARTS2_READY=True.")
```

## PART S3: Held-out 2024 Validation + Stratified Bootstrap 95% CI

```python
# ============================================================
# PART S3 VERSION 2.0
# HELD-OUT 2024 VALIDATION AND STRATIFIED BOOTSTRAP 95% CI
# ============================================================
#
# Purpose:
# - Evaluate the two outcome-specific M2 AdaBoost pipelines in the
#   untouched held-out 2024 cohort.
# - Apply the outcome-specific thresholds locked in PART S2.
# - Quantify uncertainty using fixed-prediction stratified
#   participant-level percentile bootstrap confidence intervals.
#
# Safeguards:
# - No model is fitted or refitted.
# - Held-out observations are not resampled for model fitting.
# - No prediction threshold is selected or modified.
# - No recalibration is performed.
# - Participant-level predictions and bootstrap distributions remain
#   in memory and are not exported.
# ============================================================

import numpy as np
import pandas as pd
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
)


# ============================================================
# 1. SETTINGS
# ============================================================

N_BOOTSTRAPS = 2000
MINIMUM_VALID_BOOTSTRAPS = 1800
CI_LEVEL = 0.95
LOWER_PERCENTILE = 2.5
UPPER_PERCENTILE = 97.5
BASE_BOOTSTRAP_RANDOM_STATE = 31000
THRESHOLD_ATOL = 1e-12

EXPECTED_HELDOUT_N = 1227
EXPECTED_HELDOUT_MDD_EVENTS = 14

BOOTSTRAP_METRICS = [
    "AUC",
    "Average_Precision",
    "Brier_Score",
    "Sensitivity",
    "Specificity",
    "Precision",
    "F1",
]


# ============================================================
# 2. VERIFY PART S2
# ============================================================

required_objects = [
    "PARTS2_READY",
    "partS2_verification",
    "secondary_fitted_pipelines",
    "secondary_locked_thresholds",
    "secondary_heldout_2024",
    "M2_FEATURES",
    "SECONDARY_TARGETS",
    "SECONDARY_MODEL_NAME",
    "SECONDARY_ALGORITHM_NAME",
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required PART S2 objects: "
        + ", ".join(missing_objects)
    )

if not bool(PARTS2_READY):
    raise RuntimeError("PART S2 is not ready.")

if not partS2_verification["Outcome_Ready"].all():
    raise RuntimeError("One or more PART S2 outcomes are not ready.")

if set(secondary_fitted_pipelines) != set(SECONDARY_TARGETS):
    raise RuntimeError("The fitted secondary pipelines are incomplete.")

if set(secondary_locked_thresholds) != set(SECONDARY_TARGETS):
    raise RuntimeError("The locked secondary thresholds are incomplete.")


# ============================================================
# 3. PREPARE AND VERIFY HELD-OUT INPUTS
# ============================================================

required_heldout_columns = (
    M2_FEATURES
    + SECONDARY_TARGETS
    + ["mdddx", "Source_Row_Position"]
)

missing_heldout_columns = [
    column for column in required_heldout_columns
    if column not in secondary_heldout_2024.columns
]

if missing_heldout_columns:
    raise KeyError(
        "Held-out cohort is missing: "
        + ", ".join(missing_heldout_columns)
    )

X_heldout_secondary = (
    secondary_heldout_2024[M2_FEATURES]
    .copy()
    .reset_index(drop=True)
)

heldout_source_positions = (
    secondary_heldout_2024["Source_Row_Position"]
    .to_numpy(dtype=int)
)

if len(X_heldout_secondary) != EXPECTED_HELDOUT_N:
    raise RuntimeError("Unexpected held-out cohort size.")

if X_heldout_secondary.columns.tolist() != M2_FEATURES:
    raise RuntimeError("Held-out M2 feature order is incorrect.")

if X_heldout_secondary.isna().any().any():
    raise RuntimeError("Held-out predictors contain missing values.")

for feature in M2_FEATURES:
    converted = pd.to_numeric(
        X_heldout_secondary[feature],
        errors="raise",
    )
    if not np.isfinite(converted.to_numpy(dtype=float)).all():
        raise RuntimeError(f"Held-out {feature} contains nonfinite values.")
    X_heldout_secondary[feature] = converted

heldout_mdd = pd.to_numeric(
    secondary_heldout_2024["mdddx"],
    errors="raise",
).astype(int)

if int(heldout_mdd.sum()) != EXPECTED_HELDOUT_MDD_EVENTS:
    raise RuntimeError("Unexpected held-out MDD event count.")


# ============================================================
# 4. PROBABILITY AND METRIC HELPERS
# ============================================================

def positive_probability(estimator, X):
    matrix = np.asarray(estimator.predict_proba(X), dtype=float)
    classes = np.asarray(estimator.classes_)
    positive_positions = np.flatnonzero(classes == 1)

    if matrix.ndim != 2 or len(positive_positions) != 1:
        raise ValueError("Unable to identify class-1 probability.")

    probability = matrix[:, int(positive_positions[0])]

    if (
        not np.isfinite(probability).all()
        or (probability < 0).any()
        or (probability > 1).any()
    ):
        raise ValueError("Invalid held-out probabilities.")

    return probability


def calculate_fixed_metrics(y_true, y_prob, threshold):
    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)
    y_pred = (y_prob >= float(threshold)).astype(int)

    if len(y_true) != len(y_prob):
        raise ValueError("Outcome and probability lengths differ.")
    if set(np.unique(y_true)) != {0, 1}:
        raise ValueError("Both outcome classes are required.")

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1],
    ).ravel()

    sensitivity = recall_score(y_true, y_pred, zero_division=0)
    specificity = tn / (tn + fp) if (tn + fp) else np.nan
    precision = precision_score(y_true, y_pred, zero_division=0)

    return {
        "AUC": float(roc_auc_score(y_true, y_prob)),
        "Average_Precision": float(
            average_precision_score(y_true, y_prob)
        ),
        "Brier_Score": float(brier_score_loss(y_true, y_prob)),
        "Sensitivity": float(sensitivity),
        "Specificity": float(specificity),
        "Precision": float(precision),
        "F1": float(f1_score(y_true, y_pred, zero_division=0)),
        "Predicted_Positive_N": int(y_pred.sum()),
        "Predicted_Positive_Rate": float(y_pred.mean()),
        "TP": int(tp),
        "FP": int(fp),
        "FN": int(fn),
        "TN": int(tn),
        "y_pred": y_pred,
    }


# ============================================================
# 5. STRATIFIED FIXED-PREDICTION BOOTSTRAP
# ============================================================

def stratified_bootstrap(
    y_true,
    y_prob,
    threshold,
    n_bootstraps,
    random_state,
):
    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)

    positive_positions = np.flatnonzero(y_true == 1)
    negative_positions = np.flatnonzero(y_true == 0)

    if len(positive_positions) == 0 or len(negative_positions) == 0:
        raise ValueError("Both classes are required for bootstrap.")

    rng = np.random.default_rng(random_state)
    distributions = {
        metric: np.full(n_bootstraps, np.nan, dtype=float)
        for metric in BOOTSTRAP_METRICS
    }

    for bootstrap_index in range(n_bootstraps):
        sampled_positive = rng.choice(
            positive_positions,
            size=len(positive_positions),
            replace=True,
        )
        sampled_negative = rng.choice(
            negative_positions,
            size=len(negative_positions),
            replace=True,
        )
        sampled_positions = np.concatenate([
            sampled_positive,
            sampled_negative,
        ])
        rng.shuffle(sampled_positions)

        try:
            replicate_metrics = calculate_fixed_metrics(
                y_true[sampled_positions],
                y_prob[sampled_positions],
                threshold,
            )
            for metric in BOOTSTRAP_METRICS:
                distributions[metric][bootstrap_index] = (
                    replicate_metrics[metric]
                )
        except Exception:
            continue

    return distributions


# ============================================================
# 6. FIXED HELD-OUT PREDICTIONS AND BOOTSTRAP ANALYSES
# ============================================================

secondary_heldout_predictions = {}
secondary_heldout_bootstrap_distributions = {}

heldout_result_rows = []
bootstrap_result_rows = []
bootstrap_status_rows = []
input_audit_rows = []

for outcome_index, outcome in enumerate(SECONDARY_TARGETS):
    y_heldout = (
        secondary_heldout_2024[outcome]
        .astype(int)
        .copy()
        .reset_index(drop=True)
    )

    if set(y_heldout.unique()) != {0, 1}:
        raise RuntimeError(f"{outcome}: held-out outcome is not binary.")

    fitted_pipeline = secondary_fitted_pipelines[outcome]
    locked_threshold = float(secondary_locked_thresholds[outcome])

    if not hasattr(fitted_pipeline, "named_steps"):
        raise TypeError(f"{outcome}: fitted object is not a pipeline.")

    expected_steps = ["over_sampler", "under_sampler", "estimator"]
    if list(fitted_pipeline.named_steps) != expected_steps:
        raise RuntimeError(f"{outcome}: fitted pipeline step order changed.")

    if type(
        fitted_pipeline.named_steps["over_sampler"]
    ).__name__ != "SMOTENC":
        raise RuntimeError(f"{outcome}: over-sampler is not SMOTENC.")

    if type(
        fitted_pipeline.named_steps["under_sampler"]
    ).__name__ != "RandomUnderSampler":
        raise RuntimeError(
            f"{outcome}: under-sampler is not RandomUnderSampler."
        )

    y_probability = positive_probability(
        fitted_pipeline,
        X_heldout_secondary,
    )

    point_metrics = calculate_fixed_metrics(
        y_heldout,
        y_probability,
        locked_threshold,
    )

    secondary_heldout_predictions[outcome] = {
        "validation_set": "Held-out 2024",
        "y_true": y_heldout.to_numpy(dtype=int),
        "y_prob": y_probability.copy(),
        "y_pred": point_metrics["y_pred"].copy(),
        "source_row_position": heldout_source_positions.copy(),
        "locked_threshold": locked_threshold,
        "n": len(y_heldout),
        "events": int(y_heldout.sum()),
        "prevalence": float(y_heldout.mean()),
        "model_refitted": False,
        "heldout_resampled_for_model_fitting": False,
        "threshold_modified": False,
    }

    random_state = BASE_BOOTSTRAP_RANDOM_STATE + outcome_index * 100
    distributions = stratified_bootstrap(
        y_true=y_heldout,
        y_prob=y_probability,
        threshold=locked_threshold,
        n_bootstraps=N_BOOTSTRAPS,
        random_state=random_state,
    )

    secondary_heldout_bootstrap_distributions[outcome] = distributions

    heldout_result_rows.append({
        "Outcome": outcome,
        "Screening_Model": SECONDARY_MODEL_NAME,
        "Algorithm": SECONDARY_ALGORITHM_NAME,
        "Validation_Set": "Held-out 2024",
        "N": len(y_heldout),
        "Events": int(y_heldout.sum()),
        "Prevalence": float(y_heldout.mean()),
        "Locked_Threshold": locked_threshold,
        "Threshold_Source": "PART S2 development OOF only",
        **{
            key: value
            for key, value in point_metrics.items()
            if key != "y_pred"
        },
    })

    minimum_valid = N_BOOTSTRAPS

    for metric in BOOTSTRAP_METRICS:
        valid_values = distributions[metric][
            np.isfinite(distributions[metric])
        ]
        valid_n = int(len(valid_values))
        minimum_valid = min(minimum_valid, valid_n)
        ci_computed = valid_n >= MINIMUM_VALID_BOOTSTRAPS

        if ci_computed:
            lower_ci = float(
                np.percentile(valid_values, LOWER_PERCENTILE)
            )
            upper_ci = float(
                np.percentile(valid_values, UPPER_PERCENTILE)
            )
        else:
            lower_ci = np.nan
            upper_ci = np.nan

        bootstrap_result_rows.append({
            "Outcome": outcome,
            "Screening_Model": SECONDARY_MODEL_NAME,
            "Algorithm": SECONDARY_ALGORITHM_NAME,
            "Validation_Set": "Held-out 2024",
            "N": len(y_heldout),
            "Events": int(y_heldout.sum()),
            "Locked_Threshold": locked_threshold,
            "Metric": metric,
            "Estimate": float(point_metrics[metric]),
            "CI_Lower": lower_ci,
            "CI_Upper": upper_ci,
            "CI_Level": CI_LEVEL,
            "CI_Method": (
                "Stratified percentile bootstrap "
                "(2.5th-97.5th percentiles)"
            ),
            "Requested_Bootstrap_Replicates": N_BOOTSTRAPS,
            "Valid_Bootstrap_Replicates": valid_n,
            "CI_Computed": ci_computed,
            "Random_State": random_state,
        })

    bootstrap_status_rows.append({
        "Outcome": outcome,
        "Validation_Set": "Held-out 2024",
        "Requested_Bootstrap_Replicates": N_BOOTSTRAPS,
        "Minimum_Valid_Replicates_Across_Metrics": minimum_valid,
        "Minimum_Required_Replicates": MINIMUM_VALID_BOOTSTRAPS,
        "All_CIs_Computed": minimum_valid >= MINIMUM_VALID_BOOTSTRAPS,
        "Status": (
            "Success"
            if minimum_valid >= MINIMUM_VALID_BOOTSTRAPS
            else "Insufficient valid replicates"
        ),
    })

    input_checks = {
        "N_Correct": len(y_heldout) == EXPECTED_HELDOUT_N,
        "Outcome_Binary": set(y_heldout.unique()) == {0, 1},
        "No_Outcome_Missing": not y_heldout.isna().any(),
        "No_Predictor_Missing": not X_heldout_secondary.isna().any().any(),
        "Probability_Length_Correct": len(y_probability) == len(y_heldout),
        "Probabilities_Valid": bool(
            np.isfinite(y_probability).all()
            and (y_probability >= 0).all()
            and (y_probability <= 1).all()
        ),
        "Locked_Threshold_Finite": np.isfinite(locked_threshold),
        "Predictions_Match_Locked_Threshold": np.array_equal(
            point_metrics["y_pred"],
            (y_probability >= locked_threshold).astype(int),
        ),
        "Model_Not_Refitted": True,
        "Heldout_Not_Resampled_For_Model_Fitting": True,
        "Threshold_Not_Modified": True,
    }

    input_audit_rows.append({
        "Outcome": outcome,
        **input_checks,
        "Input_Ready": bool(all(input_checks.values())),
    })


# ============================================================
# 7. AGGREGATE TABLES
# ============================================================

partS3_heldout_results = pd.DataFrame(heldout_result_rows)
partS3_bootstrap_results = pd.DataFrame(bootstrap_result_rows)
partS3_bootstrap_status = pd.DataFrame(bootstrap_status_rows)
partS3_input_audit = pd.DataFrame(input_audit_rows)


def format_ci(row):
    return (
        f"{float(row['Estimate']):.3f} "
        f"({float(row['CI_Lower']):.3f}-"
        f"{float(row['CI_Upper']):.3f})"
    )


publication_rows = []

for outcome in SECONDARY_TARGETS:
    point_row = partS3_heldout_results.loc[
        partS3_heldout_results["Outcome"] == outcome
    ].iloc[0]

    publication_row = {
        "Outcome": outcome,
        "Screening_Model": SECONDARY_MODEL_NAME,
        "Algorithm": SECONDARY_ALGORITHM_NAME,
        "Validation_Set": "Held-out 2024",
        "N": int(point_row["N"]),
        "Events": int(point_row["Events"]),
        "Locked_Threshold": float(point_row["Locked_Threshold"]),
    }

    for metric in BOOTSTRAP_METRICS:
        ci_row = partS3_bootstrap_results.loc[
            (partS3_bootstrap_results["Outcome"] == outcome)
            & (partS3_bootstrap_results["Metric"] == metric)
        ]
        if len(ci_row) != 1:
            raise RuntimeError(
                f"{outcome}: expected one bootstrap row for {metric}."
            )
        publication_row[f"{metric}_95CI"] = format_ci(ci_row.iloc[0])

    publication_row.update({
        "TP": int(point_row["TP"]),
        "FP": int(point_row["FP"]),
        "FN": int(point_row["FN"]),
        "TN": int(point_row["TN"]),
    })
    publication_rows.append(publication_row)

partS3_publication_table = pd.DataFrame(publication_rows)


# ============================================================
# 8. GLOBAL VERIFICATION
# ============================================================

all_inputs_ready = bool(
    len(partS3_input_audit) == len(SECONDARY_TARGETS)
    and partS3_input_audit["Input_Ready"].all()
)

all_bootstraps_ready = bool(
    len(partS3_bootstrap_status) == len(SECONDARY_TARGETS)
    and partS3_bootstrap_status["All_CIs_Computed"].all()
)

all_ci_rows_ready = bool(
    len(partS3_bootstrap_results)
    == len(SECONDARY_TARGETS) * len(BOOTSTRAP_METRICS)
    and partS3_bootstrap_results["CI_Computed"].all()
)

PARTS3_READY = bool(
    all_inputs_ready
    and all_bootstraps_ready
    and all_ci_rows_ready
    and len(partS3_publication_table) == len(SECONDARY_TARGETS)
)

partS3_verification = pd.DataFrame([{
    "Predictor_Strategy": SECONDARY_MODEL_NAME,
    "Algorithm": SECONDARY_ALGORITHM_NAME,
    "Validation_Set": "Held-out 2024",
    "Outcomes_Evaluated": len(SECONDARY_TARGETS),
    "Heldout_N": len(X_heldout_secondary),
    "Heldout_MDD_Events": int(heldout_mdd.sum()),
    "All_Fixed_Inputs_Ready": all_inputs_ready,
    "All_Bootstrap_Analyses_Ready": all_bootstraps_ready,
    "All_Confidence_Intervals_Computed": all_ci_rows_ready,
    "Model_Refitted": False,
    "Heldout_Resampled_For_Model_Fitting": False,
    "Threshold_Modified": False,
    "Recalibration_Performed": False,
    "Participant_Level_Predictions_Exported": False,
    "Bootstrap_Distributions_Exported": False,
    "PARTS3_Ready": PARTS3_READY,
}])


# ============================================================
# 9. DISPLAY AGGREGATE RESULTS
# ============================================================

print("=" * 110)
print("PART S3 VERSION 2.0")
print("HELD-OUT 2024 VALIDATION AND BOOTSTRAP 95% CI")
print("=" * 110)

print("\nHeld-out performance:")
display(partS3_heldout_results)

print("\nBootstrap status:")
display(partS3_bootstrap_status)

print("\nPublication table:")
display(partS3_publication_table)

print("\nInput audit:")
display(partS3_input_audit)

print("\nGlobal verification:")
display(partS3_verification)


# ============================================================
# 10. EXPORT AGGREGATE RESULTS ONLY
# ============================================================

partS3_heldout_results.to_csv(
    "partS3_heldout_2024_results.csv",
    index=False,
    encoding="utf-8-sig",
)

partS3_bootstrap_results.to_csv(
    "partS3_heldout_bootstrap_ci_results.csv",
    index=False,
    encoding="utf-8-sig",
)

partS3_bootstrap_status.to_csv(
    "partS3_heldout_bootstrap_status.csv",
    index=False,
    encoding="utf-8-sig",
)

partS3_publication_table.to_csv(
    "partS3_heldout_2024_publication_table.csv",
    index=False,
    encoding="utf-8-sig",
)

partS3_input_audit.to_csv(
    "partS3_heldout_input_audit.csv",
    index=False,
    encoding="utf-8-sig",
)

partS3_verification.to_csv(
    "partS3_verification.csv",
    index=False,
    encoding="utf-8-sig",
)


# ============================================================
# 11. FINAL SAFETY GATE AND SUMMARY
# ============================================================

if not PARTS3_READY:
    raise RuntimeError("PART S3 methodological verification failed.")

print("\n" + "=" * 110)
print("PART S3 VERSION 2.0 COMPLETED")
print("=" * 110)
print(f"Predictor strategy        : {SECONDARY_MODEL_NAME}")
print(f"Algorithm                 : {SECONDARY_ALGORITHM_NAME}")
print(f"Held-out cohort           : N={len(X_heldout_secondary):,}")
print(f"Secondary outcomes        : {len(SECONDARY_TARGETS)}")
print(f"Bootstrap replicates      : {N_BOOTSTRAPS:,} per outcome")
print(f"Minimum valid replicates  : {MINIMUM_VALID_BOOTSTRAPS:,}")
print("CI method                 : Stratified percentile bootstrap")
print("CI limits                 : 2.5th-97.5th percentiles")
print("Model refitted            : No")
print("Held-out model resampling : No")
print("Threshold modified        : No")
print("Participant rows exported : No")
print("Bootstrap distributions   : Not exported")
print("\nPARTS3_READY:", PARTS3_READY)
print("\nProceed to PART S4 only after PARTS3_READY=True.")
```

## PART S4: Temporal 2025 Validation and Bootstrap 95% Confidence Intervals

```python
# ============================================================
# PART S4 VERSION 2.0
# TEMPORAL 2025 VALIDATION AND STRATIFIED BOOTSTRAP 95% CI
# ============================================================
#
# Purpose:
# - Evaluate the two outcome-specific M2 AdaBoost pipelines in the
#   untouched temporal 2025 cohort.
# - Apply the outcome-specific thresholds locked in PART S2.
# - Quantify uncertainty using fixed-prediction stratified
#   participant-level percentile bootstrap confidence intervals.
#
# Safeguards:
# - No model is fitted or refitted.
# - Temporal observations are not resampled for model fitting.
# - No prediction threshold is selected or modified.
# - No recalibration is performed.
# - Participant-level predictions and bootstrap distributions remain
#   in memory and are not exported.
# ============================================================

import numpy as np
import pandas as pd
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
)


# ============================================================
# 1. SETTINGS
# ============================================================

N_BOOTSTRAPS = 2000
MINIMUM_VALID_BOOTSTRAPS = 1800
CI_LEVEL = 0.95
LOWER_PERCENTILE = 2.5
UPPER_PERCENTILE = 97.5
BASE_BOOTSTRAP_RANDOM_STATE = 41000
THRESHOLD_ATOL = 1e-12

EXPECTED_TEMPORAL_N = 815
EXPECTED_TEMPORAL_MDD_EVENTS = 16

BOOTSTRAP_METRICS = [
    "AUC",
    "Average_Precision",
    "Brier_Score",
    "Sensitivity",
    "Specificity",
    "Precision",
    "F1",
]


# ============================================================
# 2. VERIFY PRECEDING DEVELOPMENT AND HELD-OUT ANALYSES
# ============================================================

required_objects = [
    "PARTS2_READY",
    "partS2_verification",
    "PARTS3_READY",
    "secondary_fitted_pipelines",
    "secondary_locked_thresholds",
    "secondary_temporal_2025",
    "M2_FEATURES",
    "SECONDARY_TARGETS",
    "SECONDARY_MODEL_NAME",
    "SECONDARY_ALGORITHM_NAME",
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required PART S2 objects: "
        + ", ".join(missing_objects)
    )

if not bool(PARTS2_READY):
    raise RuntimeError("PART S2 is not ready.")

if not bool(PARTS3_READY):
    raise RuntimeError("PART S3 is not ready.")

if not partS2_verification["Outcome_Ready"].all():
    raise RuntimeError("One or more PART S2 outcomes are not ready.")

if set(secondary_fitted_pipelines) != set(SECONDARY_TARGETS):
    raise RuntimeError("The fitted secondary pipelines are incomplete.")

if set(secondary_locked_thresholds) != set(SECONDARY_TARGETS):
    raise RuntimeError("The locked secondary thresholds are incomplete.")


# ============================================================
# 3. PREPARE AND VERIFY TEMPORAL INPUTS
# ============================================================

required_temporal_columns = (
    M2_FEATURES
    + SECONDARY_TARGETS
    + ["mdddx", "Temporal_Row_Position"]
)

missing_temporal_columns = [
    column for column in required_temporal_columns
    if column not in secondary_temporal_2025.columns
]

if missing_temporal_columns:
    raise KeyError(
        "Temporal cohort is missing: "
        + ", ".join(missing_temporal_columns)
    )

X_temporal_secondary = (
    secondary_temporal_2025[M2_FEATURES]
    .copy()
    .reset_index(drop=True)
)

temporal_source_positions = (
    secondary_temporal_2025["Temporal_Row_Position"]
    .to_numpy(dtype=int)
)

if len(X_temporal_secondary) != EXPECTED_TEMPORAL_N:
    raise RuntimeError("Unexpected temporal cohort size.")

if X_temporal_secondary.columns.tolist() != M2_FEATURES:
    raise RuntimeError("Temporal M2 feature order is incorrect.")

if X_temporal_secondary.isna().any().any():
    raise RuntimeError("Temporal predictors contain missing values.")

for feature in M2_FEATURES:
    converted = pd.to_numeric(
        X_temporal_secondary[feature],
        errors="raise",
    )
    if not np.isfinite(converted.to_numpy(dtype=float)).all():
        raise RuntimeError(f"Temporal {feature} contains nonfinite values.")
    X_temporal_secondary[feature] = converted

temporal_mdd = pd.to_numeric(
    secondary_temporal_2025["mdddx"],
    errors="raise",
).astype(int)

if int(temporal_mdd.sum()) != EXPECTED_TEMPORAL_MDD_EVENTS:
    raise RuntimeError("Unexpected temporal MDD event count.")


# ============================================================
# 4. PROBABILITY AND METRIC HELPERS
# ============================================================

def positive_probability(estimator, X):
    matrix = np.asarray(estimator.predict_proba(X), dtype=float)
    classes = np.asarray(estimator.classes_)
    positive_positions = np.flatnonzero(classes == 1)

    if matrix.ndim != 2 or len(positive_positions) != 1:
        raise ValueError("Unable to identify class-1 probability.")

    probability = matrix[:, int(positive_positions[0])]

    if (
        not np.isfinite(probability).all()
        or (probability < 0).any()
        or (probability > 1).any()
    ):
        raise ValueError("Invalid temporal probabilities.")

    return probability


def calculate_fixed_metrics(y_true, y_prob, threshold):
    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)
    y_pred = (y_prob >= float(threshold)).astype(int)

    if len(y_true) != len(y_prob):
        raise ValueError("Outcome and probability lengths differ.")
    if set(np.unique(y_true)) != {0, 1}:
        raise ValueError("Both outcome classes are required.")

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1],
    ).ravel()

    sensitivity = recall_score(y_true, y_pred, zero_division=0)
    specificity = tn / (tn + fp) if (tn + fp) else np.nan
    precision = precision_score(y_true, y_pred, zero_division=0)

    return {
        "AUC": float(roc_auc_score(y_true, y_prob)),
        "Average_Precision": float(
            average_precision_score(y_true, y_prob)
        ),
        "Brier_Score": float(brier_score_loss(y_true, y_prob)),
        "Sensitivity": float(sensitivity),
        "Specificity": float(specificity),
        "Precision": float(precision),
        "F1": float(f1_score(y_true, y_pred, zero_division=0)),
        "Predicted_Positive_N": int(y_pred.sum()),
        "Predicted_Positive_Rate": float(y_pred.mean()),
        "TP": int(tp),
        "FP": int(fp),
        "FN": int(fn),
        "TN": int(tn),
        "y_pred": y_pred,
    }


# ============================================================
# 5. STRATIFIED FIXED-PREDICTION BOOTSTRAP
# ============================================================

def stratified_bootstrap(
    y_true,
    y_prob,
    threshold,
    n_bootstraps,
    random_state,
):
    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)

    positive_positions = np.flatnonzero(y_true == 1)
    negative_positions = np.flatnonzero(y_true == 0)

    if len(positive_positions) == 0 or len(negative_positions) == 0:
        raise ValueError("Both classes are required for bootstrap.")

    rng = np.random.default_rng(random_state)
    distributions = {
        metric: np.full(n_bootstraps, np.nan, dtype=float)
        for metric in BOOTSTRAP_METRICS
    }

    for bootstrap_index in range(n_bootstraps):
        sampled_positive = rng.choice(
            positive_positions,
            size=len(positive_positions),
            replace=True,
        )
        sampled_negative = rng.choice(
            negative_positions,
            size=len(negative_positions),
            replace=True,
        )
        sampled_positions = np.concatenate([
            sampled_positive,
            sampled_negative,
        ])
        rng.shuffle(sampled_positions)

        try:
            replicate_metrics = calculate_fixed_metrics(
                y_true[sampled_positions],
                y_prob[sampled_positions],
                threshold,
            )
            for metric in BOOTSTRAP_METRICS:
                distributions[metric][bootstrap_index] = (
                    replicate_metrics[metric]
                )
        except Exception:
            continue

    return distributions


# ============================================================
# 6. FIXED TEMPORAL PREDICTIONS AND BOOTSTRAP ANALYSES
# ============================================================

secondary_temporal_predictions = {}
secondary_temporal_bootstrap_distributions = {}

temporal_result_rows = []
bootstrap_result_rows = []
bootstrap_status_rows = []
input_audit_rows = []

for outcome_index, outcome in enumerate(SECONDARY_TARGETS):
    y_temporal = (
        secondary_temporal_2025[outcome]
        .astype(int)
        .copy()
        .reset_index(drop=True)
    )

    if set(y_temporal.unique()) != {0, 1}:
        raise RuntimeError(f"{outcome}: temporal outcome is not binary.")

    fitted_pipeline = secondary_fitted_pipelines[outcome]
    locked_threshold = float(secondary_locked_thresholds[outcome])

    if not hasattr(fitted_pipeline, "named_steps"):
        raise TypeError(f"{outcome}: fitted object is not a pipeline.")

    expected_steps = ["over_sampler", "under_sampler", "estimator"]
    if list(fitted_pipeline.named_steps) != expected_steps:
        raise RuntimeError(f"{outcome}: fitted pipeline step order changed.")

    if type(
        fitted_pipeline.named_steps["over_sampler"]
    ).__name__ != "SMOTENC":
        raise RuntimeError(f"{outcome}: over-sampler is not SMOTENC.")

    if type(
        fitted_pipeline.named_steps["under_sampler"]
    ).__name__ != "RandomUnderSampler":
        raise RuntimeError(
            f"{outcome}: under-sampler is not RandomUnderSampler."
        )

    y_probability = positive_probability(
        fitted_pipeline,
        X_temporal_secondary,
    )

    point_metrics = calculate_fixed_metrics(
        y_temporal,
        y_probability,
        locked_threshold,
    )

    secondary_temporal_predictions[outcome] = {
        "validation_set": "Temporal 2025",
        "y_true": y_temporal.to_numpy(dtype=int),
        "y_prob": y_probability.copy(),
        "y_pred": point_metrics["y_pred"].copy(),
        "source_row_position": temporal_source_positions.copy(),
        "locked_threshold": locked_threshold,
        "n": len(y_temporal),
        "events": int(y_temporal.sum()),
        "prevalence": float(y_temporal.mean()),
        "model_refitted": False,
        "temporal_resampled_for_model_fitting": False,
        "threshold_modified": False,
    }

    random_state = BASE_BOOTSTRAP_RANDOM_STATE + outcome_index * 100
    distributions = stratified_bootstrap(
        y_true=y_temporal,
        y_prob=y_probability,
        threshold=locked_threshold,
        n_bootstraps=N_BOOTSTRAPS,
        random_state=random_state,
    )

    secondary_temporal_bootstrap_distributions[outcome] = distributions

    temporal_result_rows.append({
        "Outcome": outcome,
        "Screening_Model": SECONDARY_MODEL_NAME,
        "Algorithm": SECONDARY_ALGORITHM_NAME,
        "Validation_Set": "Temporal 2025",
        "N": len(y_temporal),
        "Events": int(y_temporal.sum()),
        "Prevalence": float(y_temporal.mean()),
        "Locked_Threshold": locked_threshold,
        "Threshold_Source": "PART S2 development OOF only",
        **{
            key: value
            for key, value in point_metrics.items()
            if key != "y_pred"
        },
    })

    minimum_valid = N_BOOTSTRAPS

    for metric in BOOTSTRAP_METRICS:
        valid_values = distributions[metric][
            np.isfinite(distributions[metric])
        ]
        valid_n = int(len(valid_values))
        minimum_valid = min(minimum_valid, valid_n)
        ci_computed = valid_n >= MINIMUM_VALID_BOOTSTRAPS

        if ci_computed:
            lower_ci = float(
                np.percentile(valid_values, LOWER_PERCENTILE)
            )
            upper_ci = float(
                np.percentile(valid_values, UPPER_PERCENTILE)
            )
        else:
            lower_ci = np.nan
            upper_ci = np.nan

        bootstrap_result_rows.append({
            "Outcome": outcome,
            "Screening_Model": SECONDARY_MODEL_NAME,
            "Algorithm": SECONDARY_ALGORITHM_NAME,
            "Validation_Set": "Temporal 2025",
            "N": len(y_temporal),
            "Events": int(y_temporal.sum()),
            "Locked_Threshold": locked_threshold,
            "Metric": metric,
            "Estimate": float(point_metrics[metric]),
            "CI_Lower": lower_ci,
            "CI_Upper": upper_ci,
            "CI_Level": CI_LEVEL,
            "CI_Method": (
                "Stratified percentile bootstrap "
                "(2.5th-97.5th percentiles)"
            ),
            "Requested_Bootstrap_Replicates": N_BOOTSTRAPS,
            "Valid_Bootstrap_Replicates": valid_n,
            "CI_Computed": ci_computed,
            "Random_State": random_state,
        })

    bootstrap_status_rows.append({
        "Outcome": outcome,
        "Validation_Set": "Temporal 2025",
        "Requested_Bootstrap_Replicates": N_BOOTSTRAPS,
        "Minimum_Valid_Replicates_Across_Metrics": minimum_valid,
        "Minimum_Required_Replicates": MINIMUM_VALID_BOOTSTRAPS,
        "All_CIs_Computed": minimum_valid >= MINIMUM_VALID_BOOTSTRAPS,
        "Status": (
            "Success"
            if minimum_valid >= MINIMUM_VALID_BOOTSTRAPS
            else "Insufficient valid replicates"
        ),
    })

    input_checks = {
        "N_Correct": len(y_temporal) == EXPECTED_TEMPORAL_N,
        "Outcome_Binary": set(y_temporal.unique()) == {0, 1},
        "No_Outcome_Missing": not y_temporal.isna().any(),
        "No_Predictor_Missing": not X_temporal_secondary.isna().any().any(),
        "Probability_Length_Correct": len(y_probability) == len(y_temporal),
        "Probabilities_Valid": bool(
            np.isfinite(y_probability).all()
            and (y_probability >= 0).all()
            and (y_probability <= 1).all()
        ),
        "Locked_Threshold_Finite": np.isfinite(locked_threshold),
        "Predictions_Match_Locked_Threshold": np.array_equal(
            point_metrics["y_pred"],
            (y_probability >= locked_threshold).astype(int),
        ),
        "Model_Not_Refitted": True,
        "Temporal_Not_Resampled_For_Model_Fitting": True,
        "Threshold_Not_Modified": True,
    }

    input_audit_rows.append({
        "Outcome": outcome,
        **input_checks,
        "Input_Ready": bool(all(input_checks.values())),
    })


# ============================================================
# 7. AGGREGATE TABLES
# ============================================================

partS4_temporal_results = pd.DataFrame(temporal_result_rows)
partS4_bootstrap_results = pd.DataFrame(bootstrap_result_rows)
partS4_bootstrap_status = pd.DataFrame(bootstrap_status_rows)
partS4_input_audit = pd.DataFrame(input_audit_rows)


def format_ci(row):
    return (
        f"{float(row['Estimate']):.3f} "
        f"({float(row['CI_Lower']):.3f}-"
        f"{float(row['CI_Upper']):.3f})"
    )


publication_rows = []

for outcome in SECONDARY_TARGETS:
    point_row = partS4_temporal_results.loc[
        partS4_temporal_results["Outcome"] == outcome
    ].iloc[0]

    publication_row = {
        "Outcome": outcome,
        "Screening_Model": SECONDARY_MODEL_NAME,
        "Algorithm": SECONDARY_ALGORITHM_NAME,
        "Validation_Set": "Temporal 2025",
        "N": int(point_row["N"]),
        "Events": int(point_row["Events"]),
        "Locked_Threshold": float(point_row["Locked_Threshold"]),
    }

    for metric in BOOTSTRAP_METRICS:
        ci_row = partS4_bootstrap_results.loc[
            (partS4_bootstrap_results["Outcome"] == outcome)
            & (partS4_bootstrap_results["Metric"] == metric)
        ]
        if len(ci_row) != 1:
            raise RuntimeError(
                f"{outcome}: expected one bootstrap row for {metric}."
            )
        publication_row[f"{metric}_95CI"] = format_ci(ci_row.iloc[0])

    publication_row.update({
        "TP": int(point_row["TP"]),
        "FP": int(point_row["FP"]),
        "FN": int(point_row["FN"]),
        "TN": int(point_row["TN"]),
    })
    publication_rows.append(publication_row)

partS4_publication_table = pd.DataFrame(publication_rows)


# ============================================================
# 8. GLOBAL VERIFICATION
# ============================================================

all_inputs_ready = bool(
    len(partS4_input_audit) == len(SECONDARY_TARGETS)
    and partS4_input_audit["Input_Ready"].all()
)

all_bootstraps_ready = bool(
    len(partS4_bootstrap_status) == len(SECONDARY_TARGETS)
    and partS4_bootstrap_status["All_CIs_Computed"].all()
)

all_ci_rows_ready = bool(
    len(partS4_bootstrap_results)
    == len(SECONDARY_TARGETS) * len(BOOTSTRAP_METRICS)
    and partS4_bootstrap_results["CI_Computed"].all()
)

PARTS4_READY = bool(
    all_inputs_ready
    and all_bootstraps_ready
    and all_ci_rows_ready
    and len(partS4_publication_table) == len(SECONDARY_TARGETS)
)

partS4_verification = pd.DataFrame([{
    "Predictor_Strategy": SECONDARY_MODEL_NAME,
    "Algorithm": SECONDARY_ALGORITHM_NAME,
    "Validation_Set": "Temporal 2025",
    "Outcomes_Evaluated": len(SECONDARY_TARGETS),
    "Temporal_N": len(X_temporal_secondary),
    "Temporal_MDD_Events": int(temporal_mdd.sum()),
    "All_Fixed_Inputs_Ready": all_inputs_ready,
    "All_Bootstrap_Analyses_Ready": all_bootstraps_ready,
    "All_Confidence_Intervals_Computed": all_ci_rows_ready,
    "Model_Refitted": False,
    "Temporal_Resampled_For_Model_Fitting": False,
    "Threshold_Modified": False,
    "Recalibration_Performed": False,
    "Participant_Level_Predictions_Exported": False,
    "Bootstrap_Distributions_Exported": False,
    "PARTS4_Ready": PARTS4_READY,
}])


# ============================================================
# 9. DISPLAY AGGREGATE RESULTS
# ============================================================

print("=" * 110)
print("PART S4 VERSION 2.0")
print("TEMPORAL 2025 VALIDATION AND BOOTSTRAP 95% CI")
print("=" * 110)

print("\nTemporal performance:")
display(partS4_temporal_results)

print("\nBootstrap status:")
display(partS4_bootstrap_status)

print("\nPublication table:")
display(partS4_publication_table)

print("\nInput audit:")
display(partS4_input_audit)

print("\nGlobal verification:")
display(partS4_verification)


# ============================================================
# 10. EXPORT AGGREGATE RESULTS ONLY
# ============================================================

partS4_temporal_results.to_csv(
    "partS4_temporal_2025_results.csv",
    index=False,
    encoding="utf-8-sig",
)

partS4_bootstrap_results.to_csv(
    "partS4_temporal_bootstrap_ci_results.csv",
    index=False,
    encoding="utf-8-sig",
)

partS4_bootstrap_status.to_csv(
    "partS4_temporal_bootstrap_status.csv",
    index=False,
    encoding="utf-8-sig",
)

partS4_publication_table.to_csv(
    "partS4_temporal_2025_publication_table.csv",
    index=False,
    encoding="utf-8-sig",
)

partS4_input_audit.to_csv(
    "partS4_temporal_input_audit.csv",
    index=False,
    encoding="utf-8-sig",
)

partS4_verification.to_csv(
    "partS4_verification.csv",
    index=False,
    encoding="utf-8-sig",
)


# ============================================================
# 11. FINAL SAFETY GATE AND SUMMARY
# ============================================================

if not PARTS4_READY:
    raise RuntimeError("PART S4 methodological verification failed.")

print("\n" + "=" * 110)
print("PART S4 VERSION 2.0 COMPLETED")
print("=" * 110)
print(f"Predictor strategy        : {SECONDARY_MODEL_NAME}")
print(f"Algorithm                 : {SECONDARY_ALGORITHM_NAME}")
print(f"Temporal cohort           : N={len(X_temporal_secondary):,}")
print(f"Secondary outcomes        : {len(SECONDARY_TARGETS)}")
print(f"Bootstrap replicates      : {N_BOOTSTRAPS:,} per outcome")
print(f"Minimum valid replicates  : {MINIMUM_VALID_BOOTSTRAPS:,}")
print("CI method                 : Stratified percentile bootstrap")
print("CI limits                 : 2.5th-97.5th percentiles")
print("Model refitted            : No")
print("Temporal model resampling : No")
print("Threshold modified        : No")
print("Participant rows exported : No")
print("Bootstrap distributions   : Not exported")
print("\nPARTS4_READY:", PARTS4_READY)
print("\nProceed to PART S5 only after PARTS4_READY=True.")
```

## PART S5A: Raw Probability Calibration

```python
# ============================================================
# MDD SECONDARY OUTCOMES ENGINE VERSION 1.0
# PART S5A VERSION 2.0
# RAW CALIBRATION ANALYSIS
# HELD-OUT 2024 + TEMPORAL 2025
# ============================================================
#
# Secondary outcomes:
#   1. Depression_Risk
#   2. Suicide_Risk
#
# Prespecified strategy:
#   M2_Admin_2Q
#
# Prespecified algorithm:
#   AdaBoostClassifier
#
# Calibration principles:
#   - Use FIXED probabilities from PART S3 and PART S4
#   - No model refitting
#   - No threshold modification
#   - No probability recalibration
#   - Calibration analysis is diagnostic only
#
# Metrics:
#   - Calibration-in-the-large difference
#   - Observed / Expected ratio
#   - Calibration intercept
#   - Calibration slope
#   - Brier score
#   - Null Brier score
#   - Brier skill score
#   - ECE
#   - MCE
#   - ICI using LOWESS
#   - E50
#   - E90
#
# Plots:
#   - Quantile-binned calibration points
#   - LOWESS calibration curve
#   - Ideal 45-degree line
#
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.special import logit

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import brier_score_loss

from statsmodels.nonparametric.smoothers_lowess import lowess


# ============================================================
# 2. REQUIRED OBJECTS
# ============================================================

required_objects = [

    "secondary_heldout_predictions",
    "partS3_heldout_results",
    "partS3_verification",

    "secondary_temporal_predictions",
    "partS4_temporal_results",
    "partS4_verification"

]


missing_objects = [

    object_name

    for object_name in required_objects

    if object_name not in globals()

]


if missing_objects:

    raise NameError(

        "Missing required objects: "

        + ", ".join(
            missing_objects
        )

    )


# ============================================================
# 3. VERIFY PREVIOUS PARTS
# ============================================================

if not bool(
    partS3_verification[
        "PARTS3_Ready"
    ].all()
):

    raise RuntimeError(
        "PART S3 is not ready."
    )


if not bool(
    partS4_verification[
        "PARTS4_Ready"
    ].all()
):

    raise RuntimeError(
        "PART S4 is not ready."
    )


# ============================================================
# 4. SETTINGS
# ============================================================

SECONDARY_TARGETS = [

    "Depression_Risk",
    "Suicide_Risk"

]


SECONDARY_MODEL_NAME = (
    "M2_Admin_2Q"
)


SECONDARY_ALGORITHM_NAME = (
    "AdaBoostClassifier"
)


CALIBRATION_BINS = 10


LOWESS_FRAC = 0.40

LOWESS_ITERATIONS = 0


PROBABILITY_EPSILON = 1e-6


OUTPUT_DIR = (
    "partS5a_secondary_calibration_plots"
)


os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# 5. HELPER: CALIBRATION REGRESSION
# ============================================================
#
# Logistic calibration model:
#
# logit(P(Y=1)) =
#     intercept + slope * logit(predicted probability)
#
# Ideal:
#   intercept = 0
#   slope     = 1
#
# ============================================================

def calculate_calibration_regression(
    y_true,
    y_prob
):

    y_true = np.asarray(
        y_true,
        dtype=int
    )


    y_prob = np.asarray(
        y_prob,
        dtype=float
    )


    clipped_prob = np.clip(

        y_prob,

        PROBABILITY_EPSILON,

        1 - PROBABILITY_EPSILON

    )


    logits = logit(
        clipped_prob
    ).reshape(
        -1,
        1
    )


    model = LogisticRegression(

        penalty=None,

        solver="lbfgs",

        max_iter=10000

    )


    model.fit(
        logits,
        y_true
    )


    intercept = float(
        model.intercept_[0]
    )


    slope = float(
        model.coef_[0][0]
    )


    return (
        intercept,
        slope
    )


# ============================================================
# 6. HELPER: QUANTILE CALIBRATION BINS
# ============================================================

def create_calibration_bins(
    y_true,
    y_prob,
    requested_bins=10
):

    temp = pd.DataFrame({

        "Observed":
            np.asarray(
                y_true,
                dtype=int
            ),

        "Probability":
            np.asarray(
                y_prob,
                dtype=float
            )

    })


    unique_probability_count = (
        temp[
            "Probability"
        ]
        .nunique()
    )


    actual_bins = min(

        requested_bins,

        unique_probability_count

    )


    if actual_bins < 2:

        raise RuntimeError(

            "Insufficient unique probabilities "
            "for calibration binning."

        )


    temp[
        "Calibration_Bin"
    ] = pd.qcut(

        temp[
            "Probability"
        ],

        q=actual_bins,

        labels=False,

        duplicates="drop"

    )


    calibration_bins = (

        temp
        .groupby(
            "Calibration_Bin",
            observed=True
        )
        .agg(

            N=(
                "Observed",
                "size"
            ),

            Events=(
                "Observed",
                "sum"
            ),

            Mean_Predicted_Probability=(
                "Probability",
                "mean"
            ),

            Observed_Proportion=(
                "Observed",
                "mean"
            ),

            Min_Predicted_Probability=(
                "Probability",
                "min"
            ),

            Max_Predicted_Probability=(
                "Probability",
                "max"
            )

        )
        .reset_index()

    )


    calibration_bins[
        "Absolute_Error"
    ] = (

        calibration_bins[
            "Observed_Proportion"
        ]

        -

        calibration_bins[
            "Mean_Predicted_Probability"
        ]

    ).abs()


    return calibration_bins


# ============================================================
# 7. HELPER: ECE / MCE
# ============================================================

def calculate_ece_mce(
    calibration_bins
):

    total_n = float(
        calibration_bins[
            "N"
        ].sum()
    )


    weights = (

        calibration_bins[
            "N"
        ]

        /

        total_n

    )


    absolute_error = (

        calibration_bins[
            "Observed_Proportion"
        ]

        -

        calibration_bins[
            "Mean_Predicted_Probability"
        ]

    ).abs()


    ece = float(

        np.sum(
            weights
            *
            absolute_error
        )

    )


    mce = float(

        absolute_error.max()

    )


    return (
        ece,
        mce
    )


# ============================================================
# 8. HELPER: LOWESS CALIBRATION ERROR
# ============================================================

def calculate_lowess_calibration_metrics(
    y_true,
    y_prob,
    frac=0.40,
    it=0
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_prob = np.asarray(
        y_prob,
        dtype=float
    )


    order = np.argsort(
        y_prob
    )


    sorted_prob = y_prob[
        order
    ]


    sorted_y = y_true[
        order
    ]


    lowess_fit = lowess(

        endog=
            sorted_y,

        exog=
            sorted_prob,

        frac=
            frac,

        it=
            it,

        return_sorted=True

    )


    lowess_x = (
        lowess_fit[
            :,
            0
        ]
    )


    lowess_y = (
        lowess_fit[
            :,
            1
        ]
    )


    unique_x, unique_index = np.unique(

        lowess_x,

        return_index=True

    )


    unique_y = lowess_y[
        unique_index
    ]


    smoothed_probability = np.interp(

        y_prob,

        unique_x,

        unique_y

    )


    smoothed_probability = np.clip(

        smoothed_probability,

        0,

        1

    )


    absolute_error = np.abs(

        smoothed_probability
        -
        y_prob

    )


    ici = float(
        np.mean(
            absolute_error
        )
    )


    e50 = float(
        np.percentile(
            absolute_error,
            50
        )
    )


    e90 = float(
        np.percentile(
            absolute_error,
            90
        )
    )


    return {

        "ICI":
            ici,

        "E50":
            e50,

        "E90":
            e90,

        "LOWESS_X":
            lowess_x,

        "LOWESS_Y":
            np.clip(
                lowess_y,
                0,
                1
            )

    }


# ============================================================
# 9. HELPER: FULL RAW CALIBRATION ANALYSIS
# ============================================================

def analyze_raw_calibration(
    y_true,
    y_prob,
    requested_bins=10
):

    y_true = np.asarray(
        y_true,
        dtype=int
    )


    y_prob = np.asarray(
        y_prob,
        dtype=float
    )


    if len(
        y_true
    ) != len(
        y_prob
    ):

        raise ValueError(
            "Observed and predicted arrays have "
            "different lengths."
        )


    if np.isnan(
        y_prob
    ).any():

        raise ValueError(
            "Missing predicted probabilities detected."
        )


    if np.any(
        (y_prob < 0)
        |
        (y_prob > 1)
    ):

        raise ValueError(
            "Predicted probabilities outside [0, 1]."
        )


    n = int(
        len(
            y_true
        )
    )


    events = int(
        np.sum(
            y_true
        )
    )


    prevalence = float(

        np.mean(
            y_true
        )

    )


    mean_probability = float(

        np.mean(
            y_prob
        )

    )


    observed_events = float(
        events
    )


    expected_events = float(

        np.sum(
            y_prob
        )

    )


    calibration_in_large_difference = float(

        mean_probability
        -
        prevalence

    )


    observed_expected_ratio = (

        observed_events
        /
        expected_events

        if expected_events > 0

        else np.nan

    )


    calibration_intercept, calibration_slope = (
        calculate_calibration_regression(

            y_true,
            y_prob

        )
    )


    brier = float(

        brier_score_loss(
            y_true,
            y_prob
        )

    )


    null_probabilities = np.full(

        n,

        prevalence,

        dtype=float

    )


    null_brier = float(

        brier_score_loss(

            y_true,

            null_probabilities

        )

    )


    brier_skill_score = (

        1
        -
        (
            brier
            /
            null_brier
        )

        if null_brier > 0

        else np.nan

    )


    calibration_bins = (
        create_calibration_bins(

            y_true,
            y_prob,

            requested_bins=
                requested_bins

        )
    )


    ece, mce = (
        calculate_ece_mce(
            calibration_bins
        )
    )


    lowess_metrics = (
        calculate_lowess_calibration_metrics(

            y_true,
            y_prob,

            frac=
                LOWESS_FRAC,

            it=
                LOWESS_ITERATIONS

        )
    )


    summary = {

        "N":
            n,

        "Events":
            events,

        "Nonevents":
            n - events,

        "Prevalence":
            prevalence,

        "Mean_Predicted_Probability":
            mean_probability,

        "Calibration_In_The_Large_Difference":
            calibration_in_large_difference,

        "Observed_Events":
            observed_events,

        "Expected_Events":
            expected_events,

        "Observed_to_Expected_Ratio":
            observed_expected_ratio,

        "Calibration_Intercept":
            calibration_intercept,

        "Calibration_Slope":
            calibration_slope,

        "Brier_Score":
            brier,

        "Null_Brier_Score":
            null_brier,

        "Brier_Skill_Score":
            brier_skill_score,

        "ECE":
            ece,

        "MCE":
            mce,

        "ICI":
            lowess_metrics[
                "ICI"
            ],

        "E50":
            lowess_metrics[
                "E50"
            ],

        "E90":
            lowess_metrics[
                "E90"
            ],

        "Actual_Calibration_Bins":
            int(
                len(
                    calibration_bins
                )
            )

    }


    return (

        summary,

        calibration_bins,

        lowess_metrics

    )


# ============================================================
# 10. INPUT PREDICTION TABLES
# ============================================================

def prediction_dictionary_to_table(prediction_dictionary):

    rows = []

    for outcome in SECONDARY_TARGETS:

        if outcome not in prediction_dictionary:

            raise KeyError(
                f"Missing fixed predictions for {outcome}."
            )

        outcome_predictions = prediction_dictionary[outcome]

        y_true = np.asarray(
            outcome_predictions["y_true"],
            dtype=int,
        )

        y_prob = np.asarray(
            outcome_predictions["y_prob"],
            dtype=float,
        )

        if len(y_true) != len(y_prob):

            raise RuntimeError(
                f"{outcome}: observed and probability lengths differ."
            )

        rows.append(
            pd.DataFrame({
                "Outcome": outcome,
                "Observed": y_true,
                "Predicted_Probability": y_prob,
            })
        )

    return pd.concat(rows, ignore_index=True)

validation_sources = {

    "Held-out 2024":
        prediction_dictionary_to_table(
            secondary_heldout_predictions
        ),

    "Temporal 2025":
        prediction_dictionary_to_table(
            secondary_temporal_predictions
        )

}


# ============================================================
# 11. RESULT CONTAINERS
# ============================================================

calibration_result_rows = []

calibration_bin_frames = []

calibration_status_rows = []

plot_manifest_rows = []


# ============================================================
# 12. CALIBRATION ANALYSIS LOOP
# ============================================================

print(
    "=" * 110
)

print(
    "PART S5A VERSION 2.0: "
    "RAW CALIBRATION ANALYSIS"
)

print(
    "=" * 110
)


print(
    f"\nRequested calibration bins : "
    f"{CALIBRATION_BINS}"
)


print(
    f"ICI method                 : "
    f"LOWESS absolute calibration error "
    f"(frac={LOWESS_FRAC:.2f}, "
    f"it={LOWESS_ITERATIONS})"
)


for validation_set, prediction_table in (
    validation_sources.items()
):


    print(
        "\n" + "=" * 110
    )

    print(
        validation_set.upper()
    )

    print(
        "=" * 110
    )


    for outcome in SECONDARY_TARGETS:


        print(
            f"\n{outcome}"
        )


        outcome_predictions = (

            prediction_table[

                prediction_table[
                    "Outcome"
                ]
                ==
                outcome

            ]
            .copy()

        )


        if len(
            outcome_predictions
        ) == 0:

            raise RuntimeError(

                f"{validation_set} / {outcome}: "
                "prediction table is empty."

            )


        y_true = (

            outcome_predictions[
                "Observed"
            ]
            .astype(int)
            .to_numpy()

        )


        y_prob = (

            outcome_predictions[
                "Predicted_Probability"
            ]
            .astype(float)
            .to_numpy()

        )


        try:

            (
                summary,
                bins,
                lowess_metrics

            ) = analyze_raw_calibration(

                y_true,
                y_prob,

                requested_bins=
                    CALIBRATION_BINS

            )


            summary.update({

                "Validation_Set":
                    validation_set,

                "Outcome":
                    outcome,

                "Screening_Model":
                    SECONDARY_MODEL_NAME,

                "Algorithm":
                    SECONDARY_ALGORITHM_NAME,

                "Calibration_Type":
                    "Raw probability diagnostic",

                "LOWESS_Frac":
                    LOWESS_FRAC,

                "LOWESS_Iterations":
                    LOWESS_ITERATIONS

            })


            calibration_result_rows.append(
                summary
            )


            # ------------------------------------------------
            # Store bin-level results
            # ------------------------------------------------

            bins[
                "Validation_Set"
            ] = validation_set


            bins[
                "Outcome"
            ] = outcome


            bins[
                "Screening_Model"
            ] = SECONDARY_MODEL_NAME


            bins[
                "Algorithm"
            ] = SECONDARY_ALGORITHM_NAME


            calibration_bin_frames.append(
                bins
            )


            # ------------------------------------------------
            # Calibration plot
            # ------------------------------------------------

            safe_validation_name = (

                validation_set
                .lower()
                .replace(
                    " ",
                    "_"
                )
                .replace(
                    "-",
                    ""
                )

            )


            safe_outcome_name = (

                outcome
                .lower()

            )


            png_file = os.path.join(

                OUTPUT_DIR,

                (
                    f"{safe_validation_name}_"
                    f"{safe_outcome_name}_"
                    f"raw_calibration.png"
                )

            )


            pdf_file = os.path.join(

                OUTPUT_DIR,

                (
                    f"{safe_validation_name}_"
                    f"{safe_outcome_name}_"
                    f"raw_calibration.pdf"
                )

            )


            fig, ax = plt.subplots(
                figsize=(
                    7,
                    7
                )
            )


            # Ideal line

            ax.plot(
                [0, 1],
                [0, 1],
                linestyle="--",
                label="Ideal calibration"
            )


            # Binned points

            ax.scatter(

                bins[
                    "Mean_Predicted_Probability"
                ],

                bins[
                    "Observed_Proportion"
                ],

                s=45,

                label="Quantile bins"

            )


            # LOWESS curve

            ax.plot(

                lowess_metrics[
                    "LOWESS_X"
                ],

                lowess_metrics[
                    "LOWESS_Y"
                ],

                linewidth=2,

                label="LOWESS"

            )


            ax.set_xlim(
                0,
                1
            )


            ax.set_ylim(
                0,
                1
            )


            ax.set_xlabel(
                "Predicted probability"
            )


            ax.set_ylabel(
                "Observed outcome proportion"
            )


            ax.set_title(

                f"{outcome}: "
                f"{validation_set}\n"
                "Raw Probability Calibration"

            )


            ax.legend()


            ax.grid(
                alpha=0.25
            )


            fig.tight_layout()


            fig.savefig(

                png_file,

                dpi=300,

                bbox_inches="tight"

            )


            fig.savefig(

                pdf_file,

                bbox_inches="tight"

            )


            plt.close(
                fig
            )


            plot_created = bool(

                os.path.exists(
                    png_file
                )

                and

                os.path.exists(
                    pdf_file
                )

            )


            plot_manifest_rows.append({

                "Validation_Set":
                    validation_set,

                "Outcome":
                    outcome,

                "PNG_File":
                    png_file,

                "PDF_File":
                    pdf_file,

                "Plot_Created":
                    plot_created

            })


            calibration_status_rows.append({

                "Validation_Set":
                    validation_set,

                "Outcome":
                    outcome,

                "Status":
                    "Success",

                "N":
                    summary[
                        "N"
                    ],

                "Events":
                    summary[
                        "Events"
                    ],

                "Requested_Bins":
                    CALIBRATION_BINS,

                "Actual_Bins":
                    summary[
                        "Actual_Calibration_Bins"
                    ],

                "All_Core_Metrics_Finite":
                    bool(

                        np.isfinite(
                            [
                                summary[
                                    "Prevalence"
                                ],

                                summary[
                                    "Mean_Predicted_Probability"
                                ],

                                summary[
                                    "Calibration_Intercept"
                                ],

                                summary[
                                    "Calibration_Slope"
                                ],

                                summary[
                                    "Brier_Score"
                                ],

                                summary[
                                    "ECE"
                                ],

                                summary[
                                    "ICI"
                                ]

                            ]
                        ).all()

                    ),

                "Plot_Created":
                    plot_created,

                "Error":
                    ""

            })


            print(

                f"N={summary['N']:,} | "
                f"Events={summary['Events']:,} | "
                f"Mean p={summary['Mean_Predicted_Probability']:.4f} | "
                f"Intercept={summary['Calibration_Intercept']:.3f} | "
                f"Slope={summary['Calibration_Slope']:.3f} | "
                f"Brier={summary['Brier_Score']:.4f} | "
                f"ECE={summary['ECE']:.4f} | "
                f"ICI={summary['ICI']:.4f}"

            )


        except Exception as error:


            calibration_status_rows.append({

                "Validation_Set":
                    validation_set,

                "Outcome":
                    outcome,

                "Status":
                    "Failed",

                "N":
                    len(
                        outcome_predictions
                    ),

                "Events":
                    int(
                        outcome_predictions[
                            "Observed"
                        ]
                        .sum()
                    ),

                "Requested_Bins":
                    CALIBRATION_BINS,

                "Actual_Bins":
                    np.nan,

                "All_Core_Metrics_Finite":
                    False,

                "Plot_Created":
                    False,

                "Error":
                    str(
                        error
                    )

            })


            print(
                "FAILED"
            )

            print(
                f"Reason: {error}"
            )


# ============================================================
# 13. CREATE RESULT TABLES
# ============================================================

partS5a_calibration_results = pd.DataFrame(
    calibration_result_rows
)


if len(
    calibration_bin_frames
) > 0:

    partS5a_calibration_bins = pd.concat(

        calibration_bin_frames,

        ignore_index=True

    )

else:

    partS5a_calibration_bins = pd.DataFrame()


partS5a_calibration_status = pd.DataFrame(
    calibration_status_rows
)


partS5a_calibration_plot_manifest = pd.DataFrame(
    plot_manifest_rows
)


# ============================================================
# 14. BRIER CONSISTENCY CHECK
# ============================================================

brier_check_rows = []


for validation_set in [

    "Held-out 2024",
    "Temporal 2025"

]:


    for outcome in SECONDARY_TARGETS:


        calibration_row = (

            partS5a_calibration_results[

                (
                    partS5a_calibration_results[
                        "Validation_Set"
                    ]
                    ==
                    validation_set
                )

                &

                (
                    partS5a_calibration_results[
                        "Outcome"
                    ]
                    ==
                    outcome
                )

            ]

        )


        if len(
            calibration_row
        ) != 1:

            continue


        if validation_set == "Held-out 2024":

            source_row = (

                partS3_heldout_results[

                    partS3_heldout_results[
                        "Outcome"
                    ]
                    ==
                    outcome

                ]
                .iloc[0]

            )

        else:

            source_row = (

                partS4_temporal_results[

                    partS4_temporal_results[
                        "Outcome"
                    ]
                    ==
                    outcome

                ]
                .iloc[0]

            )


        calibration_brier = float(

            calibration_row
            .iloc[0][
                "Brier_Score"
            ]

        )


        source_brier = float(

            source_row[
                "Brier_Score"
            ]

        )


        brier_check_rows.append({

            "Validation_Set":
                validation_set,

            "Outcome":
                outcome,

            "Calibration_Brier":
                calibration_brier,

            "Source_Brier":
                source_brier,

            "Absolute_Difference":
                abs(
                    calibration_brier
                    -
                    source_brier
                ),

            "Brier_Matches_Source":
                bool(

                    np.isclose(

                        calibration_brier,

                        source_brier,

                        atol=1e-12,

                        rtol=0

                    )

                )

        })


partS5a_brier_verification = pd.DataFrame(
    brier_check_rows
)


# ============================================================
# 15. PUBLICATION TABLE
# ============================================================

publication_columns = [

    "Validation_Set",
    "Outcome",

    "N",
    "Events",
    "Prevalence",

    "Mean_Predicted_Probability",

    "Calibration_In_The_Large_Difference",

    "Observed_to_Expected_Ratio",

    "Calibration_Intercept",

    "Calibration_Slope",

    "Brier_Score",

    "Null_Brier_Score",

    "Brier_Skill_Score",

    "ECE",

    "ICI",

    "E50",

    "E90",

    "Actual_Calibration_Bins"

]


partS5a_publication_table = (

    partS5a_calibration_results[

        publication_columns

    ]
    .copy()

)


# ============================================================
# 16. VERIFICATION
# ============================================================

verification_rows = []


for validation_set in [

    "Held-out 2024",
    "Temporal 2025"

]:


    for outcome in SECONDARY_TARGETS:


        status_rows = (

            partS5a_calibration_status[

                (
                    partS5a_calibration_status[
                        "Validation_Set"
                    ]
                    ==
                    validation_set
                )

                &

                (
                    partS5a_calibration_status[
                        "Outcome"
                    ]
                    ==
                    outcome
                )

            ]

        )


        brier_rows = (

            partS5a_brier_verification[

                (
                    partS5a_brier_verification[
                        "Validation_Set"
                    ]
                    ==
                    validation_set
                )

                &

                (
                    partS5a_brier_verification[
                        "Outcome"
                    ]
                    ==
                    outcome
                )

            ]

        )


        if len(
            status_rows
        ) != 1:

            success = False

            metrics_finite = False

            plot_created = False

        else:

            status_row = (
                status_rows
                .iloc[0]
            )

            success = bool(

                status_row[
                    "Status"
                ]
                ==
                "Success"

            )

            metrics_finite = bool(

                status_row[
                    "All_Core_Metrics_Finite"
                ]

            )

            plot_created = bool(

                status_row[
                    "Plot_Created"
                ]

            )


        if len(
            brier_rows
        ) != 1:

            brier_match = False

        else:

            brier_match = bool(

                brier_rows
                .iloc[0][
                    "Brier_Matches_Source"
                ]

            )


        verification_rows.append({

            "Validation_Set":
                validation_set,

            "Outcome":
                outcome,

            "Analysis_Success":
                success,

            "All_Core_Metrics_Finite":
                metrics_finite,

            "Calibration_Plot_Created":
                plot_created,

            "Brier_Matches_S3_or_S4":
                brier_match,

            "No_Model_Refit":
                True,

            "No_Threshold_Modification":
                True,

            "No_Probability_Recalibration":
                True,

            "Fixed_Validation_Probabilities_Used":
                True,

            "Participant_Level_Predictions_Exported":
                False

        })


partS5a_verification = pd.DataFrame(
    verification_rows
)


partS5a_verification[
    "PARTS5A_Ready"
] = (

    partS5a_verification[

        [
            "Analysis_Success",
            "All_Core_Metrics_Finite",
            "Calibration_Plot_Created",
            "Brier_Matches_S3_or_S4",
            "No_Model_Refit",
            "No_Threshold_Modification",
            "No_Probability_Recalibration",
            "Fixed_Validation_Probabilities_Used"
        ]

    ]
    .all(
        axis=1
    )

)


PARTS5A_READY = bool(
    len(partS5a_verification) == 4
    and partS5a_verification["PARTS5A_Ready"].all()
)


# ============================================================
# 17. DISPLAY
# ============================================================

print(
    "\n" + "=" * 110
)

print(
    "RAW CALIBRATION SUMMARY"
)

print(
    "=" * 110
)


display(

    partS5a_calibration_results
    .round(
        4
    )

)


print(
    "\n" + "=" * 110
)

print(
    "PUBLICATION TABLE"
)

print(
    "=" * 110
)


display(

    partS5a_publication_table
    .round(
        4
    )

)


print(
    "\n" + "=" * 110
)

print(
    "BRIER CONSISTENCY CHECK"
)

print(
    "=" * 110
)


display(
    partS5a_brier_verification
)


print(
    "\n" + "=" * 110
)

print(
    "PART S5A VERIFICATION"
)

print(
    "=" * 110
)


display(
    partS5a_verification
)


# ============================================================
# 18. HARD SAFETY CHECK
# ============================================================

if not PARTS5A_READY:

    raise RuntimeError(

        "PART S5A verification failed. "
        "Do not proceed to recalibration analysis."

    )


# ============================================================
# 19. EXPORT
# ============================================================

partS5a_calibration_results.to_csv(

    "partS5a_raw_calibration_results.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5a_calibration_bins.to_csv(

    "partS5a_raw_calibration_bins.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5a_calibration_status.to_csv(

    "partS5a_raw_calibration_status.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5a_brier_verification.to_csv(

    "partS5a_brier_verification.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5a_publication_table.to_csv(

    "partS5a_raw_calibration_publication_table.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5a_calibration_plot_manifest.to_csv(

    "partS5a_calibration_plot_manifest.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5a_verification.to_csv(

    "partS5a_verification.csv",

    index=False,

    encoding="utf-8-sig"

)


# ============================================================
# 20. FINAL REPORT
# ============================================================

print(
    "\n" + "=" * 110
)

print(
    "PART S5A VERSION 2.0 COMPLETED"
)

print(
    "=" * 110
)


print(
    f"\nSecondary outcomes analyzed : "
    f"{len(SECONDARY_TARGETS)}"
)


print(
    "Validation cohorts analyzed: 2"
)


print(
    f"Requested calibration bins : "
    f"{CALIBRATION_BINS}"
)


print(
    f"ICI method                  : "
    f"LOWESS absolute calibration error "
    f"(frac={LOWESS_FRAC:.2f}, "
    f"it={LOWESS_ITERATIONS})"
)


print(
    f"Calibration plot directory  : "
    f"{OUTPUT_DIR}"
)


print(
    "\nArtifacts created:"
)

print(
    "- partS5a_calibration_results"
)

print(
    "- partS5a_calibration_bins"
)

print(
    "- partS5a_calibration_status"
)

print(
    "- partS5a_brier_verification"
)

print(
    "- partS5a_publication_table"
)

print(
    "- partS5a_calibration_plot_manifest"
)

print(
    "- partS5a_verification"
)


print(
    "\nNo model was refitted, no locked threshold was modified, "
    "and no probability was recalibrated."
)


print(
    "Calibration regression and LOWESS smoothing were used "
    "only as diagnostic summaries of fixed validation "
    "probabilities."
)


print(
    "Participant-level predictions were retained in memory and "
    "were not exported."
)


print(
    f"\nPARTS5A_READY: {PARTS5A_READY}"
)


print(
    "\nProceed to PART S5B only after "
    "PARTS5A_Ready=True for all 4 "
    "outcome-validation combinations."
)
```

```python
# ============================================================
# PART S5B PRE-RUN COMPATIBILITY FIX
# Run this cell BEFORE PART S5B
# ============================================================

import pandas as pd

if "partS2_verification" not in globals():
    raise NameError(
        "partS2_verification is not available. "
        "Run PART S2 again."
    )

if not isinstance(partS2_verification, pd.DataFrame):
    raise TypeError(
        "partS2_verification must be a pandas DataFrame."
    )

print(
    "Columns before correction:",
    partS2_verification.columns.tolist(),
)

if "Outcome_Ready" not in partS2_verification.columns:
    raise KeyError(
        "Outcome_Ready is missing from partS2_verification."
    )

# Force creation of the compatibility column
partS2_verification = partS2_verification.copy()

partS2_verification.loc[:, "PARTS2_Ready"] = (
    partS2_verification["Outcome_Ready"]
    .astype(bool)
    .to_numpy()
)

# Return the corrected object to notebook global memory
globals()["partS2_verification"] = partS2_verification

# Verification
assert "PARTS2_Ready" in partS2_verification.columns
assert bool(partS2_verification["PARTS2_Ready"].all())

print(
    "Columns after correction:",
    partS2_verification.columns.tolist(),
)

print(
    "PARTS2_Ready:",
    bool(partS2_verification["PARTS2_Ready"].all()),
)

print("PART S5B PRE-RUN FIX COMPLETED")
```

```python
# ============================================================
# MDD SECONDARY OUTCOMES ENGINE VERSION 1.0
# PART S5B VERSION 2.3 — FULL CORRECTED
# OOF-DERIVED LOGISTIC RECALIBRATION
# AND TRANSPORT TO VALIDATION COHORTS
# ============================================================
#
# Secondary outcomes:
#   1. Depression_Risk
#   2. Suicide_Risk
#
# Prespecified strategy:
#   M2_Admin_2Q
#
# Prespecified algorithm:
#   AdaBoostClassifier
#
# PURPOSE
# ------------------------------------------------------------
# To evaluate whether probability calibration can be improved
# by applying a logistic recalibration mapping learned
# EXCLUSIVELY from 2024 OOF predictions.
#
#
# RECALIBRATION PRINCIPLE
# ------------------------------------------------------------
#
# For each secondary outcome:
#
#   Genuine 2024 OOF predictions
#           ↓
#
#   Fit logistic recalibration mapping:
#
#       logit(P(Y=1))
#           =
#       intercept
#           +
#       slope * logit(raw probability)
#
#           ↓
#
#   LOCK recalibration mapping
#
#           ↓
#
#   Apply unchanged to:
#
#       - Held-out 2024 probabilities
#       - Temporal 2025 probabilities
#
#
# IMPORTANT ANALYTICAL PRINCIPLES
# ------------------------------------------------------------
# - Held-out 2024 outcomes are NOT used to fit recalibration.
# - Temporal 2025 outcomes are NOT used to fit recalibration.
# - No predictive model is refitted.
# - No classification threshold is re-optimized.
# - Raw probabilities are retained.
# - Raw threshold-based classification remains PRIMARY.
# - Recalibrated probabilities are SECONDARY / SUPPLEMENTARY.
#
#
# VERSION 1.1 NUMERICAL FIX
# ------------------------------------------------------------
# A positive logistic-recalibration slope defines a strictly
# monotonic transformation on the logit scale.
#
# However, after expit(), extreme values may numerically
# saturate near 0 or 1 and create probability ties.
# Consequently, probability-scale AUC may differ slightly
# despite mathematical rank preservation.
#
# Therefore:
#
# 1. Rank preservation is verified using the recalibrated
#    LOGIT SCORE before expit().
#
# 2. Probability-scale AUC change is retained as a numerical
#    diagnostic only.
#
# 3. A small probability-scale AUC difference does NOT cause
#    the recalibration analysis to fail.
#
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import statsmodels.api as sm

from scipy.special import (
    logit,
    expit
)

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss
)


# ============================================================
# 2. ADAPT FIXED-PREDICTION OBJECTS FROM PARTS S2-S4
# ============================================================

EXPECTED_SECONDARY_TARGETS = [
    "Depression_Risk",
    "Suicide_Risk",
]


def oof_dictionary_to_table(prediction_dictionary):

    frames = []

    for outcome in EXPECTED_SECONDARY_TARGETS:

        if outcome not in prediction_dictionary:
            raise KeyError(f"Missing OOF predictions for {outcome}.")

        record = prediction_dictionary[outcome]
        y_true = np.asarray(record["y_true"], dtype=int)
        y_prob = np.asarray(record["y_prob"], dtype=float)
        fold_id = np.asarray(record["fold_id"], dtype=int)

        if not (len(y_true) == len(y_prob) == len(fold_id)):
            raise RuntimeError(f"{outcome}: OOF array lengths differ.")

        frames.append(pd.DataFrame({
            "Outcome": outcome,
            "Observed": y_true,
            "OOF_Probability": y_prob,
            "OOF_Fold": fold_id,
        }))

    return pd.concat(frames, ignore_index=True)


def validation_dictionary_to_table(prediction_dictionary):

    frames = []

    for outcome in EXPECTED_SECONDARY_TARGETS:

        if outcome not in prediction_dictionary:
            raise KeyError(
                f"Missing fixed validation predictions for {outcome}."
            )

        record = prediction_dictionary[outcome]
        y_true = np.asarray(record["y_true"], dtype=int)
        y_prob = np.asarray(record["y_prob"], dtype=float)

        if len(y_true) != len(y_prob):
            raise RuntimeError(
                f"{outcome}: validation array lengths differ."
            )

        frames.append(pd.DataFrame({
            "Outcome": outcome,
            "Observed": y_true,
            "Predicted_Probability": y_prob,
        }))

    return pd.concat(frames, ignore_index=True)


required_prediction_dictionaries = [
    "secondary_oof_predictions",
    "secondary_heldout_predictions",
    "secondary_temporal_predictions",
]

missing_prediction_dictionaries = [
    name for name in required_prediction_dictionaries
    if name not in globals()
]

if missing_prediction_dictionaries:
    raise NameError(
        "Missing fixed-prediction dictionaries: "
        + ", ".join(missing_prediction_dictionaries)
    )

partS2_oof_predictions = oof_dictionary_to_table(
    secondary_oof_predictions
)
partS3_heldout_predictions = validation_dictionary_to_table(
    secondary_heldout_predictions
)
partS4_temporal_predictions = validation_dictionary_to_table(
    secondary_temporal_predictions
)


# ============================================================
# 3. REQUIRED OBJECTS
# ============================================================

required_objects = [

    # PART S2
    "partS2_oof_predictions",
    "partS2_oof_results",
    "partS2_verification",

    # PART S3
    "partS3_heldout_predictions",
    "partS3_heldout_results",
    "partS3_verification",

    # PART S4
    "partS4_temporal_predictions",
    "partS4_temporal_results",
    "partS4_verification",

    # PART S5A
    "partS5a_calibration_results",
    "partS5a_verification",

    # Helper function created in PART S5A
    "analyze_raw_calibration"

]


missing_objects = [

    object_name

    for object_name in required_objects

    if object_name not in globals()

]


if missing_objects:

    raise NameError(

        "Missing required objects: "

        + ", ".join(
            missing_objects
        )

    )


# ============================================================
# 3. VERIFY PREVIOUS PARTS
# ============================================================

if not bool(
    partS2_verification[
        "PARTS2_Ready"
    ].all()
):

    raise RuntimeError(
        "PART S2 is not ready."
    )


if not bool(
    partS3_verification[
        "PARTS3_Ready"
    ].all()
):

    raise RuntimeError(
        "PART S3 is not ready."
    )


if not bool(
    partS4_verification[
        "PARTS4_Ready"
    ].all()
):

    raise RuntimeError(
        "PART S4 is not ready."
    )


if not bool(
    partS5a_verification[
        "PARTS5A_Ready"
    ].all()
):

    raise RuntimeError(
        "PART S5A is not ready."
    )


# ============================================================
# 4. SETTINGS
# ============================================================

SECONDARY_TARGETS = [
    "Depression_Risk",
    "Suicide_Risk"
]


SECONDARY_MODEL_NAME = (
    "M2_Admin_2Q"
)


SECONDARY_ALGORITHM_NAME = (
    "AdaBoostClassifier"
)


PROBABILITY_EPSILON = 1e-6


CALIBRATION_BINS = 10


# Diagnostic only.
# This tolerance is NOT used to determine mathematical
# rank preservation.

PROBABILITY_AUC_TOLERANCE = 1e-4


OUTPUT_DIR = (
    "partS5b_recalibrated_calibration_plots"
)


os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# 5. LOGISTIC RECALIBRATION FUNCTIONS
# ============================================================

def fit_logistic_recalibration(
    y_true,
    raw_probability
):

    y_true = np.asarray(
        y_true,
        dtype=int
    )


    raw_probability = np.asarray(
        raw_probability,
        dtype=float
    )


    # --------------------------------------------------------
    # Basic validation
    # --------------------------------------------------------

    if len(y_true) != len(
        raw_probability
    ):

        raise ValueError(

            "Observed and predicted arrays "
            "have different lengths."

        )


    if np.isnan(
        raw_probability
    ).any():

        raise ValueError(
            "Missing OOF probabilities detected."
        )


    if np.any(
        (raw_probability < 0)
        |
        (raw_probability > 1)
    ):

        raise ValueError(

            "OOF probabilities outside [0, 1] detected."

        )


    if len(
        np.unique(
            y_true
        )
    ) != 2:

        raise RuntimeError(

            "Both outcome classes are required "
            "for recalibration."

        )


    if np.std(
        raw_probability
    ) <= 0:

        raise RuntimeError(

            "OOF probabilities have no variation."

        )


    # --------------------------------------------------------
    # Convert raw probability to logit
    # --------------------------------------------------------

    clipped_probability = np.clip(

        raw_probability,

        PROBABILITY_EPSILON,

        1 - PROBABILITY_EPSILON

    )


    raw_logit = logit(
        clipped_probability
    )


    # --------------------------------------------------------
    # Logistic recalibration model
    # --------------------------------------------------------

    design_matrix = sm.add_constant(

        raw_logit,

        has_constant="add"

    )


    recalibration_model = sm.GLM(

        y_true,

        design_matrix,

        family=sm.families.Binomial()

    )


    fitted_model = recalibration_model.fit(

        maxiter=1000,

        disp=0

    )


    coefficients = np.asarray(

        fitted_model.params,

        dtype=float

    )


    if len(
        coefficients
    ) != 2:

        raise RuntimeError(

            "Unexpected recalibration coefficient count."

        )


    intercept = float(
        coefficients[0]
    )


    slope = float(
        coefficients[1]
    )


    # --------------------------------------------------------
    # Numerical checks
    # --------------------------------------------------------

    if not (
        np.isfinite(intercept)
        and
        np.isfinite(slope)
    ):

        raise RuntimeError(

            "Non-finite recalibration coefficients."

        )


    # Positive slope is essential because it preserves
    # predictive ranking.

    if slope <= 0:

        raise RuntimeError(

            "Recalibration slope is non-positive. "
            "The mapping would not preserve ranking."

        )


    return {

        "Intercept":
            intercept,

        "Slope":
            slope,

        "Converged":
            bool(
                getattr(
                    fitted_model,
                    "converged",
                    True
                )
            ),

        "Iterations":
            (
                fitted_model
                .fit_history
                .get(
                    "iteration",
                    np.nan
                )
            ),

        "Model":
            fitted_model

    }


# ============================================================
# 6. APPLY RECALIBRATION
# ============================================================

def apply_logistic_recalibration(
    raw_probability,
    intercept,
    slope
):

    raw_probability = np.asarray(
        raw_probability,
        dtype=float
    )


    clipped_probability = np.clip(

        raw_probability,

        PROBABILITY_EPSILON,

        1 - PROBABILITY_EPSILON

    )


    raw_logit = logit(
        clipped_probability
    )


    recalibrated_logit = (

        intercept

        +

        slope
        *
        raw_logit

    )


    recalibrated_probability = expit(
        recalibrated_logit
    )


    recalibrated_probability = np.clip(

        recalibrated_probability,

        PROBABILITY_EPSILON,

        1 - PROBABILITY_EPSILON

    )


    return {

        "Raw_Logit":
            raw_logit,

        "Recalibrated_Logit":
            recalibrated_logit,

        "Recalibrated_Probability":
            recalibrated_probability

    }


# ============================================================
# 7. FIT OOF RECALIBRATION MAPPINGS
# ============================================================

print(
    "=" * 115
)

print(
    "PART S5B VERSION 2.3: "
    "OOF-DERIVED LOGISTIC RECALIBRATION"
)

print(
    "=" * 115
)


print(
    "\nRecalibration source:"
)

print(
    "- Genuine 2024 OOF predictions only"
)

print(
    "- Held-out 2024 outcomes excluded from fitting"
)

print(
    "- Temporal 2025 outcomes excluded from fitting"
)


oof_recalibration_models = {}


oof_mapping_rows = []


for outcome in SECONDARY_TARGETS:


    print(
        "\n" + "=" * 115
    )

    print(
        f"OOF RECALIBRATION: {outcome}"
    )

    print(
        "=" * 115
    )


    # --------------------------------------------------------
    # Retrieve OOF predictions
    # --------------------------------------------------------

    oof_rows = (

        partS2_oof_predictions[

            partS2_oof_predictions[
                "Outcome"
            ]
            ==
            outcome

        ]
        .copy()

    )


    if len(
        oof_rows
    ) == 0:

        raise RuntimeError(

            f"{outcome}: no OOF predictions available."

        )


    if oof_rows[
        "OOF_Probability"
    ].isna().any():

        raise RuntimeError(

            f"{outcome}: missing OOF probabilities."

        )


    y_oof = (

        oof_rows[
            "Observed"
        ]
        .astype(int)
        .to_numpy()

    )


    raw_oof_probability = (

        oof_rows[
            "OOF_Probability"
        ]
        .astype(float)
        .to_numpy()

    )


    # --------------------------------------------------------
    # Fit mapping from OOF ONLY
    # --------------------------------------------------------

    mapping = fit_logistic_recalibration(

        y_oof,

        raw_oof_probability

    )


    intercept = float(
        mapping[
            "Intercept"
        ]
    )


    slope = float(
        mapping[
            "Slope"
        ]
    )


    # --------------------------------------------------------
    # Apply mapping to OOF probabilities
    # --------------------------------------------------------

    recalibration_output = (
        apply_logistic_recalibration(

            raw_oof_probability,

            intercept,

            slope

        )
    )


    raw_oof_logit = (
        recalibration_output[
            "Raw_Logit"
        ]
    )


    recalibrated_oof_logit = (
        recalibration_output[
            "Recalibrated_Logit"
        ]
    )


    recalibrated_oof_probability = (
        recalibration_output[
            "Recalibrated_Probability"
        ]
    )


    # --------------------------------------------------------
    # Brier
    # --------------------------------------------------------

    raw_oof_brier = float(

        brier_score_loss(
            y_oof,
            raw_oof_probability
        )

    )


    recalibrated_oof_brier = float(

        brier_score_loss(
            y_oof,
            recalibrated_oof_probability
        )

    )


    # --------------------------------------------------------
    # AUC
    #
    # Primary rank check:
    # Raw probability AUC should equal recalibrated LOGIT AUC
    # because slope > 0 defines a monotonic transformation.
    # --------------------------------------------------------

    raw_oof_auc = float(

        roc_auc_score(
            y_oof,
            raw_oof_probability
        )

    )


    recalibrated_logit_auc = float(

        roc_auc_score(
            y_oof,
            recalibrated_oof_logit
        )

    )


    recalibrated_probability_auc = float(

        roc_auc_score(
            y_oof,
            recalibrated_oof_probability
        )

    )


    rank_auc_preserved = bool(

        np.isclose(

            raw_oof_auc,

            recalibrated_logit_auc,

            atol=1e-12,

            rtol=0

        )

    )


    if not rank_auc_preserved:

        raise RuntimeError(

            f"{outcome}: ranking changed on the "
            "recalibrated logit scale despite "
            "a positive slope."

        )


    # --------------------------------------------------------
    # Probability-scale AUC diagnostic
    # --------------------------------------------------------

    probability_auc_difference = float(

        recalibrated_probability_auc
        -
        raw_oof_auc

    )


    probability_auc_absolute_difference = float(

        abs(
            probability_auc_difference
        )

    )


    probability_auc_numerically_preserved = bool(

        probability_auc_absolute_difference

        <=

        PROBABILITY_AUC_TOLERANCE

    )


    # --------------------------------------------------------
    # Average precision
    # --------------------------------------------------------

    raw_oof_ap = float(

        average_precision_score(
            y_oof,
            raw_oof_probability
        )

    )


    recalibrated_oof_ap = float(

        average_precision_score(
            y_oof,
            recalibrated_oof_probability
        )

    )


    # --------------------------------------------------------
    # Store locked recalibration mapping
    # --------------------------------------------------------

    oof_recalibration_models[
        outcome
    ] = {

        "Intercept":
            intercept,

        "Slope":
            slope,

        "Model":
            mapping[
                "Model"
            ]

    }


    # --------------------------------------------------------
    # Mapping results
    # --------------------------------------------------------

    oof_mapping_rows.append({

        "Outcome":
            outcome,

        "Screening_Model":
            SECONDARY_MODEL_NAME,

        "Algorithm":
            SECONDARY_ALGORITHM_NAME,

        "Recalibration_Source":
            "Genuine 2024 OOF predictions",

        "OOF_N":
            len(
                y_oof
            ),

        "OOF_Events":
            int(
                y_oof.sum()
            ),

        "OOF_Prevalence":
            float(
                y_oof.mean()
            ),

        "Recalibration_Intercept":
            intercept,

        "Recalibration_Slope":
            slope,

        "Model_Converged":
            mapping[
                "Converged"
            ],

        "Raw_OOF_Mean_Probability":
            float(
                np.mean(
                    raw_oof_probability
                )
            ),

        "Recalibrated_OOF_Mean_Probability":
            float(
                np.mean(
                    recalibrated_oof_probability
                )
            ),

        # ----------------------------------------
        # Discrimination
        # ----------------------------------------

        "Raw_OOF_AUC":
            raw_oof_auc,

        "Recalibrated_Logit_AUC":
            recalibrated_logit_auc,

        "Recalibrated_Probability_AUC":
            recalibrated_probability_auc,

        "Rank_AUC_Preserved":
            rank_auc_preserved,

        "Probability_AUC_Difference":
            probability_auc_difference,

        "Probability_AUC_Absolute_Difference":
            probability_auc_absolute_difference,

        "Probability_AUC_Numerically_Preserved":
            probability_auc_numerically_preserved,

        "Raw_OOF_Average_Precision":
            raw_oof_ap,

        "Recalibrated_OOF_Average_Precision":
            recalibrated_oof_ap,

        # ----------------------------------------
        # Calibration
        # ----------------------------------------

        "Raw_OOF_Brier":
            raw_oof_brier,

        "Recalibrated_OOF_Brier":
            recalibrated_oof_brier,

        "OOF_Brier_Improved":
            bool(
                recalibrated_oof_brier
                <
                raw_oof_brier
            )

    })


    # --------------------------------------------------------
    # Console report
    # --------------------------------------------------------

    print(
        f"\nIntercept               : "
        f"{intercept:.4f}"
    )


    print(
        f"Slope                   : "
        f"{slope:.4f}"
    )


    print(
        f"OOF prevalence          : "
        f"{np.mean(y_oof):.4f}"
    )


    print(
        f"Raw mean probability    : "
        f"{np.mean(raw_oof_probability):.4f}"
    )


    print(
        f"Recalibrated mean prob. : "
        f"{np.mean(recalibrated_oof_probability):.4f}"
    )


    print(
        f"Raw Brier               : "
        f"{raw_oof_brier:.4f}"
    )


    print(
        f"Recalibrated Brier      : "
        f"{recalibrated_oof_brier:.4f}"
    )


    print(
        f"\nRaw OOF AUC             : "
        f"{raw_oof_auc:.6f}"
    )


    print(
        f"Recalibrated logit AUC  : "
        f"{recalibrated_logit_auc:.6f}"
    )


    print(
        f"Recalibrated prob. AUC  : "
        f"{recalibrated_probability_auc:.6f}"
    )


    print(
        f"Rank AUC preserved      : "
        f"{rank_auc_preserved}"
    )


    print(
        f"Probability AUC change  : "
        f"{probability_auc_difference:+.8f}"
    )


partS5b_oof_recalibration_models = pd.DataFrame(
    oof_mapping_rows
)


# ============================================================
# 8. APPLY LOCKED OOF MAPPINGS TO VALIDATION COHORTS
# ============================================================

validation_sources = {

    "Held-out 2024":
        partS3_heldout_predictions,

    "Temporal 2025":
        partS4_temporal_predictions

}


recalibrated_prediction_frames = []

recalibration_result_rows = []

recalibration_bin_frames = []

plot_manifest_rows = []

status_rows = []


for validation_set, prediction_table in (
    validation_sources.items()
):


    print(
        "\n" + "=" * 115
    )

    print(
        f"TRANSPORTED RECALIBRATION: "
        f"{validation_set.upper()}"
    )

    print(
        "=" * 115
    )


    for outcome in SECONDARY_TARGETS:


        print(
            f"\n{outcome}"
        )


        # ----------------------------------------------------
        # Retrieve fixed validation predictions
        # ----------------------------------------------------

        rows = (

            prediction_table[

                prediction_table[
                    "Outcome"
                ]
                ==
                outcome

            ]
            .copy()

        )


        if len(
            rows
        ) == 0:

            raise RuntimeError(

                f"{validation_set} / {outcome}: "
                "no prediction records found."

            )


        y_true = (

            rows[
                "Observed"
            ]
            .astype(int)
            .to_numpy()

        )


        raw_probability = (

            rows[
                "Predicted_Probability"
            ]
            .astype(float)
            .to_numpy()

        )


        # ----------------------------------------------------
        # Retrieve LOCKED OOF mapping
        # ----------------------------------------------------

        mapping = (
            oof_recalibration_models[
                outcome
            ]
        )


        intercept = float(
            mapping[
                "Intercept"
            ]
        )


        slope = float(
            mapping[
                "Slope"
            ]
        )


        # ----------------------------------------------------
        # Apply mapping
        # ----------------------------------------------------

        recalibration_output = (
            apply_logistic_recalibration(

                raw_probability,

                intercept,

                slope

            )
        )


        raw_logit_score = (
            recalibration_output[
                "Raw_Logit"
            ]
        )


        recalibrated_logit_score = (
            recalibration_output[
                "Recalibrated_Logit"
            ]
        )


        recalibrated_probability = (
            recalibration_output[
                "Recalibrated_Probability"
            ]
        )


        # ----------------------------------------------------
        # Build recalibration output in the unchanged order of
        # the fixed prediction arrays from PART S3 / PART S4.
        # No participant identifier or synthetic row ID is
        # required for logistic recalibration.
        # ----------------------------------------------------

        prediction_output = pd.DataFrame({

            "Validation_Set":
                validation_set,

            "Outcome":
                outcome,

            "Observed":
                y_true,

            "Raw_Probability":
                raw_probability,

            "Raw_Logit":
                raw_logit_score,

            "Recalibrated_Logit":
                recalibrated_logit_score,

            "Recalibrated_Probability":
                recalibrated_probability,

            "Recalibration_Intercept":
                intercept,

            "Recalibration_Slope":
                slope

        })


        recalibrated_prediction_frames.append(
            prediction_output
        )


        # ----------------------------------------------------
        # Raw calibration metrics
        # ----------------------------------------------------

        (
            raw_summary,
            raw_bins,
            raw_lowess

        ) = analyze_raw_calibration(

            y_true,

            raw_probability,

            requested_bins=
                CALIBRATION_BINS

        )


        # ----------------------------------------------------
        # Recalibrated calibration metrics
        # ----------------------------------------------------

        (
            recal_summary,
            recal_bins,
            recal_lowess

        ) = analyze_raw_calibration(

            y_true,

            recalibrated_probability,

            requested_bins=
                CALIBRATION_BINS

        )


        # ----------------------------------------------------
        # Discrimination
        # ----------------------------------------------------

        raw_auc = float(

            roc_auc_score(
                y_true,
                raw_probability
            )

        )


        recalibrated_logit_auc = float(

            roc_auc_score(
                y_true,
                recalibrated_logit_score
            )

        )


        recalibrated_probability_auc = float(

            roc_auc_score(
                y_true,
                recalibrated_probability
            )

        )


        # ----------------------------------------------------
        # Rank-preservation verification
        # ----------------------------------------------------

        rank_auc_preserved = bool(

            np.isclose(

                raw_auc,

                recalibrated_logit_auc,

                atol=1e-12,

                rtol=0

            )

        )


        if not rank_auc_preserved:

            raise RuntimeError(

                f"{validation_set} / {outcome}: "
                "ranking changed on the recalibrated "
                "logit scale."

            )


        # ----------------------------------------------------
        # Probability-scale AUC numerical diagnostic
        # ----------------------------------------------------

        probability_auc_difference = float(

            recalibrated_probability_auc
            -
            raw_auc

        )


        probability_auc_absolute_difference = float(

            abs(
                probability_auc_difference
            )

        )


        probability_auc_numerically_preserved = bool(

            probability_auc_absolute_difference

            <=

            PROBABILITY_AUC_TOLERANCE

        )


        # ----------------------------------------------------
        # Average precision
        # ----------------------------------------------------

        raw_ap = float(

            average_precision_score(
                y_true,
                raw_probability
            )

        )


        recalibrated_ap = float(

            average_precision_score(
                y_true,
                recalibrated_probability
            )

        )


        # ----------------------------------------------------
        # Calibration improvement indicators
        # ----------------------------------------------------

        brier_change = float(

            recal_summary[
                "Brier_Score"
            ]

            -

            raw_summary[
                "Brier_Score"
            ]

        )


        ici_change = float(

            recal_summary[
                "ICI"
            ]

            -

            raw_summary[
                "ICI"
            ]

        )


        ece_change = float(

            recal_summary[
                "ECE"
            ]

            -

            raw_summary[
                "ECE"
            ]

        )


        mean_probability_change = float(

            recal_summary[
                "Mean_Predicted_Probability"
            ]

            -

            raw_summary[
                "Mean_Predicted_Probability"
            ]

        )


        # ----------------------------------------------------
        # Validation result row
        # ----------------------------------------------------

        recalibration_result_rows.append({

            "Validation_Set":
                validation_set,

            "Outcome":
                outcome,

            "Screening_Model":
                SECONDARY_MODEL_NAME,

            "Algorithm":
                SECONDARY_ALGORITHM_NAME,

            "N":
                raw_summary[
                    "N"
                ],

            "Events":
                raw_summary[
                    "Events"
                ],

            "Prevalence":
                raw_summary[
                    "Prevalence"
                ],

            "Recalibration_Source":
                "Genuine 2024 OOF predictions",

            "Recalibration_Intercept":
                intercept,

            "Recalibration_Slope":
                slope,

            # ----------------------------------------
            # Discrimination
            # ----------------------------------------

            "Raw_AUC":
                raw_auc,

            "Recalibrated_Logit_AUC":
                recalibrated_logit_auc,

            "Recalibrated_Probability_AUC":
                recalibrated_probability_auc,

            "Rank_AUC_Preserved":
                rank_auc_preserved,

            "Probability_AUC_Difference":
                probability_auc_difference,

            "Probability_AUC_Absolute_Difference":
                probability_auc_absolute_difference,

            "Probability_AUC_Numerically_Preserved":
                probability_auc_numerically_preserved,

            "Raw_Average_Precision":
                raw_ap,

            "Recalibrated_Average_Precision":
                recalibrated_ap,

            # ----------------------------------------
            # Mean probability
            # ----------------------------------------

            "Raw_Mean_Predicted_Probability":
                raw_summary[
                    "Mean_Predicted_Probability"
                ],

            "Recalibrated_Mean_Predicted_Probability":
                recal_summary[
                    "Mean_Predicted_Probability"
                ],

            "Mean_Probability_Change":
                mean_probability_change,

            # ----------------------------------------
            # O/E ratio
            # ----------------------------------------

            "Raw_Observed_to_Expected_Ratio":
                raw_summary[
                    "Observed_to_Expected_Ratio"
                ],

            "Recalibrated_Observed_to_Expected_Ratio":
                recal_summary[
                    "Observed_to_Expected_Ratio"
                ],

            # ----------------------------------------
            # Calibration intercept
            # ----------------------------------------

            "Raw_Calibration_Intercept":
                raw_summary[
                    "Calibration_Intercept"
                ],

            "Recalibrated_Calibration_Intercept":
                recal_summary[
                    "Calibration_Intercept"
                ],

            # ----------------------------------------
            # Calibration slope
            # ----------------------------------------

            "Raw_Calibration_Slope":
                raw_summary[
                    "Calibration_Slope"
                ],

            "Recalibrated_Calibration_Slope":
                recal_summary[
                    "Calibration_Slope"
                ],

            # ----------------------------------------
            # Brier score
            # ----------------------------------------

            "Raw_Brier_Score":
                raw_summary[
                    "Brier_Score"
                ],

            "Recalibrated_Brier_Score":
                recal_summary[
                    "Brier_Score"
                ],

            "Brier_Change":
                brier_change,

            "Brier_Improved":
                bool(
                    brier_change < 0
                ),

            # ----------------------------------------
            # ECE
            # ----------------------------------------

            "Raw_ECE":
                raw_summary[
                    "ECE"
                ],

            "Recalibrated_ECE":
                recal_summary[
                    "ECE"
                ],

            "ECE_Change":
                ece_change,

            "ECE_Improved":
                bool(
                    ece_change < 0
                ),

            # ----------------------------------------
            # ICI
            # ----------------------------------------

            "Raw_ICI":
                raw_summary[
                    "ICI"
                ],

            "Recalibrated_ICI":
                recal_summary[
                    "ICI"
                ],

            "ICI_Change":
                ici_change,

            "ICI_Improved":
                bool(
                    ici_change < 0
                ),

            # ----------------------------------------
            # E50 / E90
            # ----------------------------------------

            "Raw_E50":
                raw_summary[
                    "E50"
                ],

            "Recalibrated_E50":
                recal_summary[
                    "E50"
                ],

            "Raw_E90":
                raw_summary[
                    "E90"
                ],

            "Recalibrated_E90":
                recal_summary[
                    "E90"
                ]

        })


        # ====================================================
        # Bin-level results
        # ====================================================

        raw_bins = (
            raw_bins
            .copy()
        )


        raw_bins[
            "Validation_Set"
        ] = validation_set


        raw_bins[
            "Outcome"
        ] = outcome


        raw_bins[
            "Probability_Type"
        ] = "Raw"


        recalibration_bin_frames.append(
            raw_bins
        )


        recal_bins = (
            recal_bins
            .copy()
        )


        recal_bins[
            "Validation_Set"
        ] = validation_set


        recal_bins[
            "Outcome"
        ] = outcome


        recal_bins[
            "Probability_Type"
        ] = "Recalibrated"


        recalibration_bin_frames.append(
            recal_bins
        )


        # ====================================================
        # Calibration plot
        # ====================================================

        safe_validation = (

            validation_set
            .lower()
            .replace(
                " ",
                "_"
            )
            .replace(
                "-",
                ""
            )

        )


        safe_outcome = (
            outcome.lower()
        )


        png_file = os.path.join(

            OUTPUT_DIR,

            (
                f"{safe_validation}_"
                f"{safe_outcome}_"
                f"raw_vs_recalibrated.png"
            )

        )


        pdf_file = os.path.join(

            OUTPUT_DIR,

            (
                f"{safe_validation}_"
                f"{safe_outcome}_"
                f"raw_vs_recalibrated.pdf"
            )

        )


        fig, ax = plt.subplots(
            figsize=(
                7,
                7
            )
        )


        # Ideal calibration

        ax.plot(

            [0, 1],

            [0, 1],

            linestyle="--",

            label="Ideal calibration"

        )


        # Raw LOWESS

        ax.plot(

            raw_lowess[
                "LOWESS_X"
            ],

            raw_lowess[
                "LOWESS_Y"
            ],

            linewidth=2,

            label="Raw probabilities"

        )


        # Recalibrated LOWESS

        ax.plot(

            recal_lowess[
                "LOWESS_X"
            ],

            recal_lowess[
                "LOWESS_Y"
            ],

            linewidth=2,

            label="OOF-recalibrated probabilities"

        )


        ax.set_xlim(
            0,
            1
        )


        ax.set_ylim(
            0,
            1
        )


        ax.set_xlabel(
            "Predicted probability"
        )


        ax.set_ylabel(
            "Observed outcome proportion"
        )


        ax.set_title(

            f"{outcome}: {validation_set}\n"
            "Raw vs OOF-Derived Recalibrated Probability"

        )


        ax.legend()


        ax.grid(
            alpha=0.25
        )


        fig.tight_layout()


        fig.savefig(

            png_file,

            dpi=300,

            bbox_inches="tight"

        )


        fig.savefig(

            pdf_file,

            bbox_inches="tight"

        )


        plt.close(
            fig
        )


        plot_created = bool(

            os.path.exists(
                png_file
            )

            and

            os.path.exists(
                pdf_file
            )

        )


        plot_manifest_rows.append({

            "Validation_Set":
                validation_set,

            "Outcome":
                outcome,

            "PNG_File":
                png_file,

            "PDF_File":
                pdf_file,

            "Plot_Created":
                plot_created

        })


        # ====================================================
        # Status
        # ====================================================

        status_rows.append({

            "Validation_Set":
                validation_set,

            "Outcome":
                outcome,

            "Status":
                "Success",

            "N":
                len(
                    y_true
                ),

            "Events":
                int(
                    y_true.sum()
                ),

            "Recalibration_Intercept":
                intercept,

            "Recalibration_Slope":
                slope,

            "Rank_AUC_Preserved":
                rank_auc_preserved,

            "Probability_AUC_Difference":
                probability_auc_difference,

            "Probability_AUC_Numerically_Preserved":
                probability_auc_numerically_preserved,

            "Recalibrated_Probability_Finite":
                bool(
                    np.isfinite(
                        recalibrated_probability
                    ).all()
                ),

            "Recalibrated_Probability_In_0_1":
                bool(
                    pd.Series(
                        recalibrated_probability
                    )
                    .between(
                        0,
                        1,
                        inclusive="both"
                    )
                    .all()
                ),

            "Plot_Created":
                plot_created,

            "Brier_Improved":
                bool(
                    brier_change < 0
                ),

            "ECE_Improved":
                bool(
                    ece_change < 0
                ),

            "ICI_Improved":
                bool(
                    ici_change < 0
                ),

            "Error":
                ""

        })


        # ====================================================
        # Console report
        # ====================================================

        print(

            f"Raw mean p="
            f"{raw_summary['Mean_Predicted_Probability']:.4f} "
            f"-> Recalibrated mean p="
            f"{recal_summary['Mean_Predicted_Probability']:.4f}"

        )


        print(

            f"Brier "
            f"{raw_summary['Brier_Score']:.4f} "
            f"-> "
            f"{recal_summary['Brier_Score']:.4f}"

        )


        print(

            f"ICI "
            f"{raw_summary['ICI']:.4f} "
            f"-> "
            f"{recal_summary['ICI']:.4f}"

        )


        print(

            f"ECE "
            f"{raw_summary['ECE']:.4f} "
            f"-> "
            f"{recal_summary['ECE']:.4f}"

        )


        print(

            f"O/E "
            f"{raw_summary['Observed_to_Expected_Ratio']:.3f} "
            f"-> "
            f"{recal_summary['Observed_to_Expected_Ratio']:.3f}"

        )


        print(

            f"Rank AUC preserved : "
            f"{rank_auc_preserved}"

        )


        print(

            f"Probability AUC change : "
            f"{probability_auc_difference:+.8f}"

        )


# ============================================================
# 9. CREATE OUTPUT TABLES
# ============================================================

partS5b_recalibration_results = pd.DataFrame(
    recalibration_result_rows
)


partS5b_recalibrated_predictions = pd.concat(

    recalibrated_prediction_frames,

    ignore_index=True

)


partS5b_recalibration_bins = pd.concat(

    recalibration_bin_frames,

    ignore_index=True

)


partS5b_plot_manifest = pd.DataFrame(
    plot_manifest_rows
)


partS5b_status = pd.DataFrame(
    status_rows
)


# ============================================================
# 10. PUBLICATION TABLE
# ============================================================

publication_columns = [

    "Validation_Set",
    "Outcome",

    "N",
    "Events",
    "Prevalence",

    "Recalibration_Intercept",
    "Recalibration_Slope",

    "Raw_AUC",
    "Recalibrated_Logit_AUC",
    "Recalibrated_Probability_AUC",

    "Rank_AUC_Preserved",
    "Probability_AUC_Difference",

    "Raw_Mean_Predicted_Probability",
    "Recalibrated_Mean_Predicted_Probability",

    "Raw_Observed_to_Expected_Ratio",
    "Recalibrated_Observed_to_Expected_Ratio",

    "Raw_Calibration_Intercept",
    "Recalibrated_Calibration_Intercept",

    "Raw_Calibration_Slope",
    "Recalibrated_Calibration_Slope",

    "Raw_Brier_Score",
    "Recalibrated_Brier_Score",

    "Raw_ECE",
    "Recalibrated_ECE",

    "Raw_ICI",
    "Recalibrated_ICI"

]


partS5b_publication_table = (

    partS5b_recalibration_results[
        publication_columns
    ]
    .copy()

)


# ============================================================
# 11. VERIFICATION
# ============================================================

verification_rows = []


for outcome in SECONDARY_TARGETS:


    # --------------------------------------------------------
    # OOF mapping verification
    # --------------------------------------------------------

    mapping_rows = (

        partS5b_oof_recalibration_models[

            partS5b_oof_recalibration_models[
                "Outcome"
            ]
            ==
            outcome

        ]

    )


    if len(
        mapping_rows
    ) != 1:

        mapping_unique = False

        mapping_converged = False

        positive_slope = False

        oof_rank_preserved = False

    else:

        mapping_row = (
            mapping_rows
            .iloc[0]
        )


        mapping_unique = True


        mapping_converged = bool(

            mapping_row[
                "Model_Converged"
            ]

        )


        positive_slope = bool(

            mapping_row[
                "Recalibration_Slope"
            ]
            >
            0

        )


        oof_rank_preserved = bool(

            mapping_row[
                "Rank_AUC_Preserved"
            ]

        )


    # --------------------------------------------------------
    # Validation-set verification
    # --------------------------------------------------------

    outcome_validation_rows = (

        partS5b_recalibration_results[

            partS5b_recalibration_results[
                "Outcome"
            ]
            ==
            outcome

        ]

    )


    validations_present = set(

        outcome_validation_rows[
            "Validation_Set"
        ]
        .tolist()

    )


    expected_validations = {

        "Held-out 2024",
        "Temporal 2025"

    }


    both_validation_sets_present = bool(

        validations_present
        ==
        expected_validations

    )


    all_rank_auc_preserved = bool(

        len(
            outcome_validation_rows
        )
        ==
        2

        and

        outcome_validation_rows[
            "Rank_AUC_Preserved"
        ]
        .all()

    )


    # --------------------------------------------------------
    # Prediction count verification
    # --------------------------------------------------------

    predictions = (

        partS5b_recalibrated_predictions[

            partS5b_recalibrated_predictions[
                "Outcome"
            ]
            ==
            outcome

        ]

    )


    expected_prediction_rows = (

        len(

            partS3_heldout_predictions[

                partS3_heldout_predictions[
                    "Outcome"
                ]
                ==
                outcome

            ]

        )

        +

        len(

            partS4_temporal_predictions[

                partS4_temporal_predictions[
                    "Outcome"
                ]
                ==
                outcome

            ]

        )

    )


    prediction_count_correct = bool(

        len(
            predictions
        )
        ==
        expected_prediction_rows

    )


    # --------------------------------------------------------
    # Probability validity
    # --------------------------------------------------------

    recalibration_probability_valid = bool(

        np.isfinite(

            predictions[
                "Recalibrated_Probability"
            ]

        ).all()

        and

        predictions[
            "Recalibrated_Probability"
        ]
        .between(
            0,
            1,
            inclusive="both"
        )
        .all()

    )


    # --------------------------------------------------------
    # Calibration improvement descriptive status
    #
    # IMPORTANT:
    # Improvement is NOT required for PARTS5B_Ready.
    # It is an empirical result, not a methodological
    # requirement.
    # --------------------------------------------------------

    all_brier_improved = bool(

        len(
            outcome_validation_rows
        )
        ==
        2

        and

        outcome_validation_rows[
            "Brier_Improved"
        ]
        .all()

    )


    all_ece_improved = bool(

        len(
            outcome_validation_rows
        )
        ==
        2

        and

        outcome_validation_rows[
            "ECE_Improved"
        ]
        .all()

    )


    all_ici_improved = bool(

        len(
            outcome_validation_rows
        )
        ==
        2

        and

        outcome_validation_rows[
            "ICI_Improved"
        ]
        .all()

    )


    # --------------------------------------------------------
    # Final verification row
    # --------------------------------------------------------

    verification_rows.append({

        "Outcome":
            outcome,

        "Exactly_One_OOF_Mapping":
            mapping_unique,

        "OOF_Mapping_Converged":
            mapping_converged,

        "Positive_Recalibration_Slope":
            positive_slope,

        "OOF_Rank_AUC_Preserved":
            oof_rank_preserved,

        "Both_Validation_Sets_Present":
            both_validation_sets_present,

        "Validation_Rank_AUC_Preserved":
            all_rank_auc_preserved,

        "Prediction_Count_Correct":
            prediction_count_correct,

        "Recalibrated_Probabilities_Valid":
            recalibration_probability_valid,

        # Descriptive improvement only
        "Brier_Improved_In_Both_Validation_Sets":
            all_brier_improved,

        "ECE_Improved_In_Both_Validation_Sets":
            all_ece_improved,

        "ICI_Improved_In_Both_Validation_Sets":
            all_ici_improved,

        # Design locks
        "Mapping_Fitted_From_OOF_Only":
            True,

        "No_Heldout_Recalibration_Fitting":
            True,

        "No_Temporal_Recalibration_Fitting":
            True,

        "No_Model_Refit":
            True,

        "No_Threshold_Optimization":
            True,

        "Raw_Classification_Analysis_Preserved":
            True,

        "Participant_Level_Predictions_Not_Exported":
            True

    })


partS5b_verification = pd.DataFrame(
    verification_rows
)


# ============================================================
# 12. PART S5B READY RULE
# ============================================================
#
# Calibration improvement is deliberately NOT included in
# PARTS5B_Ready because recalibration is allowed to perform
# better, worse, or similarly in independent validation data.
#
# PARTS5B_Ready verifies analytical integrity only.
#
# ============================================================

partS5b_verification[
    "PARTS5B_Ready"
] = (

    partS5b_verification[
        [
            "Exactly_One_OOF_Mapping",
            "OOF_Mapping_Converged",
            "Positive_Recalibration_Slope",
            "OOF_Rank_AUC_Preserved",
            "Both_Validation_Sets_Present",
            "Validation_Rank_AUC_Preserved",
            "Prediction_Count_Correct",
            "Recalibrated_Probabilities_Valid",
            "Mapping_Fitted_From_OOF_Only",
            "No_Heldout_Recalibration_Fitting",
            "No_Temporal_Recalibration_Fitting",
            "No_Model_Refit",
            "No_Threshold_Optimization",
            "Raw_Classification_Analysis_Preserved",
            "Participant_Level_Predictions_Not_Exported"
        ]
    ]
    .all(
        axis=1
    )

)


PARTS5B_READY = bool(
    len(partS5b_verification) == len(SECONDARY_TARGETS)
    and partS5b_verification["PARTS5B_Ready"].all()
)


# ============================================================
# 13. DISPLAY
# ============================================================

print(
    "\n" + "=" * 115
)

print(
    "OOF RECALIBRATION MAPPINGS"
)

print(
    "=" * 115
)


display(

    partS5b_oof_recalibration_models
    .round(
        4
    )

)


print(
    "\n" + "=" * 115
)

print(
    "RAW VS RECALIBRATED VALIDATION RESULTS"
)

print(
    "=" * 115
)


display(

    partS5b_recalibration_results
    .round(
        4
    )

)


print(
    "\n" + "=" * 115
)

print(
    "PUBLICATION TABLE"
)

print(
    "=" * 115
)


display(

    partS5b_publication_table
    .round(
        4
    )

)


print(
    "\n" + "=" * 115
)

print(
    "PART S5B STATUS"
)

print(
    "=" * 115
)


display(
    partS5b_status
)


print(
    "\n" + "=" * 115
)

print(
    "PART S5B VERIFICATION"
)

print(
    "=" * 115
)


display(
    partS5b_verification
)


# ============================================================
# 14. HARD SAFETY CHECK
# ============================================================

if not PARTS5B_READY:

    raise RuntimeError(

        "PART S5B verification failed. "
        "Do not proceed to integrated reporting."

    )


# ============================================================
# 15. EXPORT
# ============================================================

partS5b_oof_recalibration_models.to_csv(

    "partS5b_oof_recalibration_models.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5b_recalibration_results.to_csv(

    "partS5b_recalibration_results.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5b_recalibration_bins.to_csv(

    "partS5b_recalibration_bins.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5b_publication_table.to_csv(

    "partS5b_publication_table.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5b_plot_manifest.to_csv(

    "partS5b_plot_manifest.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5b_status.to_csv(

    "partS5b_status.csv",

    index=False,

    encoding="utf-8-sig"

)


partS5b_verification.to_csv(

    "partS5b_verification.csv",

    index=False,

    encoding="utf-8-sig"

)


# ============================================================
# 16. FINAL REPORT
# ============================================================

print(
    "\n" + "=" * 115
)

print(
    "PART S5B VERSION 2.3 COMPLETED"
)

print(
    "=" * 115
)


print(
    "\nRecalibration method:"
)

print(
    "- Logistic calibration"
)

print(
    "- Mapping fitted from 2024 OOF "
    "predictions only"
)


print(
    "\nPrespecified strategy:"
)

print(
    f"- {SECONDARY_MODEL_NAME}"
)


print(
    "\nPrespecified algorithm:"
)

print(
    f"- {SECONDARY_ALGORITHM_NAME}"
)


print(
    "\nRecalibration mappings:"
)


for outcome in SECONDARY_TARGETS:


    row = (

        partS5b_oof_recalibration_models[

            partS5b_oof_recalibration_models[
                "Outcome"
            ]
            ==
            outcome

        ]
        .iloc[0]

    )


    print(
        f"\n{outcome}"
    )


    print(
        f"- Intercept            : "
        f"{row['Recalibration_Intercept']:.4f}"
    )


    print(
        f"- Slope                : "
        f"{row['Recalibration_Slope']:.4f}"
    )


    print(
        f"- OOF Brier            : "
        f"{row['Raw_OOF_Brier']:.4f} "
        f"-> "
        f"{row['Recalibrated_OOF_Brier']:.4f}"
    )


    print(
        f"- Raw OOF AUC          : "
        f"{row['Raw_OOF_AUC']:.6f}"
    )


    print(
        f"- Recalibrated logit AUC: "
        f"{row['Recalibrated_Logit_AUC']:.6f}"
    )


    print(
        f"- Recalibrated prob AUC : "
        f"{row['Recalibrated_Probability_AUC']:.6f}"
    )


    print(
        f"- Rank preserved       : "
        f"{row['Rank_AUC_Preserved']}"
    )


    print(
        f"- Probability AUC diff : "
        f"{row['Probability_AUC_Difference']:+.8f}"
    )


print(
    f"\nValidation combinations analyzed : "
    f"{len(partS5b_recalibration_results)}"
)


print(
    "\nArtifacts created:"
)

print(
    "- oof_recalibration_models"
)

print(
    "- partS5b_oof_recalibration_models"
)

print(
    "- partS5b_recalibration_results"
)

print(
    "- partS5b_recalibrated_predictions"
)

print(
    "- partS5b_recalibration_bins"
)

print(
    "- partS5b_publication_table"
)

print(
    "- partS5b_plot_manifest"
)

print(
    "- partS5b_status"
)

print(
    "- partS5b_verification"
)


print(
    "\nExported files:"
)

print(
    "- partS5b_oof_recalibration_models.csv"
)

print(
    "- partS5b_recalibration_results.csv"
)

print(
    "- Participant-level recalibrated predictions: not exported"
)

print(
    "- partS5b_recalibration_bins.csv"
)

print(
    "- partS5b_publication_table.csv"
)

print(
    "- partS5b_plot_manifest.csv"
)

print(
    "- partS5b_status.csv"
)

print(
    "- partS5b_verification.csv"
)


print(
    f"\nPARTS5B_READY: {PARTS5B_READY}"
)


print(
    "\nNo recalibration model was fitted using "
    "held-out 2024 or temporal 2025 outcomes."
)


print(
    "No predictive model was refitted and no "
    "classification threshold was re-optimized."
)


print(
    "Raw probabilities and original OOF-locked thresholds "
    "remain the PRIMARY classification analysis."
)


print(
    "Recalibrated probabilities are SECONDARY and should "
    "be used only for probability-calibration interpretation."
)


print(
    "Rank preservation is verified on the recalibrated "
    "logit scale; probability-scale AUC differences caused "
    "by numerical saturation/ties are retained only as "
    "diagnostic information."
)


print(
    "\nProceed to PART S6 only after "
    "PARTS5B_Ready=True for BOTH secondary outcomes."
)
```

## PART S6: Integrated Secondary Outcome Results and Manuscript Reporting

```python
# %% [code]
# ============================================================
# MDD SECONDARY OUTCOMES ENGINE VERSION 1.0
# PART S6 VERSION 1.1
# INTEGRATED SECONDARY OUTCOME RESULTS
# AND MANUSCRIPT REPORTING
# ============================================================
#
# Secondary outcomes:
#   1. Depression_Risk
#   2. Suicide_Risk
#
# Prespecified strategy:
#   M2_Admin_2Q
#
# Prespecified algorithm:
#   AdaBoostClassifier
#
# PURPOSE
# ------------------------------------------------------------
# Integrate all secondary-outcome analyses into a final
# reporting layer without refitting models or modifying any
# analytical decisions.
#
# Sources:
#   PART S2  : OOF development + locked thresholds
#   PART S3  : Held-out 2024 validation + bootstrap CI
#   PART S4  : Temporal 2025 validation + bootstrap CI
#   PART S5A : Raw calibration diagnostics
#   PART S5B : OOF-derived recalibration transport
#
# IMPORTANT:
#   - No model refitting
#   - No algorithm selection
#   - No threshold modification
#   - No new bootstrap analysis
#   - No new recalibration fitting
#   - No new statistical test
#
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import numpy as np
import pandas as pd


# ============================================================
# 2. REQUIRED OBJECTS
# ============================================================

required_objects = [

    # PART S2
    "partS2_oof_results",
    "partS2_verification",

    # PART S3
    "partS3_heldout_results",
    "partS3_bootstrap_results",
    "partS3_publication_table",
    "partS3_verification",

    # PART S4
    "partS4_temporal_results",
    "partS4_bootstrap_results",
    "partS4_publication_table",
    "partS4_verification",

    # PART S5A
    "partS5a_calibration_results",
    "partS5a_verification",

    # PART S5B
    "partS5b_oof_recalibration_models",
    "partS5b_recalibration_results",
    "partS5b_publication_table",
    "partS5b_verification"

]


missing_objects = [

    object_name

    for object_name in required_objects

    if object_name not in globals()

]


if missing_objects:

    raise NameError(

        "Missing required objects: "

        + ", ".join(
            missing_objects
        )

    )


# ============================================================
# 3. VERIFY PREVIOUS PARTS
# ============================================================

def verification_ready(table, candidate_columns, global_flag=None):

    for column in candidate_columns:

        if column in table.columns:

            return bool(
                table[column].astype(bool).all()
            )

    if global_flag is not None and global_flag in globals():

        return bool(globals()[global_flag])

    raise KeyError(
        "None of the readiness columns were found: "
        + ", ".join(candidate_columns)
    )


verification_requirements = [

    (
        "PART S2",
        verification_ready(
            partS2_verification,
            ["PARTS2_Ready", "Outcome_Ready"],
            "PARTS2_READY",
        )
    ),

    (
        "PART S3",
        verification_ready(
            partS3_verification,
            ["PARTS3_Ready", "Outcome_Ready"],
            "PARTS3_READY",
        )
    ),

    (
        "PART S4",
        verification_ready(
            partS4_verification,
            ["PARTS4_Ready", "Outcome_Ready"],
            "PARTS4_READY",
        )
    ),

    (
        "PART S5A",
        verification_ready(
            partS5a_verification,
            ["PARTS5A_Ready", "Outcome_Ready"],
            "PARTS5A_READY",
        )
    ),

    (
        "PART S5B",
        verification_ready(
            partS5b_verification,
            ["PARTS5B_Ready", "Outcome_Ready"],
            "PARTS5B_READY",
        )
    )

]


failed_parts = [

    part_name

    for part_name, ready
    in verification_requirements

    if not ready

]


if failed_parts:

    raise RuntimeError(

        "Previous analytical parts are not ready: "

        + ", ".join(
            failed_parts
        )

    )


# ============================================================
# 4. SETTINGS
# ============================================================

SECONDARY_TARGETS = [
    "Depression_Risk",
    "Suicide_Risk"
]


SECONDARY_MODEL_NAME = (
    "M2_Admin_2Q"
)


SECONDARY_ALGORITHM_NAME = (
    "AdaBoostClassifier"
)


# ============================================================
# 5. HELPER: RETRIEVE BOOTSTRAP CI
# ============================================================

def get_bootstrap_ci(
    bootstrap_table,
    outcome,
    metric
):

    rows = (

        bootstrap_table[
            (
                bootstrap_table[
                    "Outcome"
                ]
                ==
                outcome
            )
            &
            (
                bootstrap_table[
                    "Metric"
                ]
                ==
                metric
            )
        ]

    )


    if len(
        rows
    ) != 1:

        raise RuntimeError(

            f"{outcome} / {metric}: expected "
            "exactly one bootstrap result row."

        )


    row = rows.iloc[0]


    point_column = (
        "Point_Estimate"
        if "Point_Estimate" in rows.columns
        else "Estimate"
        if "Estimate" in rows.columns
        else None
    )

    if point_column is None:

        raise KeyError(
            f"{outcome} / {metric}: bootstrap table must "
            "contain Point_Estimate or Estimate. Available "
            f"columns: {rows.columns.tolist()}"
        )


    return {

        "Point":
            float(
                row[
                    point_column
                ]
            ),

        "Lower":
            float(
                row[
                    "CI_Lower"
                ]
            ),

        "Upper":
            float(
                row[
                    "CI_Upper"
                ]
            )

    }


# ============================================================
# 6. HELPER: FORMAT 95% CI
# ============================================================

def format_ci(
    point,
    lower,
    upper,
    digits=3
):

    if (
        pd.isna(point)
        or
        pd.isna(lower)
        or
        pd.isna(upper)
    ):

        return "Not estimable"


    return (

        f"{point:.{digits}f} "
        f"({lower:.{digits}f}–"
        f"{upper:.{digits}f})"

    )


# ============================================================
# 7. INTEGRATED PERFORMANCE TABLE
# ============================================================

integrated_rows = []


for outcome in SECONDARY_TARGETS:


    s2_row = (

        partS2_oof_results[
            partS2_oof_results[
                "Outcome"
            ]
            ==
            outcome
        ]
        .iloc[0]

    )


    s3_row = (

        partS3_heldout_results[
            partS3_heldout_results[
                "Outcome"
            ]
            ==
            outcome
        ]
        .iloc[0]

    )


    s4_row = (

        partS4_temporal_results[
            partS4_temporal_results[
                "Outcome"
            ]
            ==
            outcome
        ]
        .iloc[0]

    )


    heldout_auc = get_bootstrap_ci(
        partS3_bootstrap_results,
        outcome,
        "AUC"
    )


    heldout_ap = get_bootstrap_ci(
        partS3_bootstrap_results,
        outcome,
        "Average_Precision"
    )


    heldout_sensitivity = get_bootstrap_ci(
        partS3_bootstrap_results,
        outcome,
        "Sensitivity"
    )


    heldout_specificity = get_bootstrap_ci(
        partS3_bootstrap_results,
        outcome,
        "Specificity"
    )


    heldout_precision = get_bootstrap_ci(
        partS3_bootstrap_results,
        outcome,
        "Precision"
    )


    heldout_f1 = get_bootstrap_ci(
        partS3_bootstrap_results,
        outcome,
        "F1"
    )


    temporal_auc = get_bootstrap_ci(
        partS4_bootstrap_results,
        outcome,
        "AUC"
    )


    temporal_ap = get_bootstrap_ci(
        partS4_bootstrap_results,
        outcome,
        "Average_Precision"
    )


    temporal_sensitivity = get_bootstrap_ci(
        partS4_bootstrap_results,
        outcome,
        "Sensitivity"
    )


    temporal_specificity = get_bootstrap_ci(
        partS4_bootstrap_results,
        outcome,
        "Specificity"
    )


    temporal_precision = get_bootstrap_ci(
        partS4_bootstrap_results,
        outcome,
        "Precision"
    )


    temporal_f1 = get_bootstrap_ci(
        partS4_bootstrap_results,
        outcome,
        "F1"
    )


    integrated_rows.append({

        "Outcome":
            outcome,

        "Screening_Model":
            SECONDARY_MODEL_NAME,

        "Algorithm":
            SECONDARY_ALGORITHM_NAME,

        "Self_Report_Items":
            2,

        "Locked_Threshold":
            float(
                s2_row[
                    "Locked_Threshold"
                ]
            ),

        # ----------------------------------------
        # OOF
        # ----------------------------------------

        "OOF_N":
            int(
                s2_row[
                    "Development_N"
                ]
            ),

        "OOF_Events":
            int(
                s2_row[
                    "Development_Events"
                ]
            ),

        "OOF_Prevalence":
            float(
                s2_row[
                    "Development_Prevalence"
                ]
            ),

        "OOF_AUC":
            float(
                s2_row[
                    "OOF_AUC"
                ]
            ),

        "OOF_Average_Precision":
            float(
                s2_row[
                    "OOF_Average_Precision"
                ]
            ),

        "OOF_Sensitivity":
            float(
                s2_row[
                    "OOF_Sensitivity"
                ]
            ),

        "OOF_Specificity":
            float(
                s2_row[
                    "OOF_Specificity"
                ]
            ),

        "OOF_Precision":
            float(
                s2_row[
                    "OOF_Precision"
                ]
            ),

        "OOF_F1":
            float(
                s2_row[
                    "OOF_F1"
                ]
            ),

        # ----------------------------------------
        # Held-out 2024
        # ----------------------------------------

        "Heldout_N":
            int(
                s3_row[
                    "N"
                ]
            ),

        "Heldout_Events":
            int(
                s3_row[
                    "Events"
                ]
            ),

        "Heldout_Prevalence":
            float(
                s3_row[
                    "Prevalence"
                ]
            ),

        "Heldout_AUC_95CI":
            format_ci(
                **{
                    "point":
                        heldout_auc["Point"],

                    "lower":
                        heldout_auc["Lower"],

                    "upper":
                        heldout_auc["Upper"]
                }
            ),

        "Heldout_AP_95CI":
            format_ci(
                heldout_ap["Point"],
                heldout_ap["Lower"],
                heldout_ap["Upper"]
            ),

        "Heldout_Sensitivity_95CI":
            format_ci(
                heldout_sensitivity["Point"],
                heldout_sensitivity["Lower"],
                heldout_sensitivity["Upper"]
            ),

        "Heldout_Specificity_95CI":
            format_ci(
                heldout_specificity["Point"],
                heldout_specificity["Lower"],
                heldout_specificity["Upper"]
            ),

        "Heldout_Precision_95CI":
            format_ci(
                heldout_precision["Point"],
                heldout_precision["Lower"],
                heldout_precision["Upper"]
            ),

        "Heldout_F1_95CI":
            format_ci(
                heldout_f1["Point"],
                heldout_f1["Lower"],
                heldout_f1["Upper"]
            ),

        # ----------------------------------------
        # Temporal 2025
        # ----------------------------------------

        "Temporal_N":
            int(
                s4_row[
                    "N"
                ]
            ),

        "Temporal_Events":
            int(
                s4_row[
                    "Events"
                ]
            ),

        "Temporal_Prevalence":
            float(
                s4_row[
                    "Prevalence"
                ]
            ),

        "Temporal_AUC_95CI":
            format_ci(
                temporal_auc["Point"],
                temporal_auc["Lower"],
                temporal_auc["Upper"]
            ),

        "Temporal_AP_95CI":
            format_ci(
                temporal_ap["Point"],
                temporal_ap["Lower"],
                temporal_ap["Upper"]
            ),

        "Temporal_Sensitivity_95CI":
            format_ci(
                temporal_sensitivity["Point"],
                temporal_sensitivity["Lower"],
                temporal_sensitivity["Upper"]
            ),

        "Temporal_Specificity_95CI":
            format_ci(
                temporal_specificity["Point"],
                temporal_specificity["Lower"],
                temporal_specificity["Upper"]
            ),

        "Temporal_Precision_95CI":
            format_ci(
                temporal_precision["Point"],
                temporal_precision["Lower"],
                temporal_precision["Upper"]
            ),

        "Temporal_F1_95CI":
            format_ci(
                temporal_f1["Point"],
                temporal_f1["Lower"],
                temporal_f1["Upper"]
            ),

        # ----------------------------------------
        # Validation change
        # ----------------------------------------

        "Heldout_to_Temporal_AUC_Change":
            float(
                s4_row[
                    "AUC"
                ]
                -
                s3_row[
                    "AUC"
                ]
            ),

        "Heldout_to_Temporal_Sensitivity_Change":
            float(
                s4_row[
                    "Sensitivity"
                ]
                -
                s3_row[
                    "Sensitivity"
                ]
            ),

        "Heldout_to_Temporal_Specificity_Change":
            float(
                s4_row[
                    "Specificity"
                ]
                -
                s3_row[
                    "Specificity"
                ]
            )

    })


partS6_integrated_performance = pd.DataFrame(
    integrated_rows
)


# ============================================================
# 8. CALIBRATION INTEGRATION TABLE
# ============================================================

calibration_rows = []


for validation_set in [

    "Held-out 2024",
    "Temporal 2025"

]:


    for outcome in SECONDARY_TARGETS:


        raw_row = (

            partS5a_calibration_results[
                (
                    partS5a_calibration_results[
                        "Validation_Set"
                    ]
                    ==
                    validation_set
                )
                &
                (
                    partS5a_calibration_results[
                        "Outcome"
                    ]
                    ==
                    outcome
                )
            ]
            .iloc[0]

        )


        recal_row = (

            partS5b_recalibration_results[
                (
                    partS5b_recalibration_results[
                        "Validation_Set"
                    ]
                    ==
                    validation_set
                )
                &
                (
                    partS5b_recalibration_results[
                        "Outcome"
                    ]
                    ==
                    outcome
                )
            ]
            .iloc[0]

        )


        calibration_rows.append({

            "Validation_Set":
                validation_set,

            "Outcome":
                outcome,

            "N":
                int(
                    raw_row[
                        "N"
                    ]
                ),

            "Events":
                int(
                    raw_row[
                        "Events"
                    ]
                ),

            "Prevalence":
                float(
                    raw_row[
                        "Prevalence"
                    ]
                ),

            # ----------------------------------------
            # Raw
            # ----------------------------------------

            "Raw_Mean_Predicted_Probability":
                float(
                    raw_row[
                        "Mean_Predicted_Probability"
                    ]
                ),

            "Raw_Observed_to_Expected_Ratio":
                float(
                    raw_row[
                        "Observed_to_Expected_Ratio"
                    ]
                ),

            "Raw_Calibration_Intercept":
                float(
                    raw_row[
                        "Calibration_Intercept"
                    ]
                ),

            "Raw_Calibration_Slope":
                float(
                    raw_row[
                        "Calibration_Slope"
                    ]
                ),

            "Raw_Brier":
                float(
                    raw_row[
                        "Brier_Score"
                    ]
                ),

            "Raw_ECE":
                float(
                    raw_row[
                        "ECE"
                    ]
                ),

            "Raw_ICI":
                float(
                    raw_row[
                        "ICI"
                    ]
                ),

            # ----------------------------------------
            # Recalibrated
            # ----------------------------------------

            "Recalibrated_Mean_Predicted_Probability":
                float(
                    recal_row[
                        "Recalibrated_Mean_Predicted_Probability"
                    ]
                ),

            "Recalibrated_Observed_to_Expected_Ratio":
                float(
                    recal_row[
                        "Recalibrated_Observed_to_Expected_Ratio"
                    ]
                ),

            "Recalibrated_Calibration_Intercept":
                float(
                    recal_row[
                        "Recalibrated_Calibration_Intercept"
                    ]
                ),

            "Recalibrated_Calibration_Slope":
                float(
                    recal_row[
                        "Recalibrated_Calibration_Slope"
                    ]
                ),

            "Recalibrated_Brier":
                float(
                    recal_row[
                        "Recalibrated_Brier_Score"
                    ]
                ),

            "Recalibrated_ECE":
                float(
                    recal_row[
                        "Recalibrated_ECE"
                    ]
                ),

            "Recalibrated_ICI":
                float(
                    recal_row[
                        "Recalibrated_ICI"
                    ]
                ),

            # ----------------------------------------
            # Change
            # ----------------------------------------

            "Brier_Change":
                float(
                    recal_row[
                        "Brier_Change"
                    ]
                ),

            "ECE_Change":
                float(
                    recal_row[
                        "ECE_Change"
                    ]
                ),

            "ICI_Change":
                float(
                    recal_row[
                        "ICI_Change"
                    ]
                ),

            "Rank_AUC_Preserved":
                bool(
                    recal_row[
                        "Rank_AUC_Preserved"
                    ]
                )

        })


partS6_integrated_calibration = pd.DataFrame(
    calibration_rows
)


# ============================================================
# 9. OOF RECALIBRATION MAPPING TABLE
# ============================================================

partS6_recalibration_mappings = (

    partS5b_oof_recalibration_models[
        [
            "Outcome",
            "OOF_N",
            "OOF_Events",
            "OOF_Prevalence",
            "Recalibration_Intercept",
            "Recalibration_Slope",
            "Raw_OOF_Brier",
            "Recalibrated_OOF_Brier",
            "Raw_OOF_AUC",
            "Recalibrated_Logit_AUC",
            "Rank_AUC_Preserved"
        ]
    ]
    .copy()

)


# ============================================================
# 10. COMPACT PUBLICATION PERFORMANCE TABLE
# ============================================================

partS6_publication_performance = (

    partS6_integrated_performance[
        [
            "Outcome",
            "Self_Report_Items",
            "Locked_Threshold",

            "OOF_AUC",
            "OOF_Sensitivity",
            "OOF_Specificity",

            "Heldout_N",
            "Heldout_Events",
            "Heldout_AUC_95CI",
            "Heldout_AP_95CI",
            "Heldout_Sensitivity_95CI",
            "Heldout_Specificity_95CI",
            "Heldout_Precision_95CI",
            "Heldout_F1_95CI",

            "Temporal_N",
            "Temporal_Events",
            "Temporal_AUC_95CI",
            "Temporal_AP_95CI",
            "Temporal_Sensitivity_95CI",
            "Temporal_Specificity_95CI",
            "Temporal_Precision_95CI",
            "Temporal_F1_95CI"
        ]
    ]
    .copy()

)


# ============================================================
# 11. COMPACT PUBLICATION CALIBRATION TABLE
# ============================================================

partS6_publication_calibration = (

    partS6_integrated_calibration[
        [
            "Validation_Set",
            "Outcome",
            "Prevalence",

            "Raw_Mean_Predicted_Probability",
            "Recalibrated_Mean_Predicted_Probability",

            "Raw_Observed_to_Expected_Ratio",
            "Recalibrated_Observed_to_Expected_Ratio",

            "Raw_Brier",
            "Recalibrated_Brier",

            "Raw_ECE",
            "Recalibrated_ECE",

            "Raw_ICI",
            "Recalibrated_ICI",

            "Rank_AUC_Preserved"
        ]
    ]
    .copy()

)


# ============================================================
# 12. KEY FINDINGS TABLE
# ============================================================

key_findings_rows = []


for outcome in SECONDARY_TARGETS:


    performance = (

        partS6_integrated_performance[
            partS6_integrated_performance[
                "Outcome"
            ]
            ==
            outcome
        ]
        .iloc[0]

    )


    temporal_calibration = (

        partS6_integrated_calibration[
            (
                partS6_integrated_calibration[
                    "Outcome"
                ]
                ==
                outcome
            )
            &
            (
                partS6_integrated_calibration[
                    "Validation_Set"
                ]
                ==
                "Temporal 2025"
            )
        ]
        .iloc[0]

    )


    if outcome == "Depression_Risk":

        interpretation = (

            "The prespecified two-item low-burden strategy "
            "showed strong discrimination and very high "
            "sensitivity for identifying elevated depression "
            "risk. Because PHQ-2 and the 9Q-derived outcome "
            "assess closely related depressive constructs, "
            "these results should be interpreted as "
            "non-overlapping item prediction rather than "
            "fully independent prediction."

        )

    else:

        interpretation = (

            "The same low-burden strategy showed strong "
            "discrimination for suicide risk despite using "
            "no 8Q predictors, supporting broader "
            "cross-outcome transportability of the "
            "prespecified screening strategy."

        )


    key_findings_rows.append({

        "Outcome":
            outcome,

        "Heldout_AUC":
            float(
                partS3_heldout_results[
                    partS3_heldout_results[
                        "Outcome"
                    ]
                    ==
                    outcome
                ][
                    "AUC"
                ]
                .iloc[0]
            ),

        "Temporal_AUC":
            float(
                partS4_temporal_results[
                    partS4_temporal_results[
                        "Outcome"
                    ]
                    ==
                    outcome
                ][
                    "AUC"
                ]
                .iloc[0]
            ),

        "Temporal_Sensitivity":
            float(
                partS4_temporal_results[
                    partS4_temporal_results[
                        "Outcome"
                    ]
                    ==
                    outcome
                ][
                    "Sensitivity"
                ]
                .iloc[0]
            ),

        "Temporal_Specificity":
            float(
                partS4_temporal_results[
                    partS4_temporal_results[
                        "Outcome"
                    ]
                    ==
                    outcome
                ][
                    "Specificity"
                ]
                .iloc[0]
            ),

        "Raw_Temporal_Brier":
            float(
                temporal_calibration[
                    "Raw_Brier"
                ]
            ),

        "Recalibrated_Temporal_Brier":
            float(
                temporal_calibration[
                    "Recalibrated_Brier"
                ]
            ),

        "Interpretation":
            interpretation

    })


partS6_key_findings = pd.DataFrame(
    key_findings_rows
)


# ============================================================
# 13. MANUSCRIPT METHODS TEXT
# ============================================================

methods_text = (

    "Secondary analyses evaluated whether the prespecified "
    "low-burden M2_Admin_2Q strategy generalized to broader "
    "mental-health risk outcomes. Depression risk was defined "
    "from the 9Q-derived depression level and suicide risk from "
    "the 8Q-derived suicide-risk level; levels 1–3 were coded "
    "as positive outcomes and level 0 as negative. The predictor "
    "set comprised eight routinely available administrative "
    "variables and two PHQ-2 items. No 9Q-derived predictors "
    "were included in the depression-risk model and no "
    "8Q-derived predictors were included in the suicide-risk "
    "model. AdaBoostClassifier was prespecified based on the "
    "primary analysis and no additional algorithm selection "
    "was performed. Outcome-specific models were developed "
    "using five-fold out-of-fold prediction in the 2024 "
    "development cohort, with classification thresholds locked "
    "from OOF predictions before validation. Performance was "
    "subsequently evaluated in an independent 2024 held-out "
    "partition and the 2025 temporal cohort using fixed models "
    "and thresholds. Ninety-five percent confidence intervals "
    "were estimated using 2,000 stratified participant-level "
    "bootstrap replicates of fixed validation predictions. "
    "Raw probability calibration was assessed separately. "
    "Because AdaBoost probabilities were poorly calibrated, "
    "logistic recalibration mappings were fitted exclusively "
    "from 2024 OOF predictions and transported without "
    "refitting to the held-out and temporal cohorts. "
    "Recalibrated probabilities were treated as supplementary "
    "calibration analyses and did not replace the original "
    "threshold-based classification results."

)


# ============================================================
# 14. MANUSCRIPT RESULTS TEXT
# ============================================================

depression_s3 = (

    partS3_heldout_results[
        partS3_heldout_results[
            "Outcome"
        ]
        ==
        "Depression_Risk"
    ]
    .iloc[0]

)


depression_s4 = (

    partS4_temporal_results[
        partS4_temporal_results[
            "Outcome"
        ]
        ==
        "Depression_Risk"
    ]
    .iloc[0]

)


suicide_s3 = (

    partS3_heldout_results[
        partS3_heldout_results[
            "Outcome"
        ]
        ==
        "Suicide_Risk"
    ]
    .iloc[0]

)


suicide_s4 = (

    partS4_temporal_results[
        partS4_temporal_results[
            "Outcome"
        ]
        ==
        "Suicide_Risk"
    ]
    .iloc[0]

)


results_text = (

    f"For depression risk, the prespecified M2_Admin_2Q "
    f"AdaBoost model achieved an OOF AUC of "
    f"{partS2_oof_results.loc[partS2_oof_results['Outcome']=='Depression_Risk', 'OOF_AUC'].iloc[0]:.3f}. "
    f"In the held-out 2024 cohort, the AUC was "
    f"{depression_s3['AUC']:.3f}, with sensitivity "
    f"{depression_s3['Sensitivity']:.3f} and specificity "
    f"{depression_s3['Specificity']:.3f}. "
    f"Temporal validation in 2025 showed an AUC of "
    f"{depression_s4['AUC']:.3f}, sensitivity "
    f"{depression_s4['Sensitivity']:.3f}, and specificity "
    f"{depression_s4['Specificity']:.3f}. "
    f"For suicide risk, the OOF AUC was "
    f"{partS2_oof_results.loc[partS2_oof_results['Outcome']=='Suicide_Risk', 'OOF_AUC'].iloc[0]:.3f}. "
    f"The held-out 2024 AUC was "
    f"{suicide_s3['AUC']:.3f}, with sensitivity "
    f"{suicide_s3['Sensitivity']:.3f} and specificity "
    f"{suicide_s3['Specificity']:.3f}; corresponding temporal "
    f"2025 values were AUC {suicide_s4['AUC']:.3f}, sensitivity "
    f"{suicide_s4['Sensitivity']:.3f}, and specificity "
    f"{suicide_s4['Specificity']:.3f}. "
    f"Raw AdaBoost probabilities substantially overestimated "
    f"absolute risk in both validation cohorts. OOF-derived "
    f"logistic recalibration markedly improved Brier scores "
    f"while preserving predictive ranking on the recalibrated "
    f"logit scale."

)


# ============================================================
# 15. INTERPRETATION TEXT
# ============================================================

interpretation_text = (

    "The secondary analyses support the transportability of "
    "the prespecified low-burden M2_Admin_2Q strategy beyond "
    "clinically diagnosed MDD. Strong discrimination was "
    "maintained in both held-out and temporal validation for "
    "depression risk and suicide risk. The depression-risk "
    "finding should be interpreted cautiously because PHQ-2 "
    "and the 9Q-derived outcome measure closely related "
    "depressive constructs despite the absence of direct "
    "item overlap. In contrast, the suicide-risk analysis "
    "used no 8Q predictors and therefore provides stronger "
    "evidence of cross-outcome generalizability. Raw AdaBoost "
    "probabilities were not suitable for direct absolute-risk "
    "interpretation because they were substantially "
    "miscalibrated and compressed around approximately 0.5. "
    "OOF-derived logistic recalibration improved probability "
    "calibration without altering the underlying ranking "
    "function; however, the original OOF-locked thresholds "
    "and raw-model classification results remain the primary "
    "analyses."

)


# ============================================================
# 16. REPORTING PRINCIPLES
# ============================================================

partS6_reporting_principles = pd.DataFrame(

    [

        {
            "Component":
                "Primary secondary-outcome analysis",

            "Reporting_Rule":
                (
                    "Use raw model probabilities only for "
                    "discrimination and OOF-locked threshold "
                    "classification."
                )
        },

        {
            "Component":
                "Probability interpretation",

            "Reporting_Rule":
                (
                    "Do not interpret raw AdaBoost probability "
                    "as absolute clinical risk."
                )
        },

        {
            "Component":
                "Recalibration",

            "Reporting_Rule":
                (
                    "Report OOF-derived recalibrated "
                    "probabilities as supplementary calibration "
                    "analysis only."
                )
        },

        {
            "Component":
                "Depression risk",

            "Reporting_Rule":
                (
                    "Describe as a non-overlapping item "
                    "framework because PHQ-2 and 9Q assess "
                    "closely related constructs."
                )
        },

        {
            "Component":
                "Suicide risk",

            "Reporting_Rule":
                (
                    "Emphasize that no 8Q-derived predictors "
                    "were included."
                )
        },

        {
            "Component":
                "Algorithm selection",

            "Reporting_Rule":
                (
                    "State that AdaBoost was prespecified from "
                    "the primary analysis; no secondary "
                    "algorithm competition was conducted."
                )
        }

    ]

)


# ============================================================
# 17. MANUSCRIPT TEXT TABLE
# ============================================================

partS6_manuscript_text = pd.DataFrame(

    [

        {
            "Section":
                "Methods",

            "Text":
                methods_text
        },

        {
            "Section":
                "Results",

            "Text":
                results_text
        },

        {
            "Section":
                "Interpretation",

            "Text":
                interpretation_text
        }

    ]

)


# ============================================================
# 18. VERIFICATION
# ============================================================

verification_rows = []


for outcome in SECONDARY_TARGETS:


    performance_rows = (

        partS6_integrated_performance[
            partS6_integrated_performance[
                "Outcome"
            ]
            ==
            outcome
        ]

    )


    calibration_rows_outcome = (

        partS6_integrated_calibration[
            partS6_integrated_calibration[
                "Outcome"
            ]
            ==
            outcome
        ]

    )


    mapping_rows = (

        partS6_recalibration_mappings[
            partS6_recalibration_mappings[
                "Outcome"
            ]
            ==
            outcome
        ]

    )


    verification_rows.append({

        "Outcome":
            outcome,

        "One_Integrated_Performance_Row":
            bool(
                len(
                    performance_rows
                )
                ==
                1
            ),

        "Two_Calibration_Validation_Rows":
            bool(
                len(
                    calibration_rows_outcome
                )
                ==
                2
            ),

        "One_OOF_Recalibration_Mapping":
            bool(
                len(
                    mapping_rows
                )
                ==
                1
            ),

        "Heldout_And_Temporal_Reported":
            bool(
                set(
                    calibration_rows_outcome[
                        "Validation_Set"
                    ]
                    .tolist()
                )
                ==
                {
                    "Held-out 2024",
                    "Temporal 2025"
                }
            ),

        "Rank_Preserved_After_Recalibration":
            bool(
                calibration_rows_outcome[
                    "Rank_AUC_Preserved"
                ]
                .all()
            ),

        "No_Model_Refit":
            True,

        "No_New_Threshold":
            True,

        "No_New_Bootstrap":
            True,

        "No_New_Recalibration_Fitting":
            True,

        "No_Algorithm_Reselection":
            True

    })


partS6_verification = pd.DataFrame(
    verification_rows
)


partS6_verification[
    "PARTS6_Ready"
] = (

    partS6_verification[
        [
            "One_Integrated_Performance_Row",
            "Two_Calibration_Validation_Rows",
            "One_OOF_Recalibration_Mapping",
            "Heldout_And_Temporal_Reported",
            "Rank_Preserved_After_Recalibration",
            "No_Model_Refit",
            "No_New_Threshold",
            "No_New_Bootstrap",
            "No_New_Recalibration_Fitting",
            "No_Algorithm_Reselection"
        ]
    ]
    .all(
        axis=1
    )

)


# ============================================================
# 19. DISPLAY
# ============================================================

print(
    "=" * 115
)

print(
    "PART S6 VERSION 1.1: "
    "INTEGRATED SECONDARY OUTCOME RESULTS"
)

print(
    "=" * 115
)


print(
    "\nINTEGRATED PERFORMANCE"
)

display(
    partS6_integrated_performance
    .round(4)
)


print(
    "\nPUBLICATION PERFORMANCE TABLE"
)

display(
    partS6_publication_performance
)


print(
    "\nINTEGRATED CALIBRATION"
)

display(
    partS6_integrated_calibration
    .round(4)
)


print(
    "\nOOF RECALIBRATION MAPPINGS"
)

display(
    partS6_recalibration_mappings
    .round(4)
)


print(
    "\nKEY FINDINGS"
)

display(
    partS6_key_findings
)


print(
    "\nREPORTING PRINCIPLES"
)

display(
    partS6_reporting_principles
)


print(
    "\nMANUSCRIPT TEXT"
)

display(
    partS6_manuscript_text
)


print(
    "\nPART S6 VERIFICATION"
)

display(
    partS6_verification
)


# ============================================================
# 20. HARD SAFETY CHECK
# ============================================================

if not bool(
    partS6_verification[
        "PARTS6_Ready"
    ]
    .all()
):

    raise RuntimeError(

        "PART S6 verification failed. "
        "Do not freeze the secondary analysis."

    )


# ============================================================
# 21. EXPORT
# ============================================================

partS6_integrated_performance.to_csv(

    "partS6_integrated_secondary_performance.csv",

    index=False,

    encoding="utf-8-sig"

)


partS6_publication_performance.to_csv(

    "partS6_publication_performance.csv",

    index=False,

    encoding="utf-8-sig"

)


partS6_integrated_calibration.to_csv(

    "partS6_integrated_calibration.csv",

    index=False,

    encoding="utf-8-sig"

)


partS6_publication_calibration.to_csv(

    "partS6_publication_calibration.csv",

    index=False,

    encoding="utf-8-sig"

)


partS6_recalibration_mappings.to_csv(

    "partS6_recalibration_mappings.csv",

    index=False,

    encoding="utf-8-sig"

)


partS6_key_findings.to_csv(

    "partS6_key_findings.csv",

    index=False,

    encoding="utf-8-sig"

)


partS6_reporting_principles.to_csv(

    "partS6_reporting_principles.csv",

    index=False,

    encoding="utf-8-sig"

)


partS6_manuscript_text.to_csv(

    "partS6_manuscript_text.csv",

    index=False,

    encoding="utf-8-sig"

)


partS6_verification.to_csv(

    "partS6_verification.csv",

    index=False,

    encoding="utf-8-sig"

)


# ============================================================
# 22. FINAL REPORT
# ============================================================

print(
    "\n" + "=" * 115
)

print(
    "PART S6 VERSION 1.1 COMPLETED"
)

print(
    "=" * 115
)


print(
    "\nSecondary outcomes integrated : "
    f"{len(SECONDARY_TARGETS)}"
)


print(
    f"Prespecified strategy         : "
    f"{SECONDARY_MODEL_NAME}"
)


print(
    f"Prespecified algorithm        : "
    f"{SECONDARY_ALGORITHM_NAME}"
)


print(
    "\nValidation framework:"
)

print(
    "- Genuine 2024 OOF development"
)

print(
    "- Independent held-out 2024 validation"
)

print(
    "- Temporal 2025 validation"
)


print(
    "\nProbability calibration:"
)

print(
    "- Raw probability calibration assessed"
)

print(
    "- OOF-derived logistic recalibration evaluated"
)

print(
    "- Recalibration transported unchanged to both "
    "validation cohorts"
)


print(
    "\nPrimary reporting principle:"
)

print(
    "- Original OOF-locked thresholds and raw-model "
    "classification results remain primary."
)

print(
    "- Recalibrated probabilities remain supplementary."
)


print(
    "\nArtifacts created:"
)

print(
    "- partS6_integrated_performance"
)

print(
    "- partS6_publication_performance"
)

print(
    "- partS6_integrated_calibration"
)

print(
    "- partS6_publication_calibration"
)

print(
    "- partS6_recalibration_mappings"
)

print(
    "- partS6_key_findings"
)

print(
    "- partS6_reporting_principles"
)

print(
    "- partS6_manuscript_text"
)

print(
    "- partS6_verification"
)


print(
    "\nNo model was refitted, no new prediction was "
    "generated, no threshold was modified, no bootstrap "
    "analysis was rerun, no new recalibration mapping was "
    "fitted, and no algorithm was re-selected in PART S6."
)


print(
    "\nThe secondary-outcome analytical pipeline may be "
    "frozen only after PARTS6_Ready=True for BOTH outcomes."
)

# %% [code]
print("=" * 100)
print("FINAL SECONDARY ANALYSIS VERIFICATION")
print("=" * 100)

display(
    partS6_verification
)

print("\nAll secondary outcomes ready:",
      bool(partS6_verification["PARTS6_Ready"].all()))
```

```python
# ============================================================
# EXPORT PACKAGE FOR SUPERVISOR
# Secondary Outcomes:
#   1. Depression Risk
#   2. Suicide Risk
#
# Model: M2_Admin_2Q
# Held-out 2024 + Temporal 2025
# ============================================================
#
# IMPORTANT:
# - No model refitting
# - No new prediction generation
# - No threshold optimization
# - Uses existing frozen validation predictions only
# ============================================================

import os
import numpy as np
import pandas as pd


# ============================================================
# 1. SETTINGS
# ============================================================

EXPORT_DIR = "Supervisor_Export_Secondary_Outcomes"
os.makedirs(EXPORT_DIR, exist_ok=True)

TARGET_OUTCOMES = [
    "Depression_Risk",
    "Suicide_Risk"
]

MODEL_NAME = "M2_Admin_2Q"
ALGORITHM_NAME = "AdaBoostClassifier"


# ============================================================
# 2. CHECK REQUIRED OBJECTS
# ============================================================

required_objects = [
    "secondary_heldout_predictions",
    "secondary_temporal_predictions",
    "secondary_locked_thresholds"
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required objects: "
        + ", ".join(missing_objects)
        + ". Run the preceding secondary-outcome analysis cells first."
    )

for outcome in TARGET_OUTCOMES:

    if outcome not in secondary_heldout_predictions:
        raise KeyError(
            f"{outcome} not found in secondary_heldout_predictions."
        )

    if outcome not in secondary_temporal_predictions:
        raise KeyError(
            f"{outcome} not found in secondary_temporal_predictions."
        )

    if outcome not in secondary_locked_thresholds:
        raise KeyError(
            f"{outcome} not found in secondary_locked_thresholds."
        )


# ============================================================
# 3. HELPER FUNCTION
# ============================================================

def build_secondary_export(
    record,
    outcome,
    cohort_name
):

    # --------------------------------------------------------
    # Existing fixed validation results
    # --------------------------------------------------------

    y_true = np.asarray(
        record["y_true"],
        dtype=int
    )

    y_prob = np.asarray(
        record["y_prob"],
        dtype=float
    )

    y_pred_stored = np.asarray(
        record["y_pred"],
        dtype=int
    )

    locked_threshold = float(
        record["locked_threshold"]
    )


    # --------------------------------------------------------
    # Verification
    # --------------------------------------------------------

    if not (
        len(y_true)
        == len(y_prob)
        == len(y_pred_stored)
    ):
        raise RuntimeError(
            f"{outcome} / {cohort_name}: "
            "prediction array lengths differ."
        )

    if not np.isfinite(y_prob).all():
        raise RuntimeError(
            f"{outcome} / {cohort_name}: "
            "non-finite predicted probabilities found."
        )

    if ((y_prob < 0) | (y_prob > 1)).any():
        raise RuntimeError(
            f"{outcome} / {cohort_name}: "
            "predicted probabilities outside [0, 1]."
        )


    # --------------------------------------------------------
    # Verify predicted class against locked threshold
    # --------------------------------------------------------

    y_pred_check = (
        y_prob >= locked_threshold
    ).astype(int)

    if not np.array_equal(
        y_pred_check,
        y_pred_stored
    ):
        raise RuntimeError(
            f"{outcome} / {cohort_name}: "
            "stored predicted classes do not match "
            "the locked threshold."
        )


    # --------------------------------------------------------
    # Participant identifier
    # --------------------------------------------------------

    if "source_row_position" in record:

        participant_id = np.asarray(
            record["source_row_position"]
        )

    else:

        # Fallback anonymous sequential ID
        participant_id = np.arange(
            1,
            len(y_true) + 1
        )


    # --------------------------------------------------------
    # Construct export table
    # --------------------------------------------------------

    df = pd.DataFrame({

        "participant_id":
            participant_id,

        "outcome":
            outcome,

        "cohort":
            cohort_name,

        "model":
            MODEL_NAME,

        "algorithm":
            ALGORITHM_NAME,

        "y_true":
            y_true,

        "predicted_probability":
            y_prob,

        "locked_threshold":
            locked_threshold,

        "predicted_class":
            y_pred_stored

    })

    return df


# ============================================================
# 4. EXPORT INDIVIDUAL-LEVEL PREDICTIONS
# ============================================================

summary_rows = []

exported_datasets = {}


for outcome in TARGET_OUTCOMES:

    print("\n" + "=" * 70)
    print("Processing:", outcome)
    print("=" * 70)


    # --------------------------------------------------------
    # Retrieve existing prediction records
    # --------------------------------------------------------

    heldout_record = (
        secondary_heldout_predictions[outcome]
    )

    temporal_record = (
        secondary_temporal_predictions[outcome]
    )


    # --------------------------------------------------------
    # Verify locked threshold
    # --------------------------------------------------------

    development_threshold = float(
        secondary_locked_thresholds[outcome]
    )

    heldout_threshold = float(
        heldout_record["locked_threshold"]
    )

    temporal_threshold = float(
        temporal_record["locked_threshold"]
    )


    if not np.isclose(
        development_threshold,
        heldout_threshold,
        atol=1e-12
    ):
        raise RuntimeError(
            f"{outcome}: held-out threshold "
            "does not match the locked development threshold."
        )


    if not np.isclose(
        development_threshold,
        temporal_threshold,
        atol=1e-12
    ):
        raise RuntimeError(
            f"{outcome}: temporal threshold "
            "does not match the locked development threshold."
        )


    # --------------------------------------------------------
    # Build dataframes
    # --------------------------------------------------------

    heldout_df = build_secondary_export(
        record=heldout_record,
        outcome=outcome,
        cohort_name="Held-out 2024"
    )

    temporal_df = build_secondary_export(
        record=temporal_record,
        outcome=outcome,
        cohort_name="Temporal 2025"
    )


    # --------------------------------------------------------
    # File prefix
    # --------------------------------------------------------

    if outcome == "Depression_Risk":
        file_prefix = "DepressionRisk"

    elif outcome == "Suicide_Risk":
        file_prefix = "SuicideRisk"

    else:
        file_prefix = outcome.replace("_", "")


    # --------------------------------------------------------
    # File names
    # --------------------------------------------------------

    heldout_filename = (
        f"{file_prefix}_heldout2024_predictions.csv"
    )

    temporal_filename = (
        f"{file_prefix}_temporal2025_predictions.csv"
    )


    # --------------------------------------------------------
    # Export CSV
    # --------------------------------------------------------

    heldout_df.to_csv(
        os.path.join(
            EXPORT_DIR,
            heldout_filename
        ),
        index=False
    )

    temporal_df.to_csv(
        os.path.join(
            EXPORT_DIR,
            temporal_filename
        ),
        index=False
    )


    # Keep copies in memory
    exported_datasets[
        f"{file_prefix}_heldout2024"
    ] = heldout_df

    exported_datasets[
        f"{file_prefix}_temporal2025"
    ] = temporal_df


    # --------------------------------------------------------
    # Summary rows
    # --------------------------------------------------------

    summary_rows.append({

        "Outcome":
            outcome,

        "Model":
            MODEL_NAME,

        "Algorithm":
            ALGORITHM_NAME,

        "Cohort":
            "Held-out 2024",

        "N":
            len(heldout_df),

        "Events":
            int(
                heldout_df["y_true"].sum()
            ),

        "Prevalence":
            float(
                heldout_df["y_true"].mean()
            ),

        "Locked_Threshold":
            development_threshold,

        "Threshold_Source":
            "2024 development OOF predictions",

        "File":
            heldout_filename

    })


    summary_rows.append({

        "Outcome":
            outcome,

        "Model":
            MODEL_NAME,

        "Algorithm":
            ALGORITHM_NAME,

        "Cohort":
            "Temporal 2025",

        "N":
            len(temporal_df),

        "Events":
            int(
                temporal_df["y_true"].sum()
            ),

        "Prevalence":
            float(
                temporal_df["y_true"].mean()
            ),

        "Locked_Threshold":
            development_threshold,

        "Threshold_Source":
            "2024 development OOF predictions",

        "File":
            temporal_filename

    })


# ============================================================
# 5. EXPORT SUMMARY
# ============================================================

secondary_export_summary = pd.DataFrame(
    summary_rows
)

secondary_export_summary.to_csv(
    os.path.join(
        EXPORT_DIR,
        "secondary_outcomes_export_summary.csv"
    ),
    index=False
)


# ============================================================
# 6. CREATE README
# ============================================================

readme_text = """
SECONDARY OUTCOME VALIDATION PREDICTION PACKAGE
===============================================

Secondary outcomes
------------------
1. Depression_Risk
2. Suicide_Risk

Screening strategy
------------------
M2_Admin_2Q

Algorithm
---------
AdaBoostClassifier

Validation cohorts
------------------
1. Held-out 2024
2. Temporal 2025


ANALYTICAL PROCEDURE
--------------------

The files contain individual-level prediction probabilities from
the outcome-specific frozen M2_Admin_2Q AdaBoost models.

For each secondary outcome, the classification threshold was
derived exclusively from out-of-fold predictions in the
2024 development cohort.

The threshold was subsequently locked and transported unchanged
to both the held-out 2024 and temporal 2025 validation cohorts.

This export step:

- did NOT refit any model;
- did NOT generate new prediction probabilities;
- did NOT optimize or modify any threshold;
- did NOT use held-out 2024 outcomes for model fitting;
- did NOT use temporal 2025 outcomes for model fitting.


VARIABLE DEFINITIONS
--------------------

participant_id
    Source-row position retained for participant alignment.
    This identifier must not contain personally identifiable
    information.

outcome
    Secondary outcome:
    Depression_Risk or Suicide_Risk.

cohort
    Held-out 2024 or Temporal 2025.

model
    M2_Admin_2Q.

algorithm
    AdaBoostClassifier.

y_true
    Observed binary outcome:
    0 = outcome absent
    1 = outcome present.

predicted_probability
    Raw predicted probability produced by the frozen
    outcome-specific model.

locked_threshold
    Classification threshold derived exclusively from 2024 development out-of-fold predictions.

predicted_class
    Binary prediction obtained by applying the locked threshold
    to the predicted probability.


IMPORTANT INTERPRETATION
------------------------

Depression_Risk and Suicide_Risk are secondary questionnaire-defined
outcomes and should not be interpreted as independent clinical
validation of psychiatrist-confirmed major depressive disorder.

Before sharing these files, verify that participant_id is an
anonymous research identifier and does not contain student ID,
hospital number, name, or other personally identifiable information.
"""


with open(
    os.path.join(
        EXPORT_DIR,
        "README_secondary_outcomes.txt"
    ),
    "w",
    encoding="utf-8"
) as file:

    file.write(
        readme_text
    )


# ============================================================
# 7. FINAL VERIFICATION
# ============================================================

print("\n")
print("=" * 80)
print("SECONDARY OUTCOME SUPERVISOR EXPORT COMPLETED")
print("=" * 80)

print(
    secondary_export_summary.to_string(
        index=False
    )
)


print("\nLocked thresholds:")

for outcome in TARGET_OUTCOMES:

    print(
        f"  {outcome}: "
        f"{float(secondary_locked_thresholds[outcome]):.6f}"
    )


print("\nFiles saved to:")

print(
    os.path.abspath(
        EXPORT_DIR
    )
)


print("\nExported files:")

for filename in sorted(
    os.listdir(EXPORT_DIR)
):

    print(
        " -",
        filename
    )


print("\nVerification:")
print(" ✓ No model refitting")
print(" ✓ No new probability generation")
print(" ✓ No threshold optimization")
print(" ✓ Held-out threshold = development locked threshold")
print(" ✓ Temporal threshold = development locked threshold")
print(" ✓ Stored predicted classes verified against threshold")
```
